```python
import stim
import pymatching
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

SEED = 20261113
rng = np.random.default_rng(SEED)

print("Stim:", stim.__version__)
print("PyMatching:", pymatching.__version__)
print("NumPy:", np.__version__)
```

```python
def build_surface_code(
    distance,
    physical_error,
    rounds=None
):
    if rounds is None:
        rounds = distance

    circuit = stim.Circuit.generated(
        "surface_code:rotated_memory_x",
        distance=distance,
        rounds=rounds,
        after_clifford_depolarization=physical_error,
        before_round_data_depolarization=physical_error,
        after_reset_flip_probability=physical_error,
        before_measure_flip_probability=physical_error
    )

    return circuit
```

```python
def run_baseline(
    distance,
    physical_error,
    shots=100_000
):
    circuit = build_surface_code(
        distance=distance,
        physical_error=physical_error
    )

    detector_error_model = circuit.detector_error_model(
        decompose_errors=True
    )

    decoder = pymatching.Matching.from_detector_error_model(
        detector_error_model
    )

    sampler = circuit.compile_detector_sampler()

    syndrome, actual_observables = sampler.sample(
        shots=shots,
        separate_observables=True
    )

    predicted_observables = decoder.decode_batch(syndrome)

    failures = np.any(
        predicted_observables != actual_observables,
        axis=1
    )

    failure_count = int(np.sum(failures))
    logical_error_rate = failure_count / shots

    return {
        "distance": distance,
        "physical_error": physical_error,
        "shots": shots,
        "failures": failure_count,
        "logical_error_rate": logical_error_rate
    }
```

```python
distances = [3, 5, 7]

physical_error_rates = [
    0.001,
    0.002,
    0.005,
    0.0075,
    0.01
]

results = []

for d in distances:
    for p in physical_error_rates:
        result = run_baseline(
            distance=d,
            physical_error=p,
            shots=100_000
        )

        results.append(result)

baseline_df = pd.DataFrame(results)
baseline_df
```

```python
from scipy.stats import norm
import numpy as np

def wilson_interval(failures, shots, confidence=0.95):
    z = norm.ppf(1 - (1 - confidence) / 2)
    p = failures / shots

    denominator = 1 + z**2 / shots

    center = (
        p + z**2 / (2 * shots)
    ) / denominator

    margin = (
        z * np.sqrt(
            p * (1 - p) / shots
            + z**2 / (4 * shots**2)
        ) / denominator
    )

    return center - margin, center + margin


baseline_df["ci_lower"] = baseline_df.apply(
    lambda r: wilson_interval(
        r["failures"], r["shots"]
    )[0],
    axis=1
)

baseline_df["ci_upper"] = baseline_df.apply(
    lambda r: wilson_interval(
        r["failures"], r["shots"]
    )[1],
    axis=1
)

baseline_df
```

```python

import numpy as np
from math import comb

def finite_data_resampling(counts, alpha):
    """
    Estimate the normalized power distribution:
        Q_alpha(s) proportional to P(s)^alpha

    using the combinatorial finite-data estimator.
    """
    counts = np.asarray(counts, dtype=np.int64)
    N = int(np.sum(counts))

    if alpha < 1 or alpha != int(alpha):
        raise ValueError("alpha must be a positive integer")

    if N < alpha:
        raise ValueError("Need at least alpha samples")

    weights = np.array([
        comb(int(c), alpha) if c >= alpha else 0
        for c in counts
    ], dtype=float)

    if weights.sum() == 0:
        raise ValueError(
            "No syndrome appeared at least alpha times"
        )

    return weights / weights.sum()


# Known synthetic distribution
true_probabilities = np.array([
    0.50, 0.25, 0.15, 0.10
])

rng = np.random.default_rng(42)

N = 100_000
samples = rng.choice(
    len(true_probabilities),
    size=N,
    p=true_probabilities
)

counts = np.bincount(
    samples,
    minlength=len(true_probabilities)
)

print("Observed counts:", counts)

for alpha in [1, 2, 3]:
    estimated = finite_data_resampling(
        counts,
        alpha
    )

    exact = (
        true_probabilities ** alpha
        / np.sum(true_probabilities ** alpha)
    )

    print(f"\nAlpha = {alpha}")
    print("Estimated:", estimated)
    print("Exact:    ", exact)
    print("Absolute error:",
          np.abs(estimated - exact))
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

sample_sizes = [100, 1000, 10000, 100000]
alphas = [1, 2, 3]
repetitions = 100

rng = np.random.default_rng(123)
results = []

for N in sample_sizes:
    for alpha in alphas:
        exact = true_probabilities ** alpha
        exact /= exact.sum()

        errors = []

        for _ in range(repetitions):
            counts = rng.multinomial(
                N, true_probabilities
            )

            try:
                estimated = finite_data_resampling(
                    counts, alpha
                )

                tv = 0.5 * np.sum(
                    np.abs(estimated - exact)
                )

                errors.append(tv)

            except ValueError:
                # Undefined if no syndrome repeats alpha times
                pass

        results.append({
            "N": N,
            "alpha": alpha,
            "mean_TV": np.mean(errors),
            "std_TV": np.std(errors, ddof=1),
            "valid_runs": len(errors)
        })

df = pd.DataFrame(results)
print(df.to_string(index=False))

for alpha in alphas:
    subset = df[df["alpha"] == alpha]

    plt.plot(
        subset["N"],
        subset["mean_TV"],
        marker="o",
        label=f"alpha={alpha}"
    )

plt.xscale("log")
plt.xlabel("Calibration sample size (N)")
plt.ylabel("Mean total variation distance")
plt.title("Finite-data resampling convergence")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

df.to_csv(
    "synthetic_resampling_convergence.csv",
    index=False
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(2026)

syndrome_spaces = [4, 16, 64, 256, 1024]
sample_sizes = [1000, 10000, 100000]
alphas = [2, 3]
repetitions = 100

results = []

for K in syndrome_spaces:
    # Uniform distribution isolates support-size effects
    true_p = np.ones(K) / K

    for N in sample_sizes:
        for alpha in alphas:
            exact = true_p ** alpha
            exact /= exact.sum()

            errors = []
            undefined = 0
            unique_counts = []

            for _ in range(repetitions):
                counts = rng.multinomial(N, true_p)

                unique_counts.append(
                    np.count_nonzero(counts)
                )

                try:
                    estimated = finite_data_resampling(
                        counts, alpha
                    )

                    tv = 0.5 * np.sum(
                        np.abs(estimated - exact)
                    )

                    errors.append(tv)

                except ValueError:
                    undefined += 1

            results.append({
                "K": K,
                "N": N,
                "alpha": alpha,
                "mean_TV": (
                    np.mean(errors) if errors else np.nan
                ),
                "std_TV": (
                    np.std(errors, ddof=1)
                    if len(errors) > 1 else np.nan
                ),
                "mean_unique": np.mean(unique_counts),
                "undefined_runs": undefined
            })

df_sparse = pd.DataFrame(results)

print(df_sparse.to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

for ax, alpha in zip(axes, alphas):
    for N in sample_sizes:
        subset = df_sparse[
            (df_sparse["alpha"] == alpha) &
            (df_sparse["N"] == N)
        ]

        ax.plot(
            subset["K"],
            subset["mean_TV"],
            marker="o",
            label=f"N={N:,}"
        )

    ax.set_xscale("log", base=2)
    ax.set_xlabel("Number of possible syndromes (K)")
    ax.set_ylabel("Mean TV distance")
    ax.set_title(f"Alpha = {alpha}")
    ax.legend()
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()

df_sparse.to_csv(
    "syndrome_sparsity_experiment.csv",
    index=False
)
```

```python

import numpy as np
import pandas as pd
from collections import Counter

def analyze_detector_sparsity(
    distance,
    physical_error=0.005,
    shots=100_000,
    batch_size=10_000
):
    circuit = build_surface_code(
        distance=distance,
        physical_error=physical_error
    )

    sampler = circuit.compile_detector_sampler()
    counts = Counter()

    remaining = shots

    while remaining > 0:
        n = min(batch_size, remaining)

        detectors = sampler.sample(
            shots=n,
            bit_packed=True
        )

        counts.update(
            map(bytes, detectors)
        )

        remaining -= n

    frequencies = np.fromiter(
        counts.values(),
        dtype=np.int64
    )

    return {
        "distance": distance,
        "shots": shots,
        "num_detectors": circuit.num_detectors,
        "unique_records": len(counts),
        "unique_fraction": len(counts) / shots,
        "singleton_records": int(
            np.sum(frequencies == 1)
        ),
        "singleton_shot_fraction": float(
            np.sum(frequencies[frequencies == 1])
            / shots
        ),
        "pair_eligible_shot_fraction": float(
            np.sum(frequencies[frequencies >= 2])
            / shots
        ),
        "triple_eligible_shot_fraction": float(
            np.sum(frequencies[frequencies >= 3])
            / shots
        ),
        "max_frequency": int(frequencies.max()),
        "distinct_pair_eligible": int(
            np.sum(frequencies >= 2)
        ),
        "distinct_triple_eligible": int(
            np.sum(frequencies >= 3)
        )
    }


results = []

for distance in [3, 5, 7]:
    result = analyze_detector_sparsity(
        distance=distance,
        physical_error=0.005,
        shots=100_000
    )

    results.append(result)
    print(f"Finished distance {distance}")

df_real = pd.DataFrame(results)

print(df_real.to_string(index=False))

df_real.to_csv(
    "circuit_detector_sparsity.csv",
    index=False
)
```

```python

import numpy as np
import pandas as pd
from collections import Counter

def collision_statistics(
    distance,
    shots=100_000,
    physical_error=0.005,
    batch_size=10_000
):
    circuit = build_surface_code(
        distance=distance,
        physical_error=physical_error
    )

    sampler = circuit.compile_detector_sampler()
    counts = Counter()

    remaining = shots

    while remaining:
        n = min(remaining, batch_size)

        detectors = sampler.sample(
            shots=n,
            bit_packed=True
        )

        counts.update(map(bytes, detectors))
        remaining -= n

    frequencies = np.array(
        list(counts.values()),
        dtype=np.int64
    )

    # Integer arithmetic avoids floating-point overflow
    pair_count = sum(
        int(c) * (int(c) - 1) // 2
        for c in frequencies if c >= 2
    )

    triple_count = sum(
        int(c) * (int(c) - 1) * (int(c) - 2) // 6
        for c in frequencies if c >= 3
    )

    N = shots

    C2 = pair_count / (N * (N - 1) / 2)

    C3 = triple_count / (
        N * (N - 1) * (N - 2) / 6
    )

    return {
        "distance": distance,
        "shots": N,
        "unique_records": len(counts),
        "pair_collisions": pair_count,
        "triple_collisions": triple_count,
        "estimated_C2": C2,
        "estimated_C3": C3,
        "pair_collision_exists": pair_count > 0,
        "triple_collision_exists": triple_count > 0
    }


collision_results = []

for d in [3, 5, 7]:
    result = collision_statistics(
        distance=d,
        shots=100_000
    )

    collision_results.append(result)
    print(f"Finished distance {d}")

df_collisions = pd.DataFrame(collision_results)

print(df_collisions.to_string(index=False))

df_collisions.to_csv(
    "collision_statistics.csv",
    index=False
)
```

```python

import numpy as np
from collections import Counter

def collision_onset(
    distance=7,
    physical_error=0.005,
    max_shots=1_000_000,
    batch_size=10_000
):
    circuit = build_surface_code(
        distance=distance,
        physical_error=physical_error
    )

    sampler = circuit.compile_detector_sampler()
    counts = Counter()

    first_pair = None
    first_triple = None
    total = 0

    while total < max_shots:
        n = min(batch_size, max_shots - total)

        detectors = sampler.sample(
            shots=n,
            bit_packed=True
        )

        for record in map(bytes, detectors):
            total += 1
            counts[record] += 1

            count = counts[record]

            if count == 2 and first_pair is None:
                first_pair = total

            if count == 3 and first_triple is None:
                first_triple = total

        if first_pair is not None and first_triple is not None:
            break

    return {
        "distance": distance,
        "shots_processed": total,
        "first_pair_at": first_pair,
        "first_triple_at": first_triple,
        "unique_records": len(counts),
        "pair_found": first_pair is not None,
        "triple_found": first_triple is not None
    }


result = collision_onset(
    distance=7,
    physical_error=0.005,
    max_shots=1_000_000
)

print(result)
```

```python

from collections import Counter
import numpy as np
import pandas as pd

def fixed_budget_collision_scaling(
    distance=7,
    physical_error=0.005,
    checkpoints=(100_000, 250_000, 500_000, 1_000_000,10_000_000),
    batch_size=10_000
):
    circuit = build_surface_code(
        distance=distance,
        physical_error=physical_error
    )

    sampler = circuit.compile_detector_sampler()
    counts = Counter()
    results = []

    total = 0

    for target in checkpoints:
        while total < target:
            n = min(batch_size, target - total)

            records = sampler.sample(
                shots=n,
                bit_packed=True
            )

            counts.update(map(bytes, records))
            total += n

        frequencies = np.fromiter(
            counts.values(),
            dtype=np.int64
        )

        pair_weights = (
            frequencies * (frequencies - 1) // 2
        )

        triple_weights = (
            frequencies
            * (frequencies - 1)
            * (frequencies - 2) // 6
        )

        pair_total = int(pair_weights.sum())
        triple_total = int(triple_weights.sum())

        results.append({
            "shots": total,
            "unique_records": len(counts),
            "pair_collisions": pair_total,
            "triple_collisions": triple_total,
            "pair_supported_records": int(
                np.count_nonzero(pair_weights)
            ),
            "triple_supported_records": int(
                np.count_nonzero(triple_weights)
            ),
            "alpha2_max_weight": (
                float(pair_weights.max() / pair_total)
                if pair_total else np.nan
            ),
            "alpha3_max_weight": (
                float(triple_weights.max() / triple_total)
                if triple_total else np.nan
            )
        })

        print(f"Completed {total:,} shots")

    return pd.DataFrame(results)


df_scaling = fixed_budget_collision_scaling()

print(df_scaling.to_string(index=False))

df_scaling.to_csv(
    "distance7_collision_scaling.csv",
    index=False
)
```

```python

from collections import Counter
import numpy as np
import pandas as pd

def inspect_dominant_records(
    distance=7,
    physical_error=0.005,
    shots=1_000_000,
    batch_size=10_000,
    top_k=10
):
    circuit = build_surface_code(
        distance=distance,
        physical_error=physical_error
    )

    sampler = circuit.compile_detector_sampler()
    counts = Counter()

    remaining = shots

    while remaining > 0:
        n = min(batch_size, remaining)

        records = sampler.sample(
            shots=n,
            bit_packed=True
        )

        counts.update(map(bytes, records))
        remaining -= n

    rows = []

    for record, count in counts.most_common(top_k):
        bits = np.unpackbits(
            np.frombuffer(record, dtype=np.uint8),
            bitorder="little"
        )[:circuit.num_detectors]

        rows.append({
            "count": count,
            "frequency": count / shots,
            "detector_weight": int(bits.sum()),
            "is_all_zero": bool(bits.sum() == 0),
            "record_hex": record.hex()
        })

    return pd.DataFrame(rows)


df_top = inspect_dominant_records()

print(df_top.to_string(index=False))
```

```python

import numpy as np
import pandas as pd
import pymatching

def conditional_failure_analysis(
    distance=7,
    physical_error=0.005,
    shots=10_000_000,
    batch_size=10_000
):
    circuit = build_surface_code(
        distance=distance,
        physical_error=physical_error
    )

    dem = circuit.detector_error_model(
        decompose_errors=True
    )

    decoder = pymatching.Matching.from_detector_error_model(
        dem
    )

    sampler = circuit.compile_detector_sampler()

    total_failures = 0
    zero_records = 0
    zero_failures = 0
    nonzero_records = 0
    nonzero_failures = 0

    remaining = shots

    while remaining > 0:
        n = min(batch_size, remaining)

        detectors, actual = sampler.sample(
            shots=n,
            separate_observables=True
        )

        predicted = decoder.decode_batch(detectors)

        failures = np.any(
            predicted != actual,
            axis=1
        )

        is_zero = ~np.any(detectors, axis=1)

        zero_records += int(np.sum(is_zero))
        zero_failures += int(
            np.sum(failures & is_zero)
        )

        nonzero_records += int(np.sum(~is_zero))
        nonzero_failures += int(
            np.sum(failures & ~is_zero)
        )

        total_failures += int(np.sum(failures))
        remaining -= n

    return pd.DataFrame([{
        "distance": distance,
        "shots": shots,
        "total_failures": total_failures,
        "baseline_LER": total_failures / shots,
        "zero_records": zero_records,
        "zero_failures": zero_failures,
        "zero_conditional_LER": (
            zero_failures / zero_records
            if zero_records else np.nan
        ),
        "nonzero_records": nonzero_records,
        "nonzero_failures": nonzero_failures,
        "nonzero_conditional_LER": (
            nonzero_failures / nonzero_records
            if nonzero_records else np.nan
        )
    }])


df_conditional = conditional_failure_analysis()

print(df_conditional.to_string(index=False))

df_conditional.to_csv(
    "conditional_logical_failure.csv",
    index=False
)
```

```python

import numpy as np
import pandas as pd
import pymatching
from collections import Counter

def heldout_resampling_experiment(
    distance,
    physical_error=0.005,
    calibration_shots=10_000_000,
    evaluation_shots=10_000_000,
    batch_size=10_000
):
    circuit = build_surface_code(
        distance=distance,
        physical_error=physical_error
    )

    dem = circuit.detector_error_model(
        decompose_errors=True
    )

    decoder = pymatching.Matching.from_detector_error_model(dem)
    sampler = circuit.compile_detector_sampler()

    calibration_counts = Counter()

    # Independent calibration data
    remaining = calibration_shots

    while remaining:
        n = min(batch_size, remaining)

        records = sampler.sample(
            shots=n,
            bit_packed=True
        )

        calibration_counts.update(map(bytes, records))
        remaining -= n

    # Evaluation data and logical failure labels
    evaluation_counts = Counter()
    evaluation_failures = Counter()

    remaining = evaluation_shots
    total_failures = 0

    while remaining:
        n = min(batch_size, remaining)

        detectors, actual = sampler.sample(
            shots=n,
            separate_observables=True
        )

        predicted = decoder.decode_batch(detectors)
        failures = np.any(predicted != actual, axis=1)

        packed = np.packbits(
            detectors,
            axis=1,
            bitorder="little"
        )

        for record, failure in zip(
            map(bytes, packed), failures
        ):
            evaluation_counts[record] += 1
            evaluation_failures[record] += int(failure)

        total_failures += int(failures.sum())
        remaining -= n

    results = []

    for alpha in [1, 2, 3]:
        numerator = 0.0
        denominator = 0.0
        matched_evaluation_shots = 0

        for record, count in calibration_counts.items():
            if count < alpha:
                continue

            if alpha == 1:
                weight = float(count)
            elif alpha == 2:
                weight = float(count * (count - 1) // 2)
            else:
                weight = float(
                    count * (count - 1) * (count - 2) // 6
                )

            eval_count = evaluation_counts[record]

            if eval_count == 0:
                continue

            failure_rate = (
                evaluation_failures[record] / eval_count
            )

            numerator += weight * failure_rate
            denominator += weight
            matched_evaluation_shots += eval_count

        results.append({
            "distance": distance,
            "alpha": alpha,
            "baseline_LER": total_failures / evaluation_shots,
            "heldout_resampled_LER": (
                numerator / denominator
                if denominator else np.nan
            ),
            "matched_evaluation_shots": matched_evaluation_shots,
            "calibration_shots": calibration_shots,
            "evaluation_shots": evaluation_shots
        })

    return pd.DataFrame(results)


results = pd.concat([
    heldout_resampling_experiment(distance=3),
    heldout_resampling_experiment(distance=5)
], ignore_index=True)

print(results.to_string(index=False))

results.to_csv(
    "heldout_resampling_initial.csv",
    index=False
)
```

```python

import numpy as np
import pandas as pd
import pymatching
from collections import Counter

def resampling_coverage_diagnostic(
    distance,
    physical_error=0.005,
    calibration_shots=100_000,
    evaluation_shots=100_000,
    batch_size=10_000
):
    circuit = build_surface_code(distance, physical_error)
    dem = circuit.detector_error_model(decompose_errors=True)

    decoder = pymatching.Matching.from_detector_error_model(dem)
    sampler = circuit.compile_detector_sampler()

    calibration = Counter()

    for start in range(0, calibration_shots, batch_size):
        n = min(batch_size, calibration_shots - start)
        records = sampler.sample(shots=n, bit_packed=True)
        calibration.update(map(bytes, records))

    eval_counts = Counter()
    eval_failures = Counter()
    total_failures = 0

    for start in range(0, evaluation_shots, batch_size):
        n = min(batch_size, evaluation_shots - start)

        detectors, actual = sampler.sample(
            shots=n,
            separate_observables=True
        )

        predicted = decoder.decode_batch(detectors)
        failures = np.any(predicted != actual, axis=1)

        packed = np.packbits(
            detectors, axis=1, bitorder="little"
        )

        for record, failed in zip(map(bytes, packed), failures):
            eval_counts[record] += 1
            eval_failures[record] += int(failed)

        total_failures += int(failures.sum())

    rows = []

    for alpha in [1, 2, 3]:
        total_weight = 0
        covered_weight = 0
        weighted_failure = 0.0
        supported_records = 0
        covered_records = 0

        for record, c in calibration.items():
            if c < alpha:
                continue

            if alpha == 1:
                w = c
            elif alpha == 2:
                w = c * (c - 1) // 2
            else:
                w = c * (c - 1) * (c - 2) // 6

            total_weight += w
            supported_records += 1

            m = eval_counts[record]

            if m > 0:
                covered_weight += w
                covered_records += 1
                weighted_failure += (
                    w * eval_failures[record] / m
                )

        if total_weight:
            coverage = covered_weight / total_weight
            lower = weighted_failure / total_weight
            upper = lower + (1 - coverage)
        else:
            coverage = np.nan
            lower = np.nan
            upper = np.nan

        rows.append({
            "distance": distance,
            "alpha": alpha,
            "baseline_LER": total_failures / evaluation_shots,
            "supported_records": supported_records,
            "covered_records": covered_records,
            "weight_coverage": coverage,
            "covered_contribution": lower,
            "unmatched_mass": (
                1 - coverage if total_weight else np.nan
            ),
            "plug_in_lower": lower,
            "plug_in_upper": upper
        })

    return pd.DataFrame(rows)


results = pd.concat([
    resampling_coverage_diagnostic(distance=3),
    resampling_coverage_diagnostic(distance=5)
], ignore_index=True)

print(results.to_string(index=False))

results.to_csv(
    "resampling_coverage_diagnostic.csv",
    index=False
)
```

```python

import numpy as np
import pandas as pd
import pymatching
from collections import Counter
from math import comb

def dominant_record_diagnostic(
    distance=5,
    physical_error=0.005,
    calibration_shots=10_000_000,
    evaluation_shots=10_000_000,
    batch_size=10_000,
    top_k=10
):
    circuit = build_surface_code(distance, physical_error)

    decoder = pymatching.Matching.from_detector_error_model(
        circuit.detector_error_model(decompose_errors=True)
    )

    sampler = circuit.compile_detector_sampler()

    calibration = Counter()
    evaluation = Counter()
    failures_by_record = Counter()

    # Calibration
    for start in range(0, calibration_shots, batch_size):
        n = min(batch_size, calibration_shots - start)
        records = sampler.sample(shots=n, bit_packed=True)
        calibration.update(map(bytes, records))

    # Independent evaluation
    for start in range(0, evaluation_shots, batch_size):
        n = min(batch_size, evaluation_shots - start)

        detectors, actual = sampler.sample(
            shots=n,
            separate_observables=True
        )

        predicted = decoder.decode_batch(detectors)
        failed = np.any(predicted != actual, axis=1)

        packed = np.packbits(
            detectors, axis=1, bitorder="little"
        )

        for record, outcome in zip(map(bytes, packed), failed):
            evaluation[record] += 1
            failures_by_record[record] += int(outcome)

    output = {}

    for alpha in [2, 3]:
        weights = {
            s: comb(c, alpha)
            for s, c in calibration.items()
            if c >= alpha
        }

        total_weight = sum(weights.values())

        if total_weight == 0:
            output[alpha] = pd.DataFrame()
            continue

        q = np.array(
            list(weights.values()), dtype=float
        ) / total_weight

        effective_support = 1 / np.sum(q ** 2)

        rows = []

        for record, weight in sorted(
            weights.items(),
            key=lambda x: x[1],
            reverse=True
        )[:top_k]:

            m = evaluation[record]
            e = failures_by_record[record]

            rows.append({
                "alpha": alpha,
                "calibration_count": calibration[record],
                "evaluation_count": m,
                "evaluation_failures": e,
                "conditional_LER": e / m if m else np.nan,
                "resampling_weight": weight / total_weight,
                "is_all_zero": all(b == 0 for b in record),
                "effective_support": effective_support
            })

        output[alpha] = pd.DataFrame(rows)

    return output


diagnostics = dominant_record_diagnostic()

for alpha, table in diagnostics.items():
    print(f"\nAlpha = {alpha}")
    print(table.to_string(index=False))
```

```python

# ============================================================
# CIRCUIT-LEVEL SYNDROME RESAMPLING
# Held-out evaluation + conservative confidence bounds
# ============================================================

%pip -q install stim pymatching numpy pandas scipy matplotlib

import os
import json
import math
import gzip
import pickle
from collections import Counter
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import stim
import pymatching
from scipy.stats import beta

# -------------------- CONFIGURATION -------------------------

DISTANCES = [3, 5, 7]
PHYSICAL_ERROR = 0.005

CALIBRATION_SHOTS = 100_000
EVALUATION_SHOTS = 100_000

BATCH_SIZE = 10_000
ALPHAS = [1, 2, 3]
CONFIDENCE = 0.95

# Increase shot budgets only after checking initial results.
OUTPUT_DIR = "qec_resampling_validation"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# -------------------- CIRCUIT -------------------------------

def build_circuit(distance, p):
    return stim.Circuit.generated(
        "surface_code:rotated_memory_x",
        distance=distance,
        rounds=distance,
        after_clifford_depolarization=p,
        before_round_data_depolarization=p,
        after_reset_flip_probability=p,
        before_measure_flip_probability=p
    )

# -------------------- SAMPLING ------------------------------

def collect_calibration(circuit, shots):
    sampler = circuit.compile_detector_sampler()
    counts = Counter()

    for start in range(0, shots, BATCH_SIZE):
        n = min(BATCH_SIZE, shots - start)
        records = sampler.sample(shots=n, bit_packed=True)
        counts.update(map(bytes, records))

    assert sum(counts.values()) == shots
    return counts


def collect_evaluation(circuit, shots):
    dem = circuit.detector_error_model(
        decompose_errors=True
    )

    decoder = pymatching.Matching.from_detector_error_model(dem)
    sampler = circuit.compile_detector_sampler()

    counts = Counter()
    failures_by_record = Counter()
    total_failures = 0

    for start in range(0, shots, BATCH_SIZE):
        n = min(BATCH_SIZE, shots - start)

        detectors, actual = sampler.sample(
            shots=n,
            separate_observables=True
        )

        predicted = decoder.decode_batch(detectors)

        failed = np.any(predicted != actual, axis=1)

        packed = np.packbits(
            detectors,
            axis=1,
            bitorder="little"
        )

        for record, outcome in zip(map(bytes, packed), failed):
            counts[record] += 1
            failures_by_record[record] += int(outcome)

        total_failures += int(failed.sum())

    assert sum(counts.values()) == shots
    assert sum(failures_by_record.values()) == total_failures

    return counts, failures_by_record, total_failures

# -------------------- STATISTICS ----------------------------

def clopper_pearson(k, n, error_probability):
    """
    Two-sided Clopper-Pearson interval.
    error_probability is the total interval error level.
    """
    if n == 0:
        return 0.0, 1.0

    lower = (
        0.0 if k == 0
        else beta.ppf(error_probability / 2, k, n-k+1)
    )

    upper = (
        1.0 if k == n
        else beta.ppf(1-error_probability/2, k+1, n-k)
    )

    return float(lower), float(upper)


def calculate_resampling(
    calibration,
    evaluation,
    failures,
    alpha,
    confidence=0.95
):
    weights = {
        record: math.comb(count, alpha)
        for record, count in calibration.items()
        if count >= alpha
    }

    if not weights:
        return {
            "alpha": alpha,
            "defined": False,
            "supported_records": 0
        }

    total_weight = sum(weights.values())

    probabilities = np.array(
        [w / total_weight for w in weights.values()],
        dtype=float
    )

    effective_support = 1 / np.sum(probabilities ** 2)

    covered_weight = 0
    plug_in_contribution = 0.0

    # Bonferroni correction over covered records.
    # Conditional on calibration, this provides simultaneous
    # coverage for their true conditional failure rates.
    covered_records = [
        s for s in weights if evaluation[s] > 0
    ]

    number_of_tests = max(1, len(covered_records))
    error_per_test = (
        (1-confidence) / number_of_tests
    )

    confidence_lower = 0.0
    confidence_upper = 0.0

    for record, weight in weights.items():
        q = weight / total_weight
        n = evaluation[record]

        if n == 0:
            # No evaluation data: failure probability in [0, 1]
            confidence_upper += q
            continue

        covered_weight += weight
        k = failures[record]

        plug_in_contribution += q * k / n

        lo, hi = clopper_pearson(
            k, n, error_per_test
        )

        confidence_lower += q * lo
        confidence_upper += q * hi

    coverage = covered_weight / total_weight
    unmatched_mass = 1 - coverage

    return {
        "alpha": alpha,
        "defined": True,
        "supported_records": len(weights),
        "covered_records": len(covered_records),
        "weight_coverage": coverage,
        "unmatched_mass": unmatched_mass,
        "effective_support": effective_support,
        "max_weight": float(probabilities.max()),
        "covered_contribution": plug_in_contribution,
        "plug_in_lower": plug_in_contribution,
        "plug_in_upper": (
            plug_in_contribution + unmatched_mass
        ),
        "conditional_CI_lower": confidence_lower,
        "conditional_CI_upper": confidence_upper,
        "confidence_level": confidence
    }

# -------------------- MAIN EXPERIMENT -----------------------

all_results = []
run_config = {
    "distances": DISTANCES,
    "physical_error": PHYSICAL_ERROR,
    "calibration_shots": CALIBRATION_SHOTS,
    "evaluation_shots": EVALUATION_SHOTS,
    "batch_size": BATCH_SIZE,
    "alphas": ALPHAS,
    "confidence": CONFIDENCE,
    "stim_version": stim.__version__,
    "timestamp_utc": datetime.now(timezone.utc).isoformat()
}

with open(
    os.path.join(OUTPUT_DIR, "config.json"), "w"
) as f:
    json.dump(run_config, f, indent=2)

for distance in DISTANCES:
    print(f"\n{'='*55}")
    print(f"Distance {distance}, p={PHYSICAL_ERROR}")
    print(f"{'='*55}")

    circuit = build_circuit(distance, PHYSICAL_ERROR)

    print("Detectors:", circuit.num_detectors)
    print("Collecting calibration data...")

    calibration = collect_calibration(
        circuit, CALIBRATION_SHOTS
    )

    print("Collecting evaluation data...")

    evaluation, failures, total_failures = collect_evaluation(
        circuit, EVALUATION_SHOTS
    )

    baseline = total_failures / EVALUATION_SHOTS

    baseline_lo, baseline_hi = clopper_pearson(
        total_failures,
        EVALUATION_SHOTS,
        1-CONFIDENCE
    )

    print(f"Baseline LER: {baseline:.8f}")
    print(f"Baseline 95% CI: [{baseline_lo:.8f}, "
          f"{baseline_hi:.8f}]")

    # Preserve complete record-level data
    archive_path = os.path.join(
        OUTPUT_DIR,
        f"distance_{distance}_counts.pkl.gz"
    )

    with gzip.open(archive_path, "wb") as f:
        pickle.dump({
            "calibration": calibration,
            "evaluation": evaluation,
            "failures": failures,
            "total_failures": total_failures
        }, f, protocol=pickle.HIGHEST_PROTOCOL)

    for alpha in ALPHAS:
        result = calculate_resampling(
            calibration,
            evaluation,
            failures,
            alpha,
            CONFIDENCE
        )

        result.update({
            "distance": distance,
            "physical_error": PHYSICAL_ERROR,
            "calibration_shots": CALIBRATION_SHOTS,
            "evaluation_shots": EVALUATION_SHOTS,
            "baseline_LER": baseline,
            "baseline_CI_lower": baseline_lo,
            "baseline_CI_upper": baseline_hi
        })

        all_results.append(result)

    # Save partial progress after every distance
    pd.DataFrame(all_results).to_csv(
        os.path.join(OUTPUT_DIR, "results.csv"),
        index=False
    )

# -------------------- RESULTS -------------------------------

df = pd.DataFrame(all_results)

display_columns = [
    "distance",
    "alpha",
    "baseline_LER",
    "weight_coverage",
    "effective_support",
    "max_weight",
    "covered_contribution",
    "conditional_CI_lower",
    "conditional_CI_upper"
]

print("\nFINAL RESULTS")
print(df[display_columns].to_string(index=False))

print("\nSaved files to:", OUTPUT_DIR)

# Optional Colab download as a single archive
import shutil
shutil.make_archive(
    OUTPUT_DIR,
    "zip",
    OUTPUT_DIR
)

print("Archive:", OUTPUT_DIR + ".zip")
```

```python

# ============================================================
# DOMINANT-RECORD CONFIDENCE ANALYSIS
# Reuses saved calibration/evaluation counters
# No new quantum circuit simulations
# ============================================================

import os
import gzip
import pickle
import math
import numpy as np
import pandas as pd
from scipy.stats import beta

OUTPUT_DIR = "qec_resampling_validation"
CONFIDENCE = 0.95
ALPHAS = [1, 2, 3]
TOP_K_VALUES = [1, 2, 5, 10, 20, 50, 100, 250, 500]

# A separate 95% interval is constructed for each candidate
# K. We adjust across all candidate K and alpha values
# within each distance to protect against selecting K
# after looking at the evaluation data.

DISTANCES = [3, 5, 7]

def cp_interval(k, n, delta):
    if n == 0:
        return 0.0, 1.0

    lo = (
        0.0 if k == 0
        else beta.ppf(delta / 2, k, n-k+1)
    )

    hi = (
        1.0 if k == n
        else beta.ppf(1-delta/2, k+1, n-k)
    )

    return float(lo), float(hi)


def analyze_top_k(
    calibration,
    evaluation,
    failures,
    alpha,
    top_k,
    delta
):
    weights = [
        (record, math.comb(count, alpha))
        for record, count in calibration.items()
        if count >= alpha
    ]

    if not weights:
        return None

    total_weight = sum(w for _, w in weights)

    weights.sort(
        key=lambda item: item[1],
        reverse=True
    )

    selected = weights[:top_k]
    selected_count = len(selected)

    # Simultaneous confidence intervals for selected records.
    per_record_delta = delta / selected_count

    covered_mass = 0.0
    lower = 0.0
    upper_selected = 0.0
    point_contribution = 0.0

    selected_evaluation_shots = 0
    selected_failures = 0

    for record, weight in selected:
        q = weight / total_weight
        n = evaluation.get(record, 0)
        k = failures.get(record, 0)

        covered_mass += q
        selected_evaluation_shots += n
        selected_failures += k

        if n > 0:
            lo, hi = cp_interval(
                k, n, per_record_delta
            )
            point_contribution += q * k / n
        else:
            lo, hi = 0.0, 1.0

        lower += q * lo
        upper_selected += q * hi

    residual_mass = max(0.0, 1.0-covered_mass)

    upper = min(
        1.0,
        upper_selected + residual_mass
    )

    return {
        "alpha": alpha,
        "top_k": top_k,
        "selected_records": selected_count,
        "covered_mass": covered_mass,
        "residual_mass": residual_mass,
        "selected_evaluation_shots":
            selected_evaluation_shots,
        "selected_failures": selected_failures,
        "point_contribution": point_contribution,
        "confidence_lower": lower,
        "confidence_upper": upper,
        "dominant_weight":
            weights[0][1] / total_weight
    }


all_rows = []

# Bonferroni adjustment across the candidate K values
# and alpha values within each distance.
number_of_candidates = len(TOP_K_VALUES) * len(ALPHAS)
delta_per_candidate = (
    (1-CONFIDENCE) / number_of_candidates
)

for distance in DISTANCES:
    path = os.path.join(
        OUTPUT_DIR,
        f"distance_{distance}_counts.pkl.gz"
    )

    if not os.path.exists(path):
        print(f"Skipping distance {distance}: no saved data")
        continue

    with gzip.open(path, "rb") as f:
        data = pickle.load(f)

    calibration = data["calibration"]
    evaluation = data["evaluation"]
    failures = data["failures"]

    print(f"\nAnalyzing distance {distance}...")

    for alpha in ALPHAS:
        for top_k in TOP_K_VALUES:
            result = analyze_top_k(
                calibration,
                evaluation,
                failures,
                alpha,
                top_k,
                delta_per_candidate
            )

            if result is None:
                continue

            result["distance"] = distance
            result["confidence_level"] = CONFIDENCE
            all_rows.append(result)


df = pd.DataFrame(all_rows)

if df.empty:
    raise RuntimeError(
        "No valid resampling distributions were found."
    )

# Select the smallest valid upper bound among candidate K.
# The multiple-comparison adjustment above permits this
# selection without invalidating the simultaneous guarantee.

best = (
    df.sort_values("confidence_upper")
      .groupby(["distance", "alpha"], as_index=False)
      .first()
      .sort_values(["distance", "alpha"])
)

print("\nBEST CONFIDENCE BOUNDS")
columns = [
    "distance",
    "alpha",
    "top_k",
    "dominant_weight",
    "covered_mass",
    "residual_mass",
    "selected_evaluation_shots",
    "selected_failures",
    "confidence_lower",
    "confidence_upper"
]

print(best[columns].to_string(index=False))

print("\nALL CANDIDATE RESULTS")
print(
    df[[
        "distance", "alpha", "top_k",
        "covered_mass", "confidence_upper"
    ]].to_string(index=False)
)

best.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "dominant_record_best_bounds.csv"
    ),
    index=False
)

df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "dominant_record_all_candidates.csv"
    ),
    index=False
)

print("\nResults saved successfully.")
```

```python

# ============================================================
# CALIBRATION-AWARE SYNDROME RESAMPLING BOUNDS
# No new Stim simulations required
# ============================================================

import os
import gzip
import pickle
import math
import numpy as np
import pandas as pd
from scipy.stats import beta

OUTPUT_DIR = "qec_resampling_validation"
DISTANCES = [3, 5, 7]
ALPHAS = [1, 2, 3]
TOP_K_VALUES = [1, 2, 5, 10, 20, 50, 100, 250, 500]

CONFIDENCE = 0.95

def cp_bounds(k, n, delta):
    if n == 0:
        return 0.0, 1.0

    lo = (
        0.0 if k == 0
        else beta.ppf(delta / 2, k, n-k+1)
    )
    hi = (
        1.0 if k == n
        else beta.ppf(1-delta/2, k+1, n-k)
    )
    return float(lo), float(hi)


def calibration_aware_bound(
    calibration,
    evaluation,
    failures,
    alpha,
    top_k,
    delta
):
    N = sum(calibration.values())

    # Select records by calibration frequency only.
    selected = calibration.most_common(top_k)

    if not selected:
        return None

    m = len(selected)

    # Allocate error budget to calibration and evaluation.
    # Bonferroni over all selected-record intervals.
    per_interval_delta = delta / (2 * m)

    lower_p = []
    upper_p = []
    lower_f = []
    upper_f = []

    for record, count in selected:
        p_lo, p_hi = cp_bounds(
            count, N, per_interval_delta
        )

        n_eval = evaluation.get(record, 0)
        k_fail = failures.get(record, 0)

        f_lo, f_hi = cp_bounds(
            k_fail,
            n_eval,
            per_interval_delta
        )

        lower_p.append(p_lo)
        upper_p.append(p_hi)
        lower_f.append(f_lo)
        upper_f.append(f_hi)

    lower_p = np.asarray(lower_p)
    upper_p = np.asarray(upper_p)
    lower_f = np.asarray(lower_f)
    upper_f = np.asarray(upper_f)

    # Upper bound on remaining original probability mass
    residual_p_upper = min(
        1.0,
        max(0.0, 1.0 - lower_p.sum())
    )

    # Lower bound on remaining probability mass
    residual_p_lower = max(
        0.0,
        1.0 - upper_p.sum()
    )

    # Selected-record power sums
    selected_power_lower = np.sum(lower_p ** alpha)
    selected_power_upper = np.sum(upper_p ** alpha)

    # Upper bound on unselected power sum.
    residual_power_upper = residual_p_upper ** alpha

    # Lower bound on the true denominator
    denominator_lower = selected_power_lower

    # Upper bound on the true denominator
    denominator_upper = (
        selected_power_upper + residual_power_upper
    )

    # Conservative numerator bounds
    numerator_lower = np.sum(
        (lower_p ** alpha) * lower_f
    )

    numerator_upper = (
        np.sum((upper_p ** alpha) * upper_f)
        + residual_power_upper
    )

    lower_bound = (
        numerator_lower / denominator_upper
        if denominator_upper > 0 else 0.0
    )

    upper_bound = (
        min(1.0, numerator_upper / denominator_lower)
        if denominator_lower > 0 else 1.0
    )

    empirical_selected_mass = (
        sum(c for _, c in selected) / N
    )

    return {
        "alpha": alpha,
        "top_k": top_k,
        "selected_records": m,
        "empirical_selected_P_mass":
            empirical_selected_mass,
        "residual_P_upper":
            residual_p_upper,
        "selected_power_lower":
            selected_power_lower,
        "residual_power_upper":
            residual_power_upper,
        "calibration_aware_lower":
            max(0.0, lower_bound),
        "calibration_aware_upper":
            upper_bound
    }


all_results = []

# Adjustment across all candidate alpha/K combinations
# within each distance.
candidate_count = len(ALPHAS) * len(TOP_K_VALUES)
delta_per_candidate = (
    (1-CONFIDENCE) / candidate_count
)

for distance in DISTANCES:
    path = os.path.join(
        OUTPUT_DIR,
        f"distance_{distance}_counts.pkl.gz"
    )

    if not os.path.exists(path):
        print(f"Skipping distance {distance}")
        continue

    with gzip.open(path, "rb") as f:
        data = pickle.load(f)

    calibration = data["calibration"]
    evaluation = data["evaluation"]
    failures = data["failures"]

    print(f"\nAnalyzing distance {distance}")

    for alpha in ALPHAS:
        for top_k in TOP_K_VALUES:
            result = calibration_aware_bound(
                calibration,
                evaluation,
                failures,
                alpha,
                top_k,
                delta_per_candidate
            )

            if result is None:
                continue

            result["distance"] = distance
            all_results.append(result)


df = pd.DataFrame(all_results)

best = (
    df.sort_values("calibration_aware_upper")
      .groupby(["distance", "alpha"], as_index=False)
      .first()
      .sort_values(["distance", "alpha"])
)

print("\nBEST CALIBRATION-AWARE BOUNDS")

columns = [
    "distance",
    "alpha",
    "top_k",
    "empirical_selected_P_mass",
    "residual_P_upper",
    "calibration_aware_lower",
    "calibration_aware_upper"
]

print(best[columns].to_string(index=False))

df.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "calibration_aware_all.csv"
    ),
    index=False
)

best.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "calibration_aware_best.csv"
    ),
    index=False
)

print("\nSaved calibration-aware results.")
```

```python

%pip -q install stim pymatching numpy pandas scipy matplotlib

import os
import math
import numpy as np
import pandas as pd
import stim
import pymatching
import matplotlib.pyplot as plt
from collections import Counter
from scipy.stats import t

# ============================================================
# CONFIGURATION
# ============================================================

DISTANCES = [3, 5]
P = 0.005
ALPHAS = [1, 2, 3]

BATCHES = 20
SHOTS_PER_BATCH = 100_000
CHUNK_SIZE = 10_000

OUTPUT_DIR = "qec_collision_batches"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ============================================================
# CIRCUIT AND DECODER
# ============================================================

def build_circuit(d, p):
    return stim.Circuit.generated(
        "surface_code:rotated_memory_x",
        distance=d,
        rounds=d,
        after_clifford_depolarization=p,
        before_round_data_depolarization=p,
        after_reset_flip_probability=p,
        before_measure_flip_probability=p
    )

# ============================================================
# COLLISION ESTIMATOR
# ============================================================

def falling_factorial(x, k):
    result = 1
    for j in range(k):
        result *= (x-j)
    return result


def estimate_collision_ratio(counts, failures, alpha):
    N = sum(counts.values())

    if N < alpha:
        return np.nan, np.nan, np.nan, 0

    denominator = falling_factorial(N, alpha)

    collision_total = 0
    failure_total = 0

    for syndrome, count in counts.items():
        if count < alpha:
            continue

        collision_total += falling_factorial(count, alpha)

        failure_total += (
            failures.get(syndrome, 0)
            * falling_factorial(count-1, alpha-1)
        )

    if collision_total == 0:
        return 0.0, 0.0, np.nan, 0

    C_hat = collision_total / denominator
    F_hat = failure_total / denominator

    return (
        C_hat,
        F_hat,
        F_hat / C_hat,
        collision_total
    )

# ============================================================
# MAIN EXPERIMENT
# ============================================================

rows = []

for d in DISTANCES:
    print(f"\n{'='*60}")
    print(f"Distance {d}")
    print(f"{'='*60}")

    circuit = build_circuit(d, P)
    sampler = circuit.compile_detector_sampler()

    dem = circuit.detector_error_model(
        decompose_errors=True
    )
    decoder = pymatching.Matching.from_detector_error_model(dem)

    for batch in range(BATCHES):
        counts = Counter()
        failures = Counter()

        total_failures = 0

        for start in range(0, SHOTS_PER_BATCH, CHUNK_SIZE):
            n = min(CHUNK_SIZE, SHOTS_PER_BATCH-start)

            detectors, actual = sampler.sample(
                shots=n,
                separate_observables=True
            )

            predicted = decoder.decode_batch(detectors)
            failed = np.any(predicted != actual, axis=1)

            packed = np.packbits(
                detectors,
                axis=1,
                bitorder="little"
            )

            for syndrome, outcome in zip(
                map(bytes, packed), failed
            ):
                counts[syndrome] += 1
                failures[syndrome] += int(outcome)

            total_failures += int(failed.sum())

        baseline = total_failures / SHOTS_PER_BATCH

        for alpha in ALPHAS:
            C, F, ratio, collisions = estimate_collision_ratio(
                counts, failures, alpha
            )

            rows.append({
                "distance": d,
                "batch": batch,
                "alpha": alpha,
                "shots": SHOTS_PER_BATCH,
                "baseline_LER": baseline,
                "C_hat": C,
                "F_hat": F,
                "resampled_LER": ratio,
                "collision_total": collisions,
                "unique_records": len(counts)
            })

        print(
            f"Batch {batch+1}/{BATCHES}: "
            f"baseline={baseline:.5f}, "
            f"unique={len(counts)}"
        )

    pd.DataFrame(rows).to_csv(
        os.path.join(OUTPUT_DIR, "batch_results.csv"),
        index=False
    )

# ============================================================
# SUMMARY
# ============================================================

df = pd.DataFrame(rows)
summary_rows = []

for (d, alpha), group in df.groupby(
    ["distance", "alpha"]
):
    valid = group["resampled_LER"].dropna()
    n = len(valid)

    if n >= 2:
        mean = valid.mean()
        sd = valid.std(ddof=1)
        margin = t.ppf(0.975, n-1) * sd / np.sqrt(n)

        lo = max(0.0, mean-margin)
        hi = min(1.0, mean+margin)
    elif n == 1:
        mean = valid.iloc[0]
        lo, hi = np.nan, np.nan
    else:
        mean = np.nan
        lo, hi = np.nan, np.nan

    summary_rows.append({
        "distance": d,
        "alpha": alpha,
        "valid_batches": n,
        "mean_resampled_LER": mean,
        "batch_CI_lower": lo,
        "batch_CI_upper": hi,
        "mean_C_hat": group["C_hat"].mean(),
        "mean_F_hat": group["F_hat"].mean(),
        "mean_baseline_LER":
            group["baseline_LER"].mean()
    })

summary = pd.DataFrame(summary_rows)

print("\nFINAL SUMMARY")
print(summary.to_string(index=False))

summary.to_csv(
    os.path.join(OUTPUT_DIR, "summary.csv"),
    index=False
)

# ============================================================
# VISUALIZATION
# ============================================================

for d in DISTANCES:
    subset = summary[summary["distance"] == d]

    plt.figure(figsize=(7, 4))

    plt.errorbar(
        subset["alpha"],
        subset["mean_resampled_LER"],
        yerr=[
            subset["mean_resampled_LER"]
            - subset["batch_CI_lower"],
            subset["batch_CI_upper"]
            - subset["mean_resampled_LER"]
        ],
        fmt="o-",
        capsize=5
    )

    plt.axhline(
        subset["mean_baseline_LER"].iloc[0],
        linestyle="--",
        label="Baseline LER"
    )

    plt.xlabel("Resampling strength α")
    plt.ylabel("Estimated resampled logical error rate")
    plt.title(f"Distance {d}: collision-based estimates")
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()

print("\nSaved results to:", OUTPUT_DIR)
```

```python

import os, gzip, pickle, math
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# CERTIFICATION PILOT
# Target: true resampled LER < 1e-3
# ============================================================

DATA_PATH = "qec_resampling_validation/distance_5_counts.pkl.gz"

ALPHAS = [2, 3]
TOP_K_VALUES = [1, 2, 5, 10, 20, 50, 100]
TARGET = 1e-3
DELTA = 0.05

with gzip.open(DATA_PATH, "rb") as f:
    data = pickle.load(f)

cal = data["calibration"]
ev = data["evaluation"]
fail = data["failures"]

N = sum(cal.values())
M = sum(ev.values())

print("Calibration shots:", N)
print("Evaluation shots:", M)

def cp_upper(k, n, delta):
    if n == 0:
        return 1.0
    if k == n:
        return 1.0
    return float(beta.ppf(1-delta, k+1, n-k))

def cp_lower(k, n, delta):
    if n == 0 or k == 0:
        return 0.0
    return float(beta.ppf(delta, k, n-k+1))

# A conservative upper bound on the probability of
# any completely unseen record.
#
# If a record has probability > u, the probability
# it is absent from N independent draws is < (1-u)^N.
#
# A union bound cannot be taken over an infinite
# alphabet directly. But there can be at most 1/u
# records with probability > u.
#
# Hence P(any record with probability > u is unseen)
# <= (1/u)*(1-u)^N.
#
# We solve for u at a specified error level.

def unseen_probability_bound(n, delta):
    lo, hi = 1e-15, 1.0

    for _ in range(100):
        mid = (lo+hi)/2
        risk = (1-mid)**n / mid

        if risk > delta:
            lo = mid
        else:
            hi = mid

    return hi

# Multiple candidate configurations and alpha values.
num_tests = len(ALPHAS)*len(TOP_K_VALUES)
delta_test = DELTA / num_tests

rows = []

for alpha in ALPHAS:
    for top_k in TOP_K_VALUES:

        selected = cal.most_common(top_k)
        selected_set = {s for s,_ in selected}

        # Allocate error probability:
        # 40% calibration selected-record intervals
        # 40% evaluation intervals
        # 20% unseen-record bound
        d_cal = 0.4*delta_test
        d_eval = 0.4*delta_test
        d_unseen = 0.2*delta_test

        m = len(selected)
        cal_delta_each = d_cal/m
        eval_delta_each = d_eval/m

        p_lo = []
        p_hi = []
        f_hi = []

        for s,c in selected:
            p_lo.append(cp_lower(c,N,cal_delta_each))
            p_hi.append(cp_upper(c,N,cal_delta_each))
            f_hi.append(cp_upper(
                fail.get(s,0),
                ev.get(s,0),
                eval_delta_each
            ))

        p_lo = np.array(p_lo)
        p_hi = np.array(p_hi)
        f_hi = np.array(f_hi)

        # Residual original probability mass.
        residual_mass_upper = min(
            1.0,
            max(0.0,1.0-p_lo.sum())
        )

        # Maximum probability among unselected observed
        # records: simultaneous CP upper bounds.
        #
        # Include all observed unselected records in
        # Bonferroni correction.
        unselected = [
            (s,c) for s,c in cal.items()
            if s not in selected_set
        ]

        d_other = d_cal / 2
        d_selected = d_cal / 2

        # Recalculate selected intervals using their
        # reduced allocation.
        p_lo = np.array([
            cp_lower(c,N,d_selected/m)
            for s,c in selected
        ])

        p_hi = np.array([
            cp_upper(c,N,d_selected/m)
            for s,c in selected
        ])

        residual_mass_upper = min(
            1.0,max(0.0,1.0-p_lo.sum())
        )

        if unselected:
            per_other = d_other/len(unselected)
            max_observed_upper = max(
                cp_upper(c,N,per_other)
                for s,c in unselected
            )
        else:
            max_observed_upper = 0.0

        max_unseen_upper = unseen_probability_bound(
            N,d_unseen
        )

        max_residual_probability = max(
            max_observed_upper,
            max_unseen_upper
        )

        # For alpha >= 2:
        # sum_residual P(s)^alpha
        # <= max_residual^(alpha-1)*residual_mass
        residual_power_upper = (
            max_residual_probability**(alpha-1)
            * residual_mass_upper
        )

        denominator_lower = np.sum(p_lo**alpha)

        numerator_upper = (
            np.sum((p_hi**alpha)*f_hi)
            + residual_power_upper
        )

        upper = (
            min(1.0,numerator_upper/denominator_lower)
            if denominator_lower>0 else 1.0
        )

        rows.append({
            "alpha":alpha,
            "top_k":top_k,
            "selected_P_mass":sum(c for _,c in selected)/N,
            "max_residual_probability":
                max_residual_probability,
            "residual_power_upper":
                residual_power_upper,
            "upper_bound":upper,
            "certified_below_target":upper<TARGET
        })

results = pd.DataFrame(rows)

best = (
    results.sort_values("upper_bound")
    .groupby("alpha",as_index=False)
    .first()
)

print("\nBEST CERTIFICATION RESULTS")
print(best.to_string(index=False))

print("\nALL CANDIDATES")
print(results.to_string(index=False))

results.to_csv(
    "qec_resampling_validation/certification_candidates.csv",
    index=False
)

print("\nTarget:",TARGET)
print("Confidence level:",1-DELTA)
```

```python

import os
import numpy as np
import pandas as pd
from scipy.stats import t

# ============================================================
# INDEPENDENT-BATCH COLLISION MOMENT ANALYSIS
# ============================================================

DATA_PATH = "qec_collision_batches/batch_results.csv"
OUTPUT_DIR = "qec_collision_batches"

TARGET = 1e-3
CONFIDENCE = 0.95
DELTA = 1-CONFIDENCE

df = pd.read_csv(DATA_PATH)

required = [
    "distance", "batch", "alpha", "C_hat", "F_hat"
]

for col in required:
    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

def hoeffding_ratio_bound(C, F, delta):
    """
    Simultaneous one-sided Hoeffding bounds
    for independent bounded batch estimates.
    """
    B = len(C)

    eps = np.sqrt(
        np.log(2/delta)/(2*B)
    )

    C_lower = max(0.0, C.mean()-eps)
    F_upper = min(1.0, F.mean()+eps)

    ratio_upper = (
        min(1.0, F_upper/C_lower)
        if C_lower > 0 else 1.0
    )

    return C_lower, F_upper, ratio_upper, eps


def exploratory_bernstein(C, F, delta):
    """
    Variance-sensitive exploratory bound.

    Uses an empirical variance correction.
    Not used as the formal certificate.
    """
    B = len(C)

    if B < 2:
        return np.nan, np.nan, np.nan

    log_term = np.log(4/delta)

    def radius(x):
        variance = np.var(x, ddof=1)
        return (
            np.sqrt(2*variance*log_term/B)
            + 7*log_term/(3*(B-1))
        )

    C_lower = max(0.0, C.mean()-radius(C))
    F_upper = min(1.0, F.mean()+radius(F))

    ratio_upper = (
        min(1.0, F_upper/C_lower)
        if C_lower > 0 else 1.0
    )

    return C_lower, F_upper, ratio_upper


rows = []

# Joint 95% coverage across all distance/alpha pairs.
groups = list(df.groupby(["distance", "alpha"]))
delta_per_group = DELTA / len(groups)

for (distance, alpha), group in groups:
    C = group["C_hat"].to_numpy(dtype=float)
    F = group["F_hat"].to_numpy(dtype=float)

    if not (
        np.isfinite(C).all()
        and np.isfinite(F).all()
    ):
        raise ValueError(
            f"Nonfinite moments at d={distance}, alpha={alpha}"
        )

    if not (
        ((C >= 0) & (C <= 1)).all()
        and ((F >= 0) & (F <= 1)).all()
    ):
        raise ValueError("Moments must lie in [0,1].")

    C_lo, F_hi, H_upper, eps = (
        hoeffding_ratio_bound(
            C, F, delta_per_group
        )
    )

    B_C_lo, B_F_hi, B_upper = (
        exploratory_bernstein(
            C, F, delta_per_group
        )
    )

    ratio_of_means = (
        F.mean()/C.mean()
        if C.mean()>0 else np.nan
    )

    rows.append({
        "distance": distance,
        "alpha": alpha,
        "batches": len(group),
        "mean_C": C.mean(),
        "mean_F": F.mean(),
        "ratio_of_means": ratio_of_means,
        "zero_F_batches": int((F==0).sum()),
        "hoeffding_radius": eps,
        "hoeffding_C_lower": C_lo,
        "hoeffding_F_upper": F_hi,
        "hoeffding_ratio_upper": H_upper,
        "exploratory_bernstein_upper": B_upper,
        "certified_below_target":
            H_upper < TARGET
    })

results = pd.DataFrame(rows)

print("\nRIGOROUS HOEFFDING RESULTS")
print(results[[
    "distance", "alpha", "batches",
    "mean_C", "mean_F",
    "ratio_of_means",
    "hoeffding_C_lower",
    "hoeffding_ratio_upper",
    "certified_below_target"
]].to_string(index=False))

print("\nEXPLORATORY VARIANCE-SENSITIVE RESULTS")
print(results[[
    "distance", "alpha",
    "zero_F_batches",
    "exploratory_bernstein_upper"
]].to_string(index=False))

results.to_csv(
    os.path.join(
        OUTPUT_DIR,
        "rigorous_moment_bounds.csv"
    ),
    index=False
)

print("\nTarget:", TARGET)
print("Joint confidence:", CONFIDENCE)
print("Saved results.")
```

```python

import numpy as np
import pandas as pd
from scipy.stats import t

# ============================================================
# DIRECT TARGET-INEQUALITY TEST
# H0: F_alpha - TARGET*C_alpha >= 0
# H1: F_alpha - TARGET*C_alpha < 0
# ============================================================

PATH = "qec_collision_batches/batch_results.csv"
TARGET = 1e-3
DELTA = 0.05

df = pd.read_csv(PATH)

groups = list(df.groupby(["distance", "alpha"]))
delta_group = DELTA / len(groups)

rows = []

for (d, alpha), g in groups:
    C = g["C_hat"].to_numpy(dtype=float)
    F = g["F_hat"].to_numpy(dtype=float)

    if not (
        np.isfinite(C).all()
        and np.isfinite(F).all()
        and np.all(C >= 0)
        and np.all(F >= 0)
        and np.all(F <= C + 1e-12)
        and np.all(C <= 1 + 1e-12)
    ):
        raise ValueError(
            f"Invalid moments for d={d}, alpha={alpha}"
        )

    Z = F - TARGET*C
    B = len(Z)

    mean_Z = Z.mean()
    sd_Z = Z.std(ddof=1)
    se_Z = sd_Z / np.sqrt(B)

    # Exploratory paired Student-t upper bound.
    t_upper = (
        mean_Z
        + t.ppf(1-delta_group, B-1)*se_Z
    )

    # Rigorous one-sided Hoeffding upper bound.
    # Z lies in [-TARGET, 1].
    width = 1 + TARGET

    h_radius = width*np.sqrt(
        np.log(1/delta_group)/(2*B)
    )

    h_upper = mean_Z + h_radius

    rows.append({
        "distance":d,
        "alpha":alpha,
        "batches":B,
        "mean_C":C.mean(),
        "mean_F":F.mean(),
        "mean_Z":mean_Z,
        "sd_Z":sd_Z,
        "negative_Z_batches":int((Z<0).sum()),
        "zero_F_batches":int((F==0).sum()),
        "exploratory_t_upper":t_upper,
        "rigorous_hoeffding_upper":h_upper,
        "t_suggests_target":t_upper<0,
        "rigorously_certified":h_upper<0
    })

results = pd.DataFrame(rows)

print("\nDIRECT TARGET-INEQUALITY RESULTS")
print(results.to_string(index=False))

print("\nINTERPRETATION")
for _,r in results.iterrows():
    print(
        f"d={int(r.distance)}, alpha={int(r.alpha)}: "
        f"mean Z={r.mean_Z:.6e}, "
        f"t upper={r.exploratory_t_upper:.6e}, "
        f"Hoeffding upper={r.rigorous_hoeffding_upper:.6e}"
    )

results.to_csv(
    "qec_collision_batches/direct_target_test.csv",
    index=False
)

print("\nTarget LER:", TARGET)
print("Joint confidence level:", 1-DELTA)
print("Saved direct_target_test.csv")
```

```python

import gzip, pickle, math
import numpy as np
import pandas as pd

PATH = "qec_resampling_validation/distance_5_counts.pkl.gz"
TAU = 1e-3

with gzip.open(PATH, "rb") as f:
    data = pickle.load(f)

counts = data["evaluation"]
failures = data["failures"]
N = sum(counts.values())

def falling(x, k):
    result = 1.0
    for j in range(k):
        result *= x-j
    return result

def overlap_moment(alpha, r):
    """
    Unbiased estimate of E[h(A)h(B)]
    for two alpha-tuples sharing r observations.
    """
    m = 2*alpha-r
    denom = falling(N, m)
    total = 0.0

    for s, c in counts.items():
        if c < m:
            continue

        e = failures.get(s, 0)
        mu = e/c

        pair = e*(e-1)/(c*(c-1))

        cross = (
            r*mu +
            (alpha*alpha-r)*pair
        )/(alpha*alpha)

        kernel_product = (
            cross - 2*TAU*mu + TAU**2
        )

        total += falling(c,m)*kernel_product

    return total/denom

rows = []

for alpha in [2,3]:
    # Disjoint tuples provide an unbiased
    # estimate of theta squared.
    U0 = overlap_moment(alpha,0)

    variance = 0.0

    for r in range(1,alpha+1):
        coefficient = (
            math.comb(alpha,r)
            * math.comb(N-alpha,alpha-r)
            / math.comb(N,alpha)
        )

        Ur = overlap_moment(alpha,r)
        component = coefficient*(Ur-U0)
        variance += component

        rows.append({
            "alpha":alpha,
            "overlap_r":r,
            "coefficient":coefficient,
            "moment":Ur,
            "variance_component":component
        })

    print(
        f"alpha={alpha}: "
        f"variance={variance:.8e}, "
        f"SD={np.sqrt(max(0,variance)):.8e}"
    )

results = pd.DataFrame(rows)

print("\nOVERLAP CONTRIBUTIONS")
print(results.to_string(index=False))

results.to_csv(
    "distance_5_overlap_variance.csv",
    index=False
)
```

```python

import itertools
import math
import numpy as np
import pandas as pd
from collections import Counter

TAU = 1e-3

# Small labeled datasets.
# Each entry is (syndrome, logical_failure).
DATASETS = {
    "mixed": [
        ("A",0), ("A",1), ("A",0), ("A",1),
        ("A",0), ("A",0), ("B",1), ("B",0),
        ("B",0), ("C",1)
    ],
    "all_success": [
        ("A",0), ("A",0), ("A",0), ("A",0),
        ("B",0), ("B",0), ("B",0), ("B",0)
    ],
    "all_failure": [
        ("A",1), ("A",1), ("A",1), ("A",1),
        ("B",1), ("B",1), ("B",1), ("B",1)
    ],
    "unique": [
        (str(i), i % 2) for i in range(8)
    ],
    "single_syndrome": [
        ("A",0), ("A",1), ("A",0), ("A",1),
        ("A",1), ("A",0), ("A",0), ("A",1)
    ]
}

def falling(n, k):
    if k > n:
        return 0
    return math.prod(n-j for j in range(k))

def brute_overlap(data, alpha, r):
    """
    Average h(A)*h(B) over ordered pairs
    of alpha-subsets with overlap exactly r.
    """
    n = len(data)
    tuples = list(itertools.combinations(range(n), alpha))

    kernels = {}
    for A in tuples:
        syndromes = [data[i][0] for i in A]
        if len(set(syndromes)) != 1:
            kernels[A] = 0.0
        else:
            kernels[A] = (
                sum(data[i][1] for i in A)/alpha - TAU
            )

    total = 0.0
    pairs = 0

    for A in tuples:
        SA = set(A)
        for B in tuples:
            if len(SA.intersection(B)) == r:
                total += kernels[A]*kernels[B]
                pairs += 1

    return total/pairs if pairs else np.nan

def count_overlap(data, alpha, r):
    """
    Exact count-based calculation of the
    finite-sample overlap moment.

    r > 0: both tuples must have the same syndrome.

    r = 0: tuples may have different syndromes.
    """
    n = len(data)
    m = 2*alpha-r

    if m > n:
        return np.nan

    c = Counter(s for s, f in data)
    e = Counter()
    for s, f in data:
        e[s] += f

    denom = falling(n, m)
    total = 0.0

    # Same-syndrome contribution
    for s, count in c.items():
        if count < m:
            continue

        mu = e[s]/count
        nu = (
            e[s]*(e[s]-1)/(count*(count-1))
            if count >= 2 else 0.0
        )

        cross = (
            r*mu + (alpha**2-r)*nu
        )/(alpha**2)

        product = cross - 2*TAU*mu + TAU**2

        total += falling(count, m)*product

    # Different-syndrome contribution.
    # Possible only for disjoint tuples.
    if r == 0:
        keys = list(c.keys())

        for s in keys:
            if c[s] < alpha:
                continue

            for t in keys:
                if s == t or c[t] < alpha:
                    continue

                mu_s = e[s]/c[s]
                mu_t = e[t]/c[t]

                total += (
                    falling(c[s], alpha)
                    * falling(c[t], alpha)
                    * (mu_s-TAU)
                    * (mu_t-TAU)
                )

    return total/denom

rows = []

for name, data in DATASETS.items():
    for alpha in [2, 3]:
        for r in range(alpha+1):

            brute = brute_overlap(data, alpha, r)
            counted = count_overlap(data, alpha, r)

            error = abs(brute-counted)

            rows.append({
                "dataset": name,
                "alpha": alpha,
                "overlap_r": r,
                "brute_force": brute,
                "count_formula": counted,
                "absolute_error": error,
                "pass": bool(np.isclose(
                    brute, counted,
                    rtol=1e-9, atol=1e-13
                ))
            })

results = pd.DataFrame(rows)

print("EXACT COMBINATORIAL VALIDATION")
print(results.to_string(index=False))

print("\nSUMMARY")
print("Tests:", len(results))
print("Passed:", int(results["pass"].sum()))
print("Failed:", int((~results["pass"]).sum()))
print("Maximum absolute error:",
      results["absolute_error"].max())

results.to_csv(
    "overlap_combinatorial_validation.csv",
    index=False
)

if not results["pass"].all():
    raise AssertionError("Overlap formula validation failed!")

print("\nALL TESTS PASSED")
```

```python

import gzip
import pickle
import math
import numpy as np
import pandas as pd
from collections import Counter

PATH = "qec_resampling_validation/distance_5_counts.pkl.gz"
TAU = 1e-3

with gzip.open(PATH, "rb") as f:
    data = pickle.load(f)

counts = data["evaluation"]
failures = data["failures"]
N = sum(counts.values())

def falling(n, k):
    if n < k:
        return 0.0
    result = 1.0
    for j in range(k):
        result *= (n-j)
    return result

def overlap_moment(alpha, r):
    """
    Count-based unbiased overlap moment.

    r=0 includes disjoint tuples with
    equal OR different syndromes.
    """
    m = 2*alpha-r
    if N < m:
        return np.nan

    denom = falling(N, m)
    same = 0.0

    for s, c in counts.items():
        if c < m:
            continue

        e = failures.get(s, 0)
        mu = e/c
        nu = e*(e-1)/(c*(c-1))

        cross = (
            r*mu + (alpha*alpha-r)*nu
        )/(alpha*alpha)

        product = cross - 2*TAU*mu + TAU**2
        same += falling(c, m)*product

    if r != 0:
        return same/denom

    # Disjoint tuples with different syndromes.
    # O(number of unique records), not O(K^2).
    total_A = 0.0
    total_A2 = 0.0

    for s, c in counts.items():
        if c < alpha:
            continue

        e = failures.get(s, 0)
        A = falling(c, alpha)*(e/c-TAU)
        total_A += A
        total_A2 += A*A

    different = total_A**2-total_A2
    return (same+different)/denom

def moment_estimates(alpha):
    C_num = 0.0
    F_num = 0.0

    for s, c in counts.items():
        if c < alpha:
            continue

        e = failures.get(s, 0)

        C_num += falling(c, alpha)
        F_num += e*falling(c-1, alpha-1)

    denom = falling(N, alpha)
    return C_num/denom, F_num/denom

previous_sd = {
    2: 1.137586e-8,
    3: 2.286224e-10
}

rows = []
summary = []

for alpha in [2, 3]:
    C, F = moment_estimates(alpha)
    theta = F-TAU*C
    M0 = overlap_moment(alpha, 0)

    variance = 0.0

    for r in range(1, alpha+1):
        coefficient = (
            math.comb(alpha, r)
            * math.comb(N-alpha, alpha-r)
            / math.comb(N, alpha)
        )

        Mr = overlap_moment(alpha, r)
        component = coefficient*(Mr-M0)
        variance += component

        rows.append({
            "alpha": alpha,
            "overlap_r": r,
            "coefficient": coefficient,
            "M_r": Mr,
            "M_0": M0,
            "variance_component": component
        })

    # An unbiased variance estimate can be negative
    # in a particular finite sample.
    sd = math.sqrt(variance) if variance >= 0 else np.nan

    summary.append({
        "alpha": alpha,
        "C_hat": C,
        "F_hat": F,
        "Z_hat": theta,
        "M0_hat": M0,
        "Z_hat_squared": theta**2,
        "variance_estimate": variance,
        "corrected_SD": sd,
        "observed_batch_SD": previous_sd[alpha],
        "SD_ratio": sd/previous_sd[alpha]
            if np.isfinite(sd) else np.nan
    })

details = pd.DataFrame(rows)
results = pd.DataFrame(summary)

print("CORRECTED VARIANCE RESULTS")
print(results.to_string(index=False))

print("\nOVERLAP CONTRIBUTIONS")
print(details.to_string(index=False))

results.to_csv(
    "distance_5_corrected_variance.csv",
    index=False
)
details.to_csv(
    "distance_5_corrected_overlap_terms.csv",
    index=False
)

print("\nSaved both CSV files.")
```

```python

import numpy as np
import pandas as pd
import stim
import pymatching
from scipy.stats import beta

# ============================================================
# EXACT DISJOINT-COLLISION CERTIFICATION
# ============================================================

DISTANCES = [3, 5]
ALPHAS = [2, 3]
P = 0.005
SHOTS_PER_DISTANCE = 2_000_000
CHUNK_SIZE = 20_000
TARGET = 1e-3
DELTA = 0.05
SEED = 20261114

def build_circuit(d, p):
    return stim.Circuit.generated(
        "surface_code:rotated_memory_x",
        distance=d,
        rounds=d,
        after_clifford_depolarization=p,
        before_round_data_depolarization=p,
        after_reset_flip_probability=p,
        before_measure_flip_probability=p
    )

def cp_upper(failures, collisions, delta):
    if collisions == 0:
        return 1.0
    if failures == collisions:
        return 1.0
    return float(beta.ppf(
        1-delta,
        failures+1,
        collisions-failures
    ))

# Bonferroni adjustment for four comparisons.
delta_each = DELTA / (len(DISTANCES)*len(ALPHAS))

rows = []

for d in DISTANCES:
    print(f"\nRunning distance {d}...", flush=True)

    circuit = build_circuit(d, P)
    decoder = pymatching.Matching.from_detector_error_model(
        circuit.detector_error_model(decompose_errors=True)
    )
    sampler = circuit.compile_detector_sampler(
        seed=SEED+d
    )

    stats = {
        a: {"groups":0, "collisions":0, "failures":0}
        for a in ALPHAS
    }

    baseline_failures = 0
    processed = 0

    while processed < SHOTS_PER_DISTANCE:
        n = min(
            CHUNK_SIZE,
            SHOTS_PER_DISTANCE-processed
        )

        det, obs = sampler.sample(
            shots=n,
            separate_observables=True
        )

        predicted = decoder.decode_batch(det)
        failed = np.any(predicted != obs, axis=1)
        baseline_failures += int(failed.sum())

        packed = np.packbits(
            det, axis=1, bitorder="little"
        )

        for alpha in ALPHAS:
            groups = n // alpha
            usable = groups * alpha

            grouped = packed[:usable].reshape(
                groups, alpha, -1
            )

            collision = np.all(
                grouped == grouped[:, :1, :],
                axis=(1, 2)
            )

            first_failed = failed[:usable:alpha]

            stats[alpha]["groups"] += groups
            stats[alpha]["collisions"] += int(
                collision.sum()
            )
            stats[alpha]["failures"] += int(
                np.count_nonzero(
                    collision & first_failed
                )
            )

        processed += n

    for alpha in ALPHAS:
        s = stats[alpha]
        K = s["collisions"]
        Y = s["failures"]

        upper = cp_upper(Y, K, delta_each)

        rows.append({
            "distance":d,
            "alpha":alpha,
            "shots":processed,
            "groups":s["groups"],
            "collisions":K,
            "collision_failures":Y,
            "baseline_LER":
                baseline_failures/processed,
            "conditional_point_estimate":
                Y/K if K else np.nan,
            "exact_upper_95_joint":upper,
            "certified_below_1e-3":
                upper < TARGET
        })

results = pd.DataFrame(rows)

print("\nEXACT CERTIFICATION RESULTS")
print(results.to_string(index=False))

results.to_csv(
    "exact_disjoint_collision_certification.csv",
    index=False
)

print("\nJoint confidence:", 1-DELTA)
print("Target:", TARGET)
print("Saved CSV.")
```

```python

import gzip, pickle, math
import numpy as np
import pandas as pd
from scipy.stats import beta, binom

PATH = "qec_resampling_validation/distance_5_counts.pkl.gz"
DELTA = 0.05
TARGET = 1e-3
KS = [1, 5, 20, 100]
ALPHAS = [2, 3]

with gzip.open(PATH, "rb") as f:
    data = pickle.load(f)

cal = data["calibration"]
ev = data["evaluation"]
fail = data["failures"]
N = sum(ev.values())

# Bonferroni correction over all reported bounds.
# Each (K, alpha) uses 3K+2 individual events.
num_events = len(ALPHAS) * sum(3*k+2 for k in KS)
eps = DELTA / num_events

def lower_cp(x, n):
    if x == 0:
        return 0.0
    return float(beta.ppf(eps, x, n-x+1))

def upper_cp(x, n):
    if n == 0 or x == n:
        return 1.0
    return float(beta.ppf(1-eps, x+1, n-x))

def max_tail_upper(k, n):
    """
    Distribution-free upper confidence bound on
    the maximum probability of an unselected label.

    For any q, at most 1/q labels have probability >= q.
    A union bound gives:
       Pr(max tail count <= k, max tail p >= q)
           <= BinomCDF(k;n,q)/q

    Invert this test at level eps.
    """
    def log_bound(q):
        if q <= 0:
            return np.inf
        return binom.logcdf(k, n, q) - math.log(q)

    lo = 0.0
    hi = 1.0

    for _ in range(65):
        mid = (lo+hi)/2
        if log_bound(mid) <= math.log(eps):
            hi = mid
        else:
            lo = mid

    return hi

ranking = [s for s, _ in cal.most_common()]
rows = []

for K in KS:
    selected = ranking[:K]
    selected_set = set(selected)

    tail_count = sum(
        c for s, c in ev.items()
        if s not in selected_set
    )

    max_tail_count = max(
        (c for s, c in ev.items()
         if s not in selected_set),
        default=0
    )

    tail_mass_upper = upper_cp(tail_count, N)
    max_tail_p_upper = max_tail_upper(
        max_tail_count, N
    )

    record_bounds = []

    for s in selected:
        m = ev.get(s, 0)
        e = fail.get(s, 0)

        record_bounds.append((
            lower_cp(m, N),
            upper_cp(m, N),
            upper_cp(e, m)
        ))

    for alpha in ALPHAS:
        denominator_lower = sum(
            p_lo**alpha
            for p_lo, p_hi, f_hi in record_bounds
        )

        selected_failure_upper = sum(
            p_hi**alpha * f_hi
            for p_lo, p_hi, f_hi in record_bounds
        )

        tail_power_upper = (
            tail_mass_upper *
            max_tail_p_upper**(alpha-1)
        )

        if denominator_lower > 0:
            bound = min(
                1.0,
                (selected_failure_upper +
                 tail_power_upper) /
                denominator_lower
            )
        else:
            bound = 1.0

        rows.append({
            "K":K,
            "alpha":alpha,
            "selected_eval_shots":N-tail_count,
            "max_tail_count":max_tail_count,
            "tail_mass_upper":tail_mass_upper,
            "max_tail_p_upper":max_tail_p_upper,
            "tail_power_upper":tail_power_upper,
            "selected_failure_upper":
                selected_failure_upper,
            "denominator_lower":
                denominator_lower,
            "rigorous_upper":bound,
            "certified":bound<TARGET
        })

results = pd.DataFrame(rows)

print("TAIL-AWARE CERTIFICATION")
print(results.to_string(index=False))

print("\nBest bounds:")
print(
    results.loc[
        results.groupby("alpha")[
            "rigorous_upper"
        ].idxmin()
    ][["alpha","K","rigorous_upper","certified"]]
    .to_string(index=False)
)

results.to_csv(
    "distance_5_tail_aware_certification.csv",
    index=False
)
```

```python

import gzip
import pickle
import math
import numpy as np
import pandas as pd
from collections import Counter
from scipy.stats import beta

PATH = "qec_resampling_validation/distance_5_counts.pkl.gz"
KS = [1, 5, 20, 100]
ALPHAS = [2, 3]
DELTA = 0.05
SEED = 20261114

with gzip.open(PATH, "rb") as f:
    data = pickle.load(f)

cal = data["calibration"]
ev = data["evaluation"]

N = sum(ev.values())
ranking = [s for s, _ in cal.most_common()]

# Reconstruct the unordered evaluation sample.
# Its order is randomized independently of its values.
rng = np.random.default_rng(SEED)
labels = list(ev.keys())
label_to_id = {s: i for i, s in enumerate(labels)}

ids = np.repeat(
    np.arange(len(labels), dtype=np.int32),
    [ev[s] for s in labels]
)
rng.shuffle(ids)

# Map calibration-selected labels to evaluation IDs.
selected_ids = {
    K: set(label_to_id[s] for s in ranking[:K]
           if s in label_to_id)
    for K in KS
}

# Simultaneous coverage across all K and alpha.
delta_each = DELTA / (len(KS)*len(ALPHAS))

def cp_upper(x, n, delta):
    if n == 0 or x == n:
        return 1.0
    return float(beta.ppf(
        1-delta, x+1, n-x
    ))

def falling(n, a):
    if n < a:
        return 0
    return math.prod(n-j for j in range(a))

rows = []

for K in KS:
    selected = set(ranking[:K])

    for alpha in ALPHAS:
        groups = N // alpha
        grouped = ids[:groups*alpha].reshape(
            groups, alpha
        )

        same = np.all(
            grouped == grouped[:, :1],
            axis=1
        )

        # A collision belongs to the tail if its
        # common syndrome was not selected.
        first_ids = grouped[:, 0]

        is_selected = np.isin(
            first_ids,
            list(selected_ids[K])
        )

        tail_collision = same & (~is_selected)
        X = int(tail_collision.sum())

        exact_upper = cp_upper(
            X, groups, delta_each
        )

        # Full overlapping U-statistic estimate.
        # This is an unbiased point estimate,
        # NOT a rigorous upper confidence bound.
        numerator = sum(
            math.comb(c, alpha)
            for s, c in ev.items()
            if s not in selected and c >= alpha
        )

        denominator = math.comb(N, alpha)
        U_hat = numerator / denominator

        rows.append({
            "K": K,
            "alpha": alpha,
            "groups": groups,
            "disjoint_tail_collisions": X,
            "overlapping_tail_C_hat": U_hat,
            "exact_tail_C_upper": exact_upper,
            "upper_to_point_ratio":
                exact_upper/U_hat
                if U_hat > 0 else np.inf
        })

results = pd.DataFrame(rows)

print("TAIL COLLISION MOMENT BENCHMARK")
print(results.to_string(index=False))

print("\nBEST EXACT TAIL UPPER BOUNDS")
print(
    results.loc[
        results.groupby("alpha")[
            "exact_tail_C_upper"
        ].idxmin()
    ].to_string(index=False)
)

results.to_csv(
    "distance_5_tail_collision_benchmark.csv",
    index=False
)

print("\nSimultaneous confidence level:", 1-DELTA)
print("Saved CSV.")
```

```python

import gzip
import pickle
from collections import defaultdict
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# FREQUENCY-STRATIFIED FINITE-SAMPLE CERTIFICATION
# ============================================================

PATH = "qec_resampling_validation/distance_5_counts.pkl.gz"
DELTA = 0.05
TARGET = 1e-3
ALPHAS = [2, 3]

# Only load pickle files you trust.
with gzip.open(PATH, "rb") as f:
    data = pickle.load(f)

cal = data["calibration"]
ev = data["evaluation"]
fail = data["failures"]

N = sum(ev.values())
assert N == 100_000
assert sum(fail.values()) == data["total_failures"]

# The dominant record and every grouping decision
# are fixed by the INDEPENDENT calibration data.
dominant = cal.most_common(1)[0][0]

GROUPS = [
    "dominant",
    "cal_20_plus",
    "cal_5_to_19",
    "cal_2_to_4",
    "cal_1",
    "cal_unseen"
]

def group_of(s):
    if s == dominant:
        return "dominant"
    c = cal.get(s, 0)
    if c >= 20:
        return "cal_20_plus"
    if c >= 5:
        return "cal_5_to_19"
    if c >= 2:
        return "cal_2_to_4"
    if c == 1:
        return "cal_1"
    return "cal_unseen"

stats = {
    g: {
        "shots": 0,
        "max_count": 0,
        "failures": 0,
        "records_seen": 0
    }
    for g in GROUPS
}

for s, count in ev.items():
    g = group_of(s)
    stats[g]["shots"] += count
    stats[g]["max_count"] = max(
        stats[g]["max_count"], count
    )
    stats[g]["failures"] += fail.get(s, 0)
    stats[g]["records_seen"] += 1

assert sum(x["shots"] for x in stats.values()) == N
assert sum(x["failures"] for x in stats.values()) == sum(fail.values())

# For six groups:
# 6 maximum-atom upper bounds
# 6 joint-failure upper bounds
# 1 dominant-atom lower bound
num_events = 2 * len(GROUPS) + 1
eps = DELTA / num_events

def cp_upper(k, n, delta):
    if k == n:
        return 1.0
    return float(beta.ppf(
        1.0-delta, k+1, n-k
    ))

def cp_lower(k, n, delta):
    if k == 0:
        return 0.0
    return float(beta.ppf(
        delta, k, n-k+1
    ))

dominant_count = ev.get(dominant, 0)
dominant_p_lower = cp_lower(
    dominant_count, N, eps
)

rows = []

for g in GROUPS:
    x = stats[g]

    # Bound the largest individual syndrome
    # probability in this calibration-fixed group.
    max_p_upper = cp_upper(
        x["max_count"], N, eps
    )

    # Bound P(logical_failure AND group).
    failure_mass_upper = cp_upper(
        x["failures"], N, eps
    )

    rows.append({
        "group": g,
        **x,
        "max_p_upper": max_p_upper,
        "failure_mass_upper": failure_mass_upper
    })

groups_df = pd.DataFrame(rows)

results = []

for alpha in ALPHAS:
    numerator_upper = float(np.sum(
        groups_df["max_p_upper"].to_numpy()
        ** (alpha-1)
        * groups_df["failure_mass_upper"].to_numpy()
    ))

    denominator_lower = dominant_p_lower ** alpha

    upper = min(
        1.0,
        numerator_upper / denominator_lower
    ) if denominator_lower > 0 else 1.0

    results.append({
        "distance": 5,
        "alpha": alpha,
        "numerator_upper": numerator_upper,
        "denominator_lower": denominator_lower,
        "confidence_upper": upper,
        "target": TARGET,
        "certified": upper < TARGET
    })

results_df = pd.DataFrame(results)

print("GROUPWISE EVALUATION")
print(groups_df.to_string(index=False))

print("\nSIMULTANEOUS CERTIFICATION")
print(results_df.to_string(index=False))

print("\nConfidence coverage: at least", 1-DELTA)
print("Dominant P lower:", dominant_p_lower)
print("Bonferroni events:", num_events)
print("Per-event delta:", eps)

groups_df.to_csv(
    "distance5_groupwise_bounds.csv",
    index=False
)
results_df.to_csv(
    "distance5_stratified_certification.csv",
    index=False
)
```

```python
import gzip
import pickle
from collections import Counter
import numpy as np
import pandas as pd
import stim
import pymatching
from scipy.stats import beta

# ==========================================================
# FRESH, FIXED-GROUP FINITE-SAMPLE VALIDATION
# ==========================================================

PATH = "qec_resampling_validation/distance_5_counts.pkl.gz"
DISTANCE = 5
P = 0.005
SHOTS = 1_000_000
CHUNK = 20_000
SEED = 20261115
DELTA = 0.05
TARGET = 1e-3
ALPHAS = [2, 3]

# Independent calibration from the earlier experiment.
with gzip.open(PATH, "rb") as f:
    archive = pickle.load(f)

cal = archive["calibration"]
dominant = cal.most_common(1)[0][0]

GROUPS = [
    "dominant", "cal_20_plus", "cal_5_to_19",
    "cal_2_to_4", "cal_1", "cal_unseen"
]

def group_of(s):
    if s == dominant:
        return "dominant"
    n = cal.get(s, 0)
    if n >= 20:
        return "cal_20_plus"
    if n >= 5:
        return "cal_5_to_19"
    if n >= 2:
        return "cal_2_to_4"
    if n == 1:
        return "cal_1"
    return "cal_unseen"

def cp_upper(k, n, delta):
    if k == n:
        return 1.0
    return float(beta.ppf(1-delta, k+1, n-k))

def cp_lower(k, n, delta):
    if k == 0:
        return 0.0
    return float(beta.ppf(delta, k, n-k+1))

# The grouping rule and confidence allocation are fixed
# before any new observations are generated.
eps = DELTA / (2 * len(GROUPS) + 1)

circuit = stim.Circuit.generated(
    "surface_code:rotated_memory_x",
    distance=DISTANCE,
    rounds=DISTANCE,
    after_clifford_depolarization=P,
    before_round_data_depolarization=P,
    after_reset_flip_probability=P,
    before_measure_flip_probability=P
)

decoder = pymatching.Matching.from_detector_error_model(
    circuit.detector_error_model(decompose_errors=True)
)
sampler = circuit.compile_detector_sampler(seed=SEED)

counts = Counter()
failures = Counter()
total_failed = 0
processed = 0

while processed < SHOTS:
    n = min(CHUNK, SHOTS - processed)

    det, obs = sampler.sample(
        shots=n, separate_observables=True
    )
    prediction = decoder.decode_batch(det)
    failed = np.any(prediction != obs, axis=1)

    packed = np.packbits(det, axis=1, bitorder="little")

    counts.update(map(bytes, packed))
    failures.update(map(bytes, packed[failed]))

    total_failed += int(failed.sum())
    processed += n

    if processed % 200_000 == 0:
        print(f"Simulated {processed:,} shots")

stats = {
    g: {"shots":0, "max_count":0, "failures":0}
    for g in GROUPS
}

for s, c in counts.items():
    g = group_of(s)
    stats[g]["shots"] += c
    stats[g]["max_count"] = max(
        stats[g]["max_count"], c
    )
    stats[g]["failures"] += failures.get(s, 0)

assert sum(x["shots"] for x in stats.values()) == SHOTS
assert sum(x["failures"] for x in stats.values()) == total_failed

p0_lower = cp_lower(
    counts.get(dominant, 0), SHOTS, eps
)

group_rows = []
for g in GROUPS:
    x = stats[g]
    group_rows.append({
        "group": g,
        **x,
        "max_p_upper": cp_upper(
            x["max_count"], SHOTS, eps
        ),
        "failure_mass_upper": cp_upper(
            x["failures"], SHOTS, eps
        )
    })

group_df = pd.DataFrame(group_rows)
summary = []

for alpha in ALPHAS:
    numerator_upper = float(np.sum(
        group_df["max_p_upper"].to_numpy()**(alpha-1)
        * group_df["failure_mass_upper"].to_numpy()
    ))

    denominator_lower = p0_lower**alpha

    upper = min(
        1.0, numerator_upper / denominator_lower
    ) if denominator_lower > 0 else 1.0

    summary.append({
        "distance": DISTANCE,
        "alpha": alpha,
        "shots": SHOTS,
        "baseline_LER": total_failed / SHOTS,
        "dominant_count": counts.get(dominant, 0),
        "dominant_failures": failures.get(dominant, 0),
        "numerator_upper": numerator_upper,
        "denominator_lower": denominator_lower,
        "confidence_upper": upper,
        "certified": upper < TARGET
    })

results = pd.DataFrame(summary)

print("\nFRESH GROUPWISE EVALUATION")
print(group_df.to_string(index=False))

print("\nFRESH VALIDATION RESULTS")
print(results.to_string(index=False))

print("\nSimultaneous confidence: at least 95%")
print("Target:", TARGET)

group_df.to_csv(
    "distance5_fresh_groupwise.csv", index=False
)
results.to_csv(
    "distance5_fresh_certification.csv", index=False
)

# Save sufficient statistics to audit the run.
with gzip.open("distance5_fresh_counts.pkl.gz", "wb") as f:
    pickle.dump({
        "counts": counts,
        "failures": failures,
        "shots": SHOTS,
        "seed": SEED,
        "p": P,
        "distance": DISTANCE,
        "groups": GROUPS
    }, f)
```

```python

import os, json, gzip, pickle, hashlib
from collections import Counter
import numpy as np
import pandas as pd
import stim
import pymatching
from scipy.stats import beta

# ============================================================
# DISTANCE-5: INDEPENDENT CALIBRATION + VALIDATION
# Primary objective: alpha=2, target < 1e-3
# ============================================================

D = 5
P = 0.005
CAL_SHOTS = 1_000_000
EVAL_SHOTS = 1_000_000
CHUNK = 20_000

CAL_SEED = 20261120
EVAL_SEED = 20261121

DELTA = 0.05
TARGET = 1e-3
ALPHAS = [2, 3]
OUT = "qec_d5_alpha2_validation"

os.makedirs(OUT, exist_ok=True)

# All analysis choices fixed before evaluation.
GROUPS = [
    "dominant",
    "cal_100_plus",
    "cal_20_to_99",
    "cal_5_to_19",
    "cal_2_to_4",
    "cal_1",
    "cal_unseen"
]

EPS = DELTA / (2 * len(GROUPS) + 1)

def cp_upper(k, n, delta):
    if k == n:
        return 1.0
    return float(beta.ppf(1-delta, k+1, n-k))

def cp_lower(k, n, delta):
    if k == 0:
        return 0.0
    return float(beta.ppf(delta, k, n-k+1))

circuit = stim.Circuit.generated(
    "surface_code:rotated_memory_x",
    distance=D,
    rounds=D,
    after_clifford_depolarization=P,
    before_round_data_depolarization=P,
    after_reset_flip_probability=P,
    before_measure_flip_probability=P
)

decoder = pymatching.Matching.from_detector_error_model(
    circuit.detector_error_model(decompose_errors=True)
)

def collect(shots, seed, name):
    sampler = circuit.compile_detector_sampler(seed=seed)

    counts = Counter()
    failures = Counter()
    total_failures = 0

    for start in range(0, shots, CHUNK):
        n = min(CHUNK, shots-start)

        det, obs = sampler.sample(
            shots=n, separate_observables=True
        )

        pred = decoder.decode_batch(det)
        bad = np.any(pred != obs, axis=1)

        packed = np.packbits(
            det, axis=1, bitorder="little"
        )

        counts.update(map(bytes, packed))
        failures.update(map(bytes, packed[bad]))
        total_failures += int(bad.sum())

        done = start+n
        if done % 200_000 == 0 or done == shots:
            print(f"{name}: {done:,}/{shots:,}")

    assert sum(counts.values()) == shots
    assert sum(failures.values()) == total_failures

    return counts, failures, total_failures

# ============================================================
# STAGE 1: INDEPENDENT CALIBRATION
# ============================================================

cal, cal_fail, cal_total_fail = collect(
    CAL_SHOTS, CAL_SEED, "Calibration"
)

dominant = cal.most_common(1)[0][0]

def group_of(s):
    if s == dominant:
        return "dominant"

    n = cal.get(s, 0)

    if n >= 100:
        return "cal_100_plus"
    if n >= 20:
        return "cal_20_to_99"
    if n >= 5:
        return "cal_5_to_19"
    if n >= 2:
        return "cal_2_to_4"
    if n == 1:
        return "cal_1"
    return "cal_unseen"

# ============================================================
# STAGE 2: INDEPENDENT EVALUATION
# ============================================================

ev, fail, total_failed = collect(
    EVAL_SHOTS, EVAL_SEED, "Evaluation"
)

stats = {
    g: {
        "shots": 0,
        "max_count": 0,
        "failures": 0,
        "records_seen": 0
    }
    for g in GROUPS
}

for s, c in ev.items():
    g = group_of(s)
    x = stats[g]

    x["shots"] += c
    x["max_count"] = max(x["max_count"], c)
    x["failures"] += fail.get(s, 0)
    x["records_seen"] += 1

assert sum(x["shots"] for x in stats.values()) == EVAL_SHOTS
assert sum(x["failures"] for x in stats.values()) == total_failed

# ============================================================
# STAGE 3: SIMULTANEOUS CONFIDENCE BOUNDS
# ============================================================

p0_lower = cp_lower(
    ev.get(dominant, 0), EVAL_SHOTS, EPS
)

rows = []

for g in GROUPS:
    x = stats[g]

    rows.append({
        "group": g,
        **x,
        "max_p_upper": cp_upper(
            x["max_count"], EVAL_SHOTS, EPS
        ),
        "failure_mass_upper": cp_upper(
            x["failures"], EVAL_SHOTS, EPS
        )
    })

group_df = pd.DataFrame(rows)

summary = []

for alpha in ALPHAS:
    terms = (
        group_df["max_p_upper"].to_numpy()
        ** (alpha-1)
        * group_df["failure_mass_upper"].to_numpy()
    )

    numerator_upper = float(terms.sum())
    denominator_lower = p0_lower ** alpha

    upper = (
        min(1.0, numerator_upper / denominator_lower)
        if denominator_lower > 0 else 1.0
    )

    summary.append({
        "distance": D,
        "alpha": alpha,
        "calibration_shots": CAL_SHOTS,
        "evaluation_shots": EVAL_SHOTS,
        "baseline_LER": total_failed/EVAL_SHOTS,
        "dominant_count": ev.get(dominant, 0),
        "dominant_failures": fail.get(dominant, 0),
        "numerator_upper": numerator_upper,
        "denominator_lower": denominator_lower,
        "confidence_upper": upper,
        "certified": bool(upper < TARGET)
    })

    group_df[f"F{alpha}_upper_contribution"] = terms

result_df = pd.DataFrame(summary)

print("\nGROUPWISE EVALUATION")
print(group_df.to_string(index=False))

print("\nCERTIFICATION RESULTS")
print(result_df.to_string(index=False))

print("\nSimultaneous coverage: at least 95%")
print("Target:", TARGET)
print("Bonferroni events:", 2*len(GROUPS)+1)
print("Per-event delta:", EPS)

# ============================================================
# STAGE 4: SAVE AUDITABLE EXPERIMENT
# ============================================================

group_df.to_csv(
    f"{OUT}/groupwise_bounds.csv", index=False
)
result_df.to_csv(
    f"{OUT}/certification_results.csv", index=False
)

with gzip.open(f"{OUT}/counts.pkl.gz", "wb") as f:
    pickle.dump({
        "calibration": cal,
        "calibration_failures": cal_fail,
        "evaluation": ev,
        "evaluation_failures": fail,
        "calibration_total_failures": cal_total_fail,
        "evaluation_total_failures": total_failed,
        "dominant": dominant
    }, f, protocol=pickle.HIGHEST_PROTOCOL)

metadata = {
    "distance": D,
    "physical_error_probability": P,
    "rounds": D,
    "noise_channels": "all four channels set to P",
    "calibration_shots": CAL_SHOTS,
    "evaluation_shots": EVAL_SHOTS,
    "calibration_seed": CAL_SEED,
    "evaluation_seed": EVAL_SEED,
    "chunk_size": CHUNK,
    "alpha_values": ALPHAS,
    "delta": DELTA,
    "target": TARGET,
    "group_definitions": GROUPS,
    "dominant_record_hex": dominant.hex(),
    "stim_version": getattr(stim, "__version__", "unknown"),
    "pymatching_version": getattr(
        pymatching, "__version__", "unknown"
    ),
    "circuit_sha256": hashlib.sha256(
        str(circuit).encode()
    ).hexdigest()
}

with open(f"{OUT}/metadata.json", "w") as f:
    json.dump(metadata, f, indent=2)

print("\nExperiment saved in:", OUT)
```

```python

import gzip
import pickle
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# DISTANCE-5 ALPHA-2: SYNDROME-WEIGHT STRATIFICATION PILOT
#
# Exploratory analysis of an existing evaluation dataset.
# NOT a new prospective confidence certificate.
# ============================================================

PATH = "qec_d5_alpha2_validation/counts.pkl.gz"
DELTA = 0.05
TARGET = 1e-3
ALPHAS = [2, 3]

with gzip.open(PATH, "rb") as f:
    data = pickle.load(f)

cal = data["calibration"]
ev = data["evaluation"]
fail = data["evaluation_failures"]

N = sum(ev.values())
dominant = data["dominant"]

assert N == 1_000_000
assert sum(fail.values()) == data["evaluation_total_failures"]

# Fixed, physically interpretable Hamming-weight bands.
# Detector records contain 120 bits for this d=5 circuit.
WEIGHT_EDGES = [0, 3, 5, 7, 9, 13, 21, 121]

def weight_band(s):
    weight = int.from_bytes(s, "little").bit_count()

    for i in range(len(WEIGHT_EDGES)-1):
        lo = WEIGHT_EDGES[i]
        hi = WEIGHT_EDGES[i+1]

        if lo <= weight < hi:
            return f"unseen_w{lo}_to_{hi-1}"

    raise ValueError("Unexpected detector weight")

# Preserve the existing frequency groups.
BASE_GROUPS = [
    "dominant",
    "cal_100_plus",
    "cal_20_to_99",
    "cal_5_to_19",
    "cal_2_to_4",
    "cal_1"
]

WEIGHT_GROUPS = [
    f"unseen_w{WEIGHT_EDGES[i]}_to_{WEIGHT_EDGES[i+1]-1}"
    for i in range(len(WEIGHT_EDGES)-1)
]

GROUPS = BASE_GROUPS + WEIGHT_GROUPS

def group_of(s):
    if s == dominant:
        return "dominant"

    c = cal.get(s, 0)

    if c >= 100:
        return "cal_100_plus"
    if c >= 20:
        return "cal_20_to_99"
    if c >= 5:
        return "cal_5_to_19"
    if c >= 2:
        return "cal_2_to_4"
    if c == 1:
        return "cal_1"

    return weight_band(s)

stats = {
    g: {
        "shots": 0,
        "failures": 0,
        "max_count": 0,
        "records_seen": 0
    }
    for g in GROUPS
}

for s, c in ev.items():
    g = group_of(s)
    x = stats[g]

    x["shots"] += c
    x["failures"] += fail.get(s, 0)
    x["max_count"] = max(x["max_count"], c)
    x["records_seen"] += 1

assert sum(x["shots"] for x in stats.values()) == N
assert sum(x["failures"] for x in stats.values()) == sum(fail.values())

def cp_upper(k, n, eps):
    if k == n:
        return 1.0
    return float(beta.ppf(1-eps, k+1, n-k))

def cp_lower(k, n, eps):
    if k == 0:
        return 0.0
    return float(beta.ppf(eps, k, n-k+1))

# Simultaneous bounds for the FIXED grouping.
# Interpretation here is exploratory because this
# evaluation dataset has previously been analyzed.
num_events = 2 * len(GROUPS) + 1
eps = DELTA / num_events

p0_lower = cp_lower(ev.get(dominant, 0), N, eps)

rows = []
for g in GROUPS:
    x = stats[g]

    rows.append({
        "group": g,
        **x,
        "observed_failure_rate":
            x["failures"] / x["shots"]
            if x["shots"] else np.nan,
        "max_p_upper": cp_upper(
            x["max_count"], N, eps
        ),
        "failure_mass_upper": cp_upper(
            x["failures"], N, eps
        )
    })

groups_df = pd.DataFrame(rows)

summary = []

for alpha in ALPHAS:
    contribution = (
        groups_df["max_p_upper"].to_numpy()
        ** (alpha-1)
        * groups_df["failure_mass_upper"].to_numpy()
    )

    groups_df[f"F{alpha}_upper_contribution"] = contribution

    numerator = float(contribution.sum())
    denominator = p0_lower ** alpha

    upper = (
        min(1.0, numerator / denominator)
        if denominator > 0 else 1.0
    )

    summary.append({
        "alpha": alpha,
        "numerator_upper": numerator,
        "denominator_lower": denominator,
        "exploratory_upper": upper,
        "target": TARGET,
        "below_target": upper < TARGET
    })

results_df = pd.DataFrame(summary)

print("\nWEIGHT-STRATIFIED GROUP ANALYSIS")
print(groups_df.to_string(index=False))

print("\nEXPLORATORY BOUND COMPARISON")
print(results_df.to_string(index=False))

print("\nPREVIOUS REPORTED UPPER BOUNDS")
print("alpha=2: 0.002674")
print("alpha=3: 0.000473")

print("\nALPHA-2 CONTRIBUTION SHARE")
total_F2 = groups_df["F2_upper_contribution"].sum()

comparison = groups_df[
    ["group", "F2_upper_contribution"]
].copy()

comparison["percentage"] = (
    100 * comparison["F2_upper_contribution"] / total_F2
)

print(comparison.to_string(index=False))

groups_df.to_csv(
    "distance5_weight_stratification_pilot.csv",
    index=False
)
results_df.to_csv(
    "distance5_weight_stratification_summary.csv",
    index=False
)

print("\nExploratory pilot complete.")
print("Do not present post-hoc results as fresh certification.")
```

```python

import os, gzip, pickle, json, hashlib
from collections import Counter
import numpy as np
import pandas as pd
import stim, pymatching
from scipy.stats import beta

# ==========================================================
# PROSPECTIVE WEIGHT-STRATIFIED CERTIFICATION
# Distance 5 | Primary target: alpha 2 < 0.001
# ==========================================================

CAL_PATH = "qec_d5_alpha2_validation/counts.pkl.gz"
OUT = "qec_d5_weight_prospective"

D, P = 5, 0.005
SHOTS, CHUNK = 2_000_000, 20_000
SEED = 20261125
DELTA, TARGET = 0.05, 1e-3
ALPHAS = [2, 3]

WEIGHT_EDGES = [0, 3, 5, 7, 9, 13, 21, 121]

BASE = [
    "dominant", "cal_100_plus", "cal_20_to_99",
    "cal_5_to_19", "cal_2_to_4", "cal_1"
]

WEIGHT_GROUPS = [
    f"unseen_w{WEIGHT_EDGES[i]}_to_{WEIGHT_EDGES[i+1]-1}"
    for i in range(len(WEIGHT_EDGES)-1)
]

GROUPS = BASE + WEIGHT_GROUPS
EPS = DELTA / (2*len(GROUPS)+1)

os.makedirs(OUT, exist_ok=True)

with gzip.open(CAL_PATH, "rb") as f:
    archive = pickle.load(f)

cal = archive["calibration"]
dominant = archive["dominant"]

assert sum(cal.values()) == 1_000_000
assert len(GROUPS) == 13

def group_of(s):
    if s == dominant:
        return "dominant"

    c = cal.get(s, 0)
    if c >= 100: return "cal_100_plus"
    if c >= 20: return "cal_20_to_99"
    if c >= 5: return "cal_5_to_19"
    if c >= 2: return "cal_2_to_4"
    if c == 1: return "cal_1"

    w = int.from_bytes(s, "little").bit_count()
    for i in range(len(WEIGHT_EDGES)-1):
        if WEIGHT_EDGES[i] <= w < WEIGHT_EDGES[i+1]:
            return WEIGHT_GROUPS[i]

    raise ValueError(f"Unexpected detector weight {w}")

def cp_upper(k, n, eps):
    if k == n: return 1.0
    return float(beta.ppf(1-eps, k+1, n-k))

def cp_lower(k, n, eps):
    if k == 0: return 0.0
    return float(beta.ppf(eps, k, n-k+1))

# Freeze metadata before sampling.
configuration = {
    "distance": D, "p": P, "shots": SHOTS,
    "seed": SEED, "chunk": CHUNK,
    "delta": DELTA, "target": TARGET,
    "alpha": ALPHAS, "groups": GROUPS,
    "weight_edges": WEIGHT_EDGES,
    "dominant_hex": dominant.hex(),
    "calibration_shots": sum(cal.values()),
    "bonferroni_events": 2*len(GROUPS)+1
}

with open(f"{OUT}/frozen_config.json", "w") as f:
    json.dump(configuration, f, indent=2)

circuit = stim.Circuit.generated(
    "surface_code:rotated_memory_x",
    distance=D, rounds=D,
    after_clifford_depolarization=P,
    before_round_data_depolarization=P,
    after_reset_flip_probability=P,
    before_measure_flip_probability=P
)

configuration["circuit_sha256"] = hashlib.sha256(
    str(circuit).encode()
).hexdigest()
configuration["stim_version"] = stim.__version__
configuration["pymatching_version"] = getattr(
    pymatching, "__version__", "unknown"
)

with open(f"{OUT}/frozen_config.json", "w") as f:
    json.dump(configuration, f, indent=2)

decoder = pymatching.Matching.from_detector_error_model(
    circuit.detector_error_model(decompose_errors=True)
)
sampler = circuit.compile_detector_sampler(seed=SEED)

# Per-record counts are needed for group maximum counts.
counts = Counter()
failures = Counter()
total_failed = 0

for start in range(0, SHOTS, CHUNK):
    n = min(CHUNK, SHOTS-start)

    det, obs = sampler.sample(
        shots=n, separate_observables=True
    )

    predicted = decoder.decode_batch(det)
    bad = np.any(predicted != obs, axis=1)

    packed = np.packbits(
        det, axis=1, bitorder="little"
    )

    counts.update(map(bytes, packed))
    failures.update(map(bytes, packed[bad]))
    total_failed += int(bad.sum())

    done = start+n
    if done % 200_000 == 0 or done == SHOTS:
        print(f"Evaluation: {done:,}/{SHOTS:,}", flush=True)

assert sum(counts.values()) == SHOTS
assert sum(failures.values()) == total_failed

stats = {
    g: {
        "shots": 0,
        "max_count": 0,
        "failures": 0,
        "records_seen": 0
    } for g in GROUPS
}

for s, c in counts.items():
    x = stats[group_of(s)]
    x["shots"] += c
    x["max_count"] = max(x["max_count"], c)
    x["failures"] += failures.get(s, 0)
    x["records_seen"] += 1

assert sum(x["shots"] for x in stats.values()) == SHOTS
assert sum(x["failures"] for x in stats.values()) == total_failed

p0_lower = cp_lower(
    counts.get(dominant, 0), SHOTS, EPS
)

rows = []
for g in GROUPS:
    x = stats[g]
    rows.append({
        "group": g,
        **x,
        "max_p_upper": cp_upper(
            x["max_count"], SHOTS, EPS
        ),
        "failure_mass_upper": cp_upper(
            x["failures"], SHOTS, EPS
        )
    })

group_df = pd.DataFrame(rows)
summary = []

for a in ALPHAS:
    contributions = (
        group_df["max_p_upper"].to_numpy()**(a-1)
        * group_df["failure_mass_upper"].to_numpy()
    )

    group_df[f"F{a}_upper_contribution"] = contributions

    numerator_upper = float(contributions.sum())
    denominator_lower = p0_lower**a
    bound = (
        min(1.0, numerator_upper/denominator_lower)
        if denominator_lower > 0 else 1.0
    )

    summary.append({
        "distance": D,
        "alpha": a,
        "evaluation_shots": SHOTS,
        "baseline_LER": total_failed/SHOTS,
        "dominant_count": counts.get(dominant, 0),
        "dominant_failures": failures.get(dominant, 0),
        "numerator_upper": numerator_upper,
        "denominator_lower": denominator_lower,
        "confidence_upper": bound,
        "certified": bool(bound < TARGET)
    })

result_df = pd.DataFrame(summary)

print("\nFRESH WEIGHT-STRATIFIED GROUPS")
print(group_df.to_string(index=False))

print("\nPROSPECTIVE CERTIFICATION")
print(result_df.to_string(index=False))

print("\nJoint confidence: at least 95%")
print("Target:", TARGET)
print("Individual confidence allocation:", EPS)

group_df.to_csv(f"{OUT}/groups.csv", index=False)
result_df.to_csv(f"{OUT}/results.csv", index=False)

with gzip.open(f"{OUT}/evaluation_counts.pkl.gz", "wb") as f:
    pickle.dump({
        "counts": counts,
        "failures": failures,
        "total_failures": total_failed,
        "shots": SHOTS,
        "seed": SEED
    }, f, protocol=pickle.HIGHEST_PROTOCOL)

print("\nSaved experiment to", OUT)
```

```python

import os, json, gzip, pickle, hashlib
from collections import Counter
import numpy as np
import pandas as pd
from scipy.stats import beta
import stim
import pymatching

# ============================================================
# INDEPENDENT AUDIT: DISTANCE-5 PROSPECTIVE CERTIFICATES
# No new simulations
# ============================================================

ROOT = "qec_d5_weight_prospective"
CAL_PATH = "qec_d5_alpha2_validation/counts.pkl.gz"

# Only load pickle archives you trust.
with open(f"{ROOT}/frozen_config.json") as f:
    cfg = json.load(f)

with gzip.open(f"{ROOT}/evaluation_counts.pkl.gz", "rb") as f:
    evaluation = pickle.load(f)

with gzip.open(CAL_PATH, "rb") as f:
    calibration = pickle.load(f)

with open(f"{ROOT}/results.csv") as f:
    published = pd.read_csv(f)

cal = calibration["calibration"]
dominant = calibration["dominant"]

counts = evaluation["counts"]
failures = evaluation["failures"]

N = int(cfg["shots"])
delta = float(cfg["delta"])
target = float(cfg["target"])
alphas = [int(a) for a in cfg["alpha"]]
edges = cfg["weight_edges"]
groups = cfg["groups"]

checks = {}

def check(name, condition):
    checks[name] = bool(condition)
    if not condition:
        raise AssertionError(f"FAILED: {name}")

# ------------------------------------------------------------
# 1. Data integrity
# ------------------------------------------------------------

check("Evaluation shot count", sum(counts.values()) == N)
check("Calibration shot count", sum(cal.values()) == 1_000_000)
check("Stored shot count", evaluation["shots"] == N)
check("Evaluation seed", evaluation["seed"] == cfg["seed"])
check("Distinct experiment seeds",
      cfg["seed"] not in (20261120, 20261121))
check("Dominant calibration record",
      dominant == cal.most_common(1)[0][0])
check("Dominant record metadata",
      dominant.hex() == cfg["dominant_hex"])

check("Failure counts valid", all(
    0 <= failures.get(s, 0) <= c
    for s, c in counts.items()
))

check("No unexpected failure labels",
      all(s in counts for s in failures))

check("Failure total", sum(failures.values()) ==
      evaluation["total_failures"])

# ------------------------------------------------------------
# 2. Independently reconstruct the frozen group membership
# ------------------------------------------------------------

def assign_group(s):
    if s == dominant:
        return "dominant"

    c = cal.get(s, 0)

    if c >= 100: return "cal_100_plus"
    if c >= 20: return "cal_20_to_99"
    if c >= 5: return "cal_5_to_19"
    if c >= 2: return "cal_2_to_4"
    if c == 1: return "cal_1"

    weight = int.from_bytes(s, "little").bit_count()

    for i in range(len(edges)-1):
        if edges[i] <= weight < edges[i+1]:
            return f"unseen_w{edges[i]}_to_{edges[i+1]-1}"

    raise AssertionError("Unclassified record")

check("Group count", len(groups) == 13)
check("Group labels unique", len(set(groups)) == len(groups))
check("Detector record size", all(
    len(s) == 15 for s in counts
))

audit = {
    g: {"shots":0, "failures":0, "max_count":0,
        "records_seen":0}
    for g in groups
}

for s, c in counts.items():
    g = assign_group(s)
    check("Recognized group", g in audit)

    audit[g]["shots"] += c
    audit[g]["failures"] += failures.get(s, 0)
    audit[g]["max_count"] = max(
        audit[g]["max_count"], c
    )
    audit[g]["records_seen"] += 1

check("Group shot partition",
      sum(x["shots"] for x in audit.values()) == N)

check("Group failure partition",
      sum(x["failures"] for x in audit.values())
      == evaluation["total_failures"])

# ------------------------------------------------------------
# 3. Independent confidence-bound reconstruction
# ------------------------------------------------------------

eps = delta/(2*len(groups)+1)

def upper(k, n):
    if k == n:
        return 1.0
    return float(beta.isf(eps, k+1, n-k))

def lower(k, n):
    if k == 0:
        return 0.0
    return float(beta.ppf(eps, k, n-k+1))

p0_lower = lower(counts.get(dominant,0), N)

rows = []

for g in groups:
    x = audit[g]
    rows.append({
        "group":g,
        **x,
        "max_p_upper": upper(x["max_count"],N),
        "failure_mass_upper": upper(x["failures"],N)
    })

group_df = pd.DataFrame(rows)

recomputed = []

for alpha in alphas:
    contributions = (
        group_df["max_p_upper"].to_numpy() ** (alpha-1)
        * group_df["failure_mass_upper"].to_numpy()
    )

    group_df[f"F{alpha}_upper_contribution"] = contributions

    numerator = float(np.sum(contributions))
    denominator = float(p0_lower ** alpha)
    bound = min(1.0,numerator/denominator) \
            if denominator > 0 else 1.0

    recomputed.append({
        "alpha": alpha,
        "numerator_upper": numerator,
        "denominator_lower": denominator,
        "confidence_upper": bound,
        "certified": bound < target
    })

result_df = pd.DataFrame(recomputed)

# ------------------------------------------------------------
# 4. Compare with the saved result file
# ------------------------------------------------------------

for _, row in result_df.iterrows():
    old = published.loc[
        published["alpha"] == row["alpha"]
    ].iloc[0]

    for column in [
        "numerator_upper",
        "denominator_lower",
        "confidence_upper"
    ]:
        check(
            f"alpha={int(row['alpha'])} {column}",
            np.isclose(
                row[column], old[column],
                rtol=1e-9, atol=1e-14
            )
        )

    check(
        f"alpha={int(row['alpha'])} certification",
        bool(row["certified"]) == bool(old["certified"])
    )

# ------------------------------------------------------------
# 5. Reconstruct the circuit and verify its fingerprint
# ------------------------------------------------------------

d = int(cfg["distance"])
p = float(cfg["p"])

circuit = stim.Circuit.generated(
    "surface_code:rotated_memory_x",
    distance=d, rounds=d,
    after_clifford_depolarization=p,
    before_round_data_depolarization=p,
    after_reset_flip_probability=p,
    before_measure_flip_probability=p
)

digest = hashlib.sha256(
    str(circuit).encode()
).hexdigest()

check("Circuit fingerprint",
      digest == cfg["circuit_sha256"])

# ------------------------------------------------------------
# 6. Report and save independent audit
# ------------------------------------------------------------

print("\nINDEPENDENT AUDIT RESULTS")
print(result_df.to_string(index=False))

print("\nDATA AND IMPLEMENTATION CHECKS")
print(f"Checks passed: {sum(checks.values())}/{len(checks)}")
print(f"Evaluation shots: {N:,}")
print(f"Evaluation failures: {sum(failures.values()):,}")
print(f"Dominant-record hits: {counts.get(dominant,0):,}")
print(f"Bonferroni events: {2*len(groups)+1}")
print(f"Individual error allocation: {eps:.12f}")

print("\nSHA256 FILE MANIFEST")
manifest = {}

for path in [
    CAL_PATH,
    f"{ROOT}/frozen_config.json",
    f"{ROOT}/evaluation_counts.pkl.gz",
    f"{ROOT}/results.csv",
    f"{ROOT}/groups.csv"
]:
    with open(path, "rb") as f:
        manifest[path] = hashlib.sha256(f.read()).hexdigest()

for path, digest in manifest.items():
    print(digest, path)

result_df.to_csv(
    f"{ROOT}/independent_audit_results.csv",
    index=False
)

group_df.to_csv(
    f"{ROOT}/independent_audit_groups.csv",
    index=False
)

with open(f"{ROOT}/audit_manifest.json", "w") as f:
    json.dump({
        "file_sha256": manifest,
        "checks": checks,
        "stim_version": stim.__version__,
        "pymatching_version": getattr(
            pymatching,"__version__","unknown"
        )
    }, f, indent=2)

print("\nAUDIT PASSED")
print("Existing certificates independently reproduced.")
```

```python

import os, json, gzip, pickle, hashlib, gc
from collections import Counter
import numpy as np
import pandas as pd
import stim
import pymatching
from scipy.stats import beta

# ============================================================
# QEC FINITE-SAMPLE CERTIFICATION: NOISE GENERALIZATION
#
# Three independent calibration/evaluation experiments.
# Fixed sample budgets and familywise confidence allocation.
# ============================================================

DISTANCE = 5
P_VALUES = [0.003, 0.007, 0.010]
ALPHAS = [2, 3]

CAL_SHOTS = 1_000_000
EVAL_SHOTS = 1_000_000
CHUNK = 20_000

DELTA = 0.05
TARGET = 1e-3

OUT = "qec_noise_generalization_v1"
os.makedirs(OUT, exist_ok=True)

WEIGHT_EDGES = [0, 3, 5, 7, 9, 13, 21, 121]

BASE_GROUPS = [
    "dominant",
    "cal_100_plus",
    "cal_20_to_99",
    "cal_5_to_19",
    "cal_2_to_4",
    "cal_1"
]

WEIGHT_GROUPS = [
    f"unseen_w{WEIGHT_EDGES[i]}_to_{WEIGHT_EDGES[i+1]-1}"
    for i in range(len(WEIGHT_EDGES)-1)
]

GROUPS = BASE_GROUPS + WEIGHT_GROUPS

# 13 maximum-probability upper bounds
# 13 failure-mass upper bounds
# 1 dominant-probability lower bound
# for each of 3 physical error rates.
EVENTS_PER_P = 2*len(GROUPS)+1
TOTAL_EVENTS = EVENTS_PER_P*len(P_VALUES)
EPS = DELTA/TOTAL_EVENTS

def cp_upper(k, n):
    if k == n:
        return 1.0
    return float(beta.isf(EPS, k+1, n-k))

def cp_lower(k, n):
    if k == 0:
        return 0.0
    return float(beta.ppf(EPS, k, n-k+1))

def build_circuit(d, p):
    return stim.Circuit.generated(
        "surface_code:rotated_memory_x",
        distance=d,
        rounds=d,
        after_clifford_depolarization=p,
        before_round_data_depolarization=p,
        after_reset_flip_probability=p,
        before_measure_flip_probability=p
    )

# All simulation settings and seeds are determined now.
EXPERIMENTS = [
    {
        "p": p,
        "cal_seed": 20261201 + 100*i,
        "eval_seed": 20261202 + 100*i
    }
    for i, p in enumerate(P_VALUES)
]

# Freeze circuit fingerprints before collecting shots.
for exp in EXPERIMENTS:
    circuit = build_circuit(DISTANCE, exp["p"])
    exp["circuit_sha256"] = hashlib.sha256(
        str(circuit).encode()
    ).hexdigest()

config = {
    "distance": DISTANCE,
    "p_values": P_VALUES,
    "alphas": ALPHAS,
    "calibration_shots": CAL_SHOTS,
    "evaluation_shots": EVAL_SHOTS,
    "chunk_size": CHUNK,
    "delta": DELTA,
    "target": TARGET,
    "weight_edges": WEIGHT_EDGES,
    "groups": GROUPS,
    "events_per_p": EVENTS_PER_P,
    "total_events": TOTAL_EVENTS,
    "eps": EPS,
    "experiments": EXPERIMENTS,
    "stim_version": stim.__version__,
    "pymatching_version": getattr(
        pymatching, "__version__", "unknown"
    ),
    "decoder": "PyMatching from decomposed Stim DEM",
    "observable": "rotated_memory_x",
    "analysis_version": "fixed_weight_stratification_v1"
}

with open(f"{OUT}/frozen_config.json", "w") as f:
    json.dump(config, f, indent=2)

def collect(sampler, decoder, shots, label):
    counts = Counter()
    failures = Counter()
    total_failures = 0

    for start in range(0, shots, CHUNK):
        n = min(CHUNK, shots-start)

        det, obs = sampler.sample(
            shots=n, separate_observables=True
        )

        pred = decoder.decode_batch(det)
        bad = np.any(pred != obs, axis=1)

        packed = np.packbits(
            det, axis=1, bitorder="little"
        )

        counts.update(map(bytes, packed))
        failures.update(map(bytes, packed[bad]))
        total_failures += int(bad.sum())

        done = start+n

        if done % 200_000 == 0 or done == shots:
            print(f"{label}: {done:,}/{shots:,}",
                  flush=True)

    assert sum(counts.values()) == shots
    assert sum(failures.values()) == total_failures

    return counts, failures, total_failures

all_results = []
all_groups = []

for i, exp in enumerate(EXPERIMENTS):
    p = exp["p"]
    tag = f"p_{p:.3f}".replace(".", "_")
    folder = f"{OUT}/{tag}"
    os.makedirs(folder, exist_ok=True)

    print(f"\n{'='*55}")
    print(f"Physical error probability p={p}")
    print(f"{'='*55}")

    circuit = build_circuit(DISTANCE, p)

    decoder = pymatching.Matching.from_detector_error_model(
        circuit.detector_error_model(decompose_errors=True)
    )

    # --------------------------------------------------------
    # 1. Independent calibration
    # --------------------------------------------------------

    cal_sampler = circuit.compile_detector_sampler(
        seed=exp["cal_seed"]
    )

    cal, cal_fail, cal_total = collect(
        cal_sampler, decoder, CAL_SHOTS, "Calibration"
    )

    dominant = cal.most_common(1)[0][0]

    def assign_group(s):
        if s == dominant:
            return "dominant"

        c = cal.get(s, 0)

        if c >= 100:
            return "cal_100_plus"
        if c >= 20:
            return "cal_20_to_99"
        if c >= 5:
            return "cal_5_to_19"
        if c >= 2:
            return "cal_2_to_4"
        if c == 1:
            return "cal_1"

        w = int.from_bytes(
            s, "little"
        ).bit_count()

        for j in range(len(WEIGHT_EDGES)-1):
            if WEIGHT_EDGES[j] <= w < WEIGHT_EDGES[j+1]:
                return WEIGHT_GROUPS[j]

        raise ValueError(f"Unexpected weight {w}")

    # Save calibration before independent evaluation.
    with gzip.open(f"{folder}/calibration.pkl.gz", "wb") as f:
        pickle.dump({
            "counts": cal,
            "failures": cal_fail,
            "total_failures": cal_total,
            "shots": CAL_SHOTS,
            "seed": exp["cal_seed"],
            "dominant": dominant
        }, f, protocol=pickle.HIGHEST_PROTOCOL)

    print(
        f"Calibration: {len(cal):,} unique syndromes, "
        f"dominant count {cal[dominant]:,}"
    )

    # --------------------------------------------------------
    # 2. Independent evaluation
    # --------------------------------------------------------

    eval_sampler = circuit.compile_detector_sampler(
        seed=exp["eval_seed"]
    )

    ev, fail, total_failed = collect(
        eval_sampler, decoder, EVAL_SHOTS, "Evaluation"
    )

    with gzip.open(f"{folder}/evaluation.pkl.gz", "wb") as f:
        pickle.dump({
            "counts": ev,
            "failures": fail,
            "total_failures": total_failed,
            "shots": EVAL_SHOTS,
            "seed": exp["eval_seed"]
        }, f, protocol=pickle.HIGHEST_PROTOCOL)

    # --------------------------------------------------------
    # 3. Reconstruct fixed groups
    # --------------------------------------------------------

    stats = {
        g: {
            "shots": 0,
            "max_count": 0,
            "failures": 0,
            "records_seen": 0
        }
        for g in GROUPS
    }

    for s, c in ev.items():
        g = assign_group(s)
        x = stats[g]

        x["shots"] += c
        x["max_count"] = max(x["max_count"], c)
        x["failures"] += fail.get(s, 0)
        x["records_seen"] += 1

    assert sum(x["shots"] for x in stats.values()) == EVAL_SHOTS
    assert sum(x["failures"] for x in stats.values()) == total_failed

    # --------------------------------------------------------
    # 4. Simultaneous probability bounds
    # --------------------------------------------------------

    p0_lower = cp_lower(
        ev.get(dominant, 0), EVAL_SHOTS
    )

    group_rows = []

    for g in GROUPS:
        x = stats[g]

        group_rows.append({
            "p": p,
            "group": g,
            **x,
            "max_p_upper": cp_upper(
                x["max_count"], EVAL_SHOTS
            ),
            "failure_mass_upper": cp_upper(
                x["failures"], EVAL_SHOTS
            )
        })

    group_df = pd.DataFrame(group_rows)

    for alpha in ALPHAS:
        contributions = (
            group_df["max_p_upper"].to_numpy()
            ** (alpha-1)
            * group_df["failure_mass_upper"].to_numpy()
        )

        group_df[f"F{alpha}_upper_contribution"] = (
            contributions
        )

        numerator_upper = float(contributions.sum())
        denominator_lower = float(p0_lower**alpha)

        upper = (
            min(1.0, numerator_upper/denominator_lower)
            if denominator_lower > 0
            else 1.0
        )

        all_results.append({
            "distance": DISTANCE,
            "p": p,
            "alpha": alpha,
            "calibration_shots": CAL_SHOTS,
            "evaluation_shots": EVAL_SHOTS,
            "baseline_LER": total_failed/EVAL_SHOTS,
            "dominant_eval_count": ev.get(dominant, 0),
            "dominant_eval_failures": fail.get(dominant, 0),
            "numerator_upper": numerator_upper,
            "denominator_lower": denominator_lower,
            "confidence_upper": upper,
            "certified": bool(upper < TARGET)
        })

    group_df.to_csv(
        f"{folder}/groupwise.csv", index=False
    )

    all_groups.append(group_df)

    current = pd.DataFrame(
        [r for r in all_results if r["p"] == p]
    )

    print("\nCERTIFICATION AT p =", p)
    print(current.to_string(index=False))

    # Save cumulative results after each completed p.
    pd.DataFrame(all_results).to_csv(
        f"{OUT}/certification_results.csv",
        index=False
    )

    del cal, cal_fail, ev, fail, stats
    del cal_sampler, eval_sampler, decoder, circuit
    gc.collect()

# ------------------------------------------------------------
# 5. Combined results
# ------------------------------------------------------------

result_df = pd.DataFrame(all_results)
group_df = pd.concat(all_groups, ignore_index=True)

group_df.to_csv(
    f"{OUT}/all_groupwise.csv", index=False
)

print("\n" + "="*65)
print("FINAL NOISE GENERALIZATION RESULTS")
print("="*65)

print(result_df.to_string(index=False))

print("\nCOMPACT COMPARISON")
print(
    result_df[
        ["p", "alpha", "baseline_LER",
         "confidence_upper", "certified"]
    ].to_string(index=False)
)

print("\nJoint confidence level: at least", 1-DELTA)
print("Total one-sided events:", TOTAL_EVENTS)
print("Per-event error probability:", EPS)
print("\nSaved to:", OUT)
```

```python

import gzip, pickle, math
import numpy as np
import pandas as pd

SOURCES = [
    (0.003, "qec_noise_generalization_v1/p_0_003/evaluation.pkl.gz"),
    (0.005, "qec_d5_weight_prospective/evaluation_counts.pkl.gz"),
    (0.007, "qec_noise_generalization_v1/p_0_007/evaluation.pkl.gz"),
    (0.010, "qec_noise_generalization_v1/p_0_010/evaluation.pkl.gz"),
]

def falling(n, k):
    if n < k:
        return 0
    return math.prod(n-j for j in range(k))

rows = []

for p, path in SOURCES:
    with gzip.open(path, "rb") as f:
        data = pickle.load(f)

    counts = data["counts"]
    failures = data["failures"]
    N = sum(counts.values())

    baseline = sum(failures.values()) / N
    most_common = counts.most_common(1)[0]

    s0, c0 = most_common

    for alpha in [2, 3]:
        denominator = falling(N, alpha)

        collision_tuples = 0
        failure_tuples = 0
        total_weight_squared = 0

        for s, c in counts.items():
            if c < alpha:
                continue

            w = falling(c, alpha)

            collision_tuples += w
            total_weight_squared += w*w

            e = failures.get(s, 0)

            if e:
                failure_tuples += (
                    e * falling(c-1, alpha-1)
                )

        C_hat = collision_tuples / denominator
        F_hat = failure_tuples / denominator

        ratio = (
            F_hat / C_hat
            if C_hat > 0 else np.nan
        )

        dominant_fraction = (
            falling(c0, alpha) / collision_tuples
            if collision_tuples else np.nan
        )

        effective_support = (
            collision_tuples**2 / total_weight_squared
            if total_weight_squared else np.nan
        )

        rows.append({
            "p": p,
            "alpha": alpha,
            "shots": N,
            "baseline_LER": baseline,
            "unique_syndromes": len(counts),
            "dominant_count": c0,
            "C_hat": C_hat,
            "F_hat": F_hat,
            "ratio_estimate": ratio,
            "failure_weighted_tuples": failure_tuples,
            "dominant_collision_share": dominant_fraction,
            "empirical_effective_support": effective_support,
            "dominant_only_C_lower_point":
                (c0/N)**alpha,
            "C_to_dominant_point_ratio":
                C_hat/((c0/N)**alpha)
        })

result = pd.DataFrame(rows)

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 220)

print("\nCOLLISION MOMENT DIAGNOSTICS")
print(result.to_string(index=False))

print("\nCOMPACT COMPARISON")
print(
    result[[
        "p", "alpha", "C_hat", "F_hat",
        "ratio_estimate",
        "failure_weighted_tuples",
        "dominant_collision_share",
        "C_to_dominant_point_ratio"
    ]].to_string(index=False)
)

result.to_csv(
    "qec_collision_generalization_diagnostics.csv",
    index=False
)

print("\nSaved diagnostic CSV.")
print("Warning: Ratio estimates are not confidence bounds.")
print("Collision tuples are dependent, not independent trials.")
```

```python

import os, gzip, pickle, json, hashlib
from collections import Counter
import numpy as np
import pandas as pd
import stim, pymatching
from scipy.stats import beta

# =======================================================
# PROSPECTIVE DISTANCE-5, p=0.007 CERTIFICATION
# Independent 5-million-shot evaluation
# Both alpha=2 and alpha=3 included in one confidence family
# =======================================================

CAL_PATH = (
    "qec_noise_generalization_v1/"
    "p_0_007/calibration.pkl.gz"
)
OUT = "qec_d5_p007_prospective_v1"

D = 5
P = 0.007
N = 5_000_000
CHUNK = 20_000
SEED = 20261215

DELTA = 0.05
TARGET = 1e-3
ALPHAS = [2, 3]

EDGES = [0, 3, 5, 7, 9, 13, 21, 121]

BASE = [
    "dominant", "cal_100_plus", "cal_20_to_99",
    "cal_5_to_19", "cal_2_to_4", "cal_1"
]

WEIGHT = [
    f"unseen_w{EDGES[i]}_to_{EDGES[i+1]-1}"
    for i in range(len(EDGES)-1)
]

GROUPS = BASE + WEIGHT
EPS = DELTA / (2*len(GROUPS)+1)

os.makedirs(OUT, exist_ok=True)

with gzip.open(CAL_PATH, "rb") as f:
    calibration = pickle.load(f)

cal = calibration["counts"]
dominant = calibration["dominant"]

assert sum(cal.values()) == 1_000_000

def classify(s):
    if s == dominant:
        return "dominant"

    c = cal.get(s, 0)
    if c >= 100: return "cal_100_plus"
    if c >= 20: return "cal_20_to_99"
    if c >= 5: return "cal_5_to_19"
    if c >= 2: return "cal_2_to_4"
    if c == 1: return "cal_1"

    w = int.from_bytes(s, "little").bit_count()

    for i in range(len(EDGES)-1):
        if EDGES[i] <= w < EDGES[i+1]:
            return WEIGHT[i]

    raise ValueError("Unclassified syndrome")

def upper(k, n):
    if k == n:
        return 1.0
    return float(beta.isf(EPS, k+1, n-k))

def lower(k, n):
    if k == 0:
        return 0.0
    return float(beta.ppf(EPS, k, n-k+1))

circuit = stim.Circuit.generated(
    "surface_code:rotated_memory_x",
    distance=D, rounds=D,
    after_clifford_depolarization=P,
    before_round_data_depolarization=P,
    after_reset_flip_probability=P,
    before_measure_flip_probability=P
)

config = {
    "distance": D,
    "p": P,
    "evaluation_shots": N,
    "seed": SEED,
    "delta": DELTA,
    "target": TARGET,
    "alphas": ALPHAS,
    "groups": GROUPS,
    "weight_edges": EDGES,
    "eps": EPS,
    "calibration_path": CAL_PATH,
    "stim_version": stim.__version__,
    "pymatching_version": getattr(
        pymatching, "__version__", "unknown"
    ),
    "circuit_sha256": hashlib.sha256(
        str(circuit).encode()
    ).hexdigest()
}

with open(f"{OUT}/frozen_config.json", "w") as f:
    json.dump(config, f, indent=2)

decoder = pymatching.Matching.from_detector_error_model(
    circuit.detector_error_model(decompose_errors=True)
)

sampler = circuit.compile_detector_sampler(seed=SEED)

counts = Counter()
failures = Counter()
total_failures = 0

for start in range(0, N, CHUNK):
    n = min(CHUNK, N-start)

    det, obs = sampler.sample(
        shots=n, separate_observables=True
    )
    predicted = decoder.decode_batch(det)
    bad = np.any(predicted != obs, axis=1)

    packed = np.packbits(
        det, axis=1, bitorder="little"
    )

    counts.update(map(bytes, packed))
    failures.update(map(bytes, packed[bad]))
    total_failures += int(bad.sum())

    done = start+n
    if done % 500_000 == 0 or done == N:
        print(f"Evaluation: {done:,}/{N:,}", flush=True)

assert sum(counts.values()) == N
assert sum(failures.values()) == total_failures

stats = {
    g: {
        "shots": 0,
        "failures": 0,
        "max_count": 0,
        "records_seen": 0
    }
    for g in GROUPS
}

for s, c in counts.items():
    x = stats[classify(s)]
    x["shots"] += c
    x["failures"] += failures.get(s, 0)
    x["max_count"] = max(x["max_count"], c)
    x["records_seen"] += 1

assert sum(x["shots"] for x in stats.values()) == N
assert sum(x["failures"] for x in stats.values()) == total_failures

dominant_n = counts.get(dominant, 0)
p0_lower = lower(dominant_n, N)

rows = []

for g in GROUPS:
    x = stats[g]
    rows.append({
        "group": g,
        **x,
        "max_p_upper": upper(x["max_count"], N),
        "failure_mass_upper": upper(x["failures"], N)
    })

group_df = pd.DataFrame(rows)
results = []

for alpha in ALPHAS:
    contributions = (
        group_df["max_p_upper"].to_numpy()**(alpha-1)
        * group_df["failure_mass_upper"].to_numpy()
    )

    group_df[f"F{alpha}_upper_contribution"] = (
        contributions
    )

    numerator = float(contributions.sum())
    denominator = float(p0_lower**alpha)

    bound = (
        min(1.0, numerator/denominator)
        if denominator > 0 else 1.0
    )

    results.append({
        "distance": D,
        "p": P,
        "alpha": alpha,
        "evaluation_shots": N,
        "baseline_LER": total_failures/N,
        "dominant_count": dominant_n,
        "dominant_failures": failures.get(dominant, 0),
        "numerator_upper": numerator,
        "denominator_lower": denominator,
        "confidence_upper": bound,
        "certified": bound < TARGET
    })

result_df = pd.DataFrame(results)

print("\nPROSPECTIVE CERTIFICATION")
print(result_df.to_string(index=False))

print("\nGROUPWISE CONTRIBUTIONS")
print(group_df.to_string(index=False))

print("\nJoint confidence: at least 95%")
print("Target:", TARGET)
print("Bonferroni events:", 2*len(GROUPS)+1)

result_df.to_csv(f"{OUT}/results.csv", index=False)
group_df.to_csv(f"{OUT}/groups.csv", index=False)

with gzip.open(f"{OUT}/evaluation.pkl.gz", "wb") as f:
    pickle.dump({
        "counts": counts,
        "failures": failures,
        "shots": N,
        "seed": SEED,
        "total_failures": total_failures
    }, f, protocol=pickle.HIGHEST_PROTOCOL)

print("\nSaved to:", OUT)
```

```python

import os, gzip, pickle, math, gc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# DOMINANT-SYNDROME ABLATION AND EFFECTIVE-SUPPORT ANALYSIS
#
# Descriptive sensitivity study, NOT confidence certification.
# No additional Stim shots.
# ============================================================

OUT = "qec_dominant_ablation_v1"
os.makedirs(OUT, exist_ok=True)

SOURCES = [
    {
        "p": 0.003,
        "evaluation":
            "qec_noise_generalization_v1/p_0_003/evaluation.pkl.gz",
        "calibration":
            "qec_noise_generalization_v1/p_0_003/calibration.pkl.gz"
    },
    {
        "p": 0.005,
        "evaluation":
            "qec_d5_weight_prospective/evaluation_counts.pkl.gz",
        "calibration":
            "qec_d5_alpha2_validation/counts.pkl.gz"
    },
    {
        "p": 0.007,
        "evaluation":
            "qec_d5_p007_prospective_v1/evaluation.pkl.gz",
        "calibration":
            "qec_noise_generalization_v1/p_0_007/calibration.pkl.gz"
    },
    {
        "p": 0.010,
        "evaluation":
            "qec_noise_generalization_v1/p_0_010/evaluation.pkl.gz",
        "calibration":
            "qec_noise_generalization_v1/p_0_010/calibration.pkl.gz"
    }
]

ALPHAS = [2, 3]
TOP_K = [1, 10, 100]

def falling(n, a):
    if n < a:
        return 0
    result = 1
    for j in range(a):
        result *= (n-j)
    return result

def safe_ratio(num, den):
    return float(num / den) if den > 0 else np.nan

rows = []
top_rows = []

for spec in SOURCES:
    p = spec["p"]

    print(f"\nAnalyzing p={p:.3f}", flush=True)

    # Only load pickle archives you trust.
    with gzip.open(spec["evaluation"], "rb") as f:
        evaluation = pickle.load(f)

    with gzip.open(spec["calibration"], "rb") as f:
        calibration = pickle.load(f)

    counts = evaluation["counts"]
    failures = evaluation["failures"]

    # Both saved calibration formats are supported.
    if "dominant" in calibration:
        dominant = calibration["dominant"]
    else:
        dominant = calibration["calibration"].most_common(1)[0][0]

    N = sum(counts.values())
    E = sum(failures.values())

    assert N > 0
    assert all(
        0 <= failures.get(s, 0) <= c
        for s, c in counts.items()
    )

    dominant_count = counts.get(dominant, 0)
    dominant_failures = failures.get(dominant, 0)

    print(
        "Shots:", f"{N:,}",
        "| Unique:", f"{len(counts):,}",
        "| Dominant hits:", f"{dominant_count:,}",
        "| Dominant failures:", dominant_failures
    )

    for alpha in ALPHAS:
        total_C_tuples = 0
        total_F_tuples = 0
        sum_w_squared = 0

        # Eligible tuples only, for concentration statistics.
        weights = []

        for s, c in counts.items():
            if c < alpha:
                continue

            w = falling(c, alpha)
            e = failures.get(s, 0)
            fw = e * falling(c-1, alpha-1) if e else 0

            total_C_tuples += w
            total_F_tuples += fw
            sum_w_squared += w*w

            weights.append((w, s, fw))

        dom_C = falling(dominant_count, alpha)

        dom_F = (
            dominant_failures
            * falling(dominant_count-1, alpha-1)
            if dominant_failures and dominant_count >= alpha
            else 0
        )

        residual_C = total_C_tuples - dom_C
        residual_F = total_F_tuples - dom_F

        full_ratio = safe_ratio(
            total_F_tuples, total_C_tuples
        )
        ablated_ratio = safe_ratio(
            residual_F, residual_C
        )

        dominant_Q = safe_ratio(
            dom_C, total_C_tuples
        )

        residual_Q = safe_ratio(
            residual_C, total_C_tuples
        )

        effective_support = safe_ratio(
            total_C_tuples**2,
            sum_w_squared
        )

        residual_w2 = sum_w_squared - dom_C**2
        ablated_effective_support = safe_ratio(
            residual_C**2, residual_w2
        )

        weights.sort(key=lambda x: x[0], reverse=True)

        for k in TOP_K:
            selected = weights[:k]
            covered = sum(item[0] for item in selected)

            top_rows.append({
                "p": p,
                "alpha": alpha,
                "top_k": k,
                "cumulative_Q_mass": safe_ratio(
                    covered, total_C_tuples
                )
            })

        rows.append({
            "p": p,
            "alpha": alpha,
            "shots": N,
            "baseline_LER": E/N,
            "unique_syndromes": len(counts),
            "dominant_count": dominant_count,
            "dominant_failures": dominant_failures,
            "dominant_empirical_f":
                safe_ratio(dominant_failures, dominant_count),
            "dominant_Q_mass": dominant_Q,
            "remaining_Q_mass": residual_Q,
            "effective_support": effective_support,
            "ablated_effective_support":
                ablated_effective_support,
            "full_ratio_estimate": full_ratio,
            "ablated_ratio_estimate": ablated_ratio,
            "full_failure_weighted_tuples": total_F_tuples,
            "ablated_failure_weighted_tuples": residual_F,
            "dominant_failure_fraction":
                safe_ratio(dom_F, total_F_tuples),
            "ablated_collision_tuples": residual_C,
            "ablated_syndromes_with_collisions":
                len(weights) - int(dom_C > 0)
        })

    del evaluation, calibration, counts, failures, weights
    gc.collect()

result = pd.DataFrame(rows)
top = pd.DataFrame(top_rows)

print("\n" + "="*72)
print("DOMINANT-SYNDROME ABLATION")
print("="*72)
print(result.to_string(index=False))

print("\nCOMPACT ABLATION COMPARISON")
print(
    result[[
        "p", "alpha",
        "dominant_Q_mass",
        "remaining_Q_mass",
        "effective_support",
        "full_ratio_estimate",
        "ablated_ratio_estimate",
        "ablated_failure_weighted_tuples"
    ]].to_string(index=False)
)

print("\nTOP-K COLLISION-WEIGHT COVERAGE")
print(top.to_string(index=False))

result.to_csv(
    f"{OUT}/ablation_results.csv", index=False
)
top.to_csv(
    f"{OUT}/top_k_concentration.csv", index=False
)

# Plot 1: dominant syndrome share
plt.figure(figsize=(8, 5))
for alpha in ALPHAS:
    d = result[result.alpha == alpha].sort_values("p")
    plt.plot(
        d.p, d.dominant_Q_mass,
        marker="o", label=f"alpha={alpha}"
    )

plt.xlabel("Physical error probability")
plt.ylabel("Dominant syndrome share of empirical Q")
plt.title("Dominant-syndrome concentration")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(
    f"{OUT}/dominant_concentration.png", dpi=160
)
plt.show()

# Plot 2: full vs ablated ratios
plt.figure(figsize=(8, 5))

for alpha in ALPHAS:
    d = result[result.alpha == alpha].sort_values("p")
    plt.plot(
        d.p, d.full_ratio_estimate,
        marker="o", linestyle="-",
        label=f"Full, alpha={alpha}"
    )
    plt.plot(
        d.p, d.ablated_ratio_estimate,
        marker="s", linestyle="--",
        label=f"Ablated, alpha={alpha}"
    )

plt.yscale("symlog", linthresh=1e-10)
plt.xlabel("Physical error probability")
plt.ylabel("Logical-failure ratio estimate")
plt.title("Full vs dominant-excluded resampling")
plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()
plt.savefig(
    f"{OUT}/ablation_ratios.png", dpi=160
)
plt.show()

print("\nSaved analysis to:", OUT)
print("All ratio estimates are exploratory.")
print("Collision tuples are dependent observations.")
print("No new confidence certificates are claimed.")
```

```python

import os, gzip, pickle, json
from collections import defaultdict
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# DOMINANT-EXCLUDED FINITE-SAMPLE CERTIFICATION PILOT
#
# Uses saved independent calibration/evaluation archives.
# No new Stim shots.
#
# IMPORTANT: This is a retrospective feasibility analysis.
# New prospective validation is needed for a clean,
# independently confirmed research milestone.
# ============================================================

OUT = "qec_ablated_certification_pilot"
os.makedirs(OUT, exist_ok=True)

SOURCES = [
    (
        0.003,
        "qec_noise_generalization_v1/p_0_003/calibration.pkl.gz",
        "qec_noise_generalization_v1/p_0_003/evaluation.pkl.gz"
    ),
    (
        0.005,
        "qec_d5_alpha2_validation/counts.pkl.gz",
        "qec_d5_weight_prospective/evaluation_counts.pkl.gz"
    ),
    (
        0.007,
        "qec_noise_generalization_v1/p_0_007/calibration.pkl.gz",
        "qec_d5_p007_prospective_v1/evaluation.pkl.gz"
    ),
    (
        0.010,
        "qec_noise_generalization_v1/p_0_010/calibration.pkl.gz",
        "qec_noise_generalization_v1/p_0_010/evaluation.pkl.gz"
    )
]

ALPHAS = [2, 3]
KS = [10, 100, 1000]
DELTA = 0.05
TARGET = 1e-3

EDGES = [0, 3, 5, 7, 9, 13, 21, 121]

BASE = [
    "cal_100_plus",
    "cal_20_to_99",
    "cal_5_to_19",
    "cal_2_to_4",
    "cal_1"
]

WEIGHT = [
    f"unseen_w{EDGES[i]}_to_{EDGES[i+1]-1}"
    for i in range(len(EDGES)-1)
]

GROUPS = BASE + WEIGHT
MAX_K = max(KS)

# 12 group maximum-probability bounds
# 12 group failure-mass bounds
# 1000 selected-syndrome lower bounds
#
# All K and both alpha values reuse the same events.
EVENTS_PER_P = 2*len(GROUPS) + MAX_K
TOTAL_EVENTS = EVENTS_PER_P * len(SOURCES)
EPS = DELTA / TOTAL_EVENTS

def cp_upper(k, n):
    if k >= n:
        return 1.0
    return float(beta.isf(EPS, k+1, n-k))

def cp_lower(k, n):
    if k <= 0:
        return 0.0
    return float(beta.ppf(EPS, k, n-k+1))

def falling(n, a):
    result = 1
    for j in range(a):
        if n-j <= 0:
            return 0
        result *= (n-j)
    return result

all_results = []
all_groups = []

for p, cal_path, ev_path in SOURCES:
    print(f"\nAnalyzing p={p:.3f}", flush=True)

    # Only load trusted pickle files.
    with gzip.open(cal_path, "rb") as f:
        cd = pickle.load(f)

    with gzip.open(ev_path, "rb") as f:
        ed = pickle.load(f)

    if "calibration" in cd:
        cal = cd["calibration"]
        dominant = cd["dominant"]
    else:
        cal = cd["counts"]
        dominant = cd["dominant"]

    ev = ed["counts"]
    fail = ed["failures"]
    N = sum(ev.values())

    # Selection uses calibration ONLY.
    selected = [
        s for s, count in cal.most_common(MAX_K + 1)
        if s != dominant
    ][:MAX_K]

    assert len(selected) == MAX_K
    assert len(set(selected)) == MAX_K

    def classify(s):
        if s == dominant:
            return None

        c = cal.get(s, 0)

        if c >= 100:
            return "cal_100_plus"
        if c >= 20:
            return "cal_20_to_99"
        if c >= 5:
            return "cal_5_to_19"
        if c >= 2:
            return "cal_2_to_4"
        if c == 1:
            return "cal_1"

        w = int.from_bytes(s, "little").bit_count()

        for i in range(len(EDGES)-1):
            if EDGES[i] <= w < EDGES[i+1]:
                return WEIGHT[i]

        raise ValueError("Unclassified syndrome")

    stats = {
        g: {
            "shots": 0,
            "failures": 0,
            "max_count": 0,
            "records_seen": 0
        }
        for g in GROUPS
    }

    for s, c in ev.items():
        g = classify(s)
        if g is None:
            continue

        x = stats[g]
        x["shots"] += c
        x["failures"] += fail.get(s, 0)
        x["max_count"] = max(x["max_count"], c)
        x["records_seen"] += 1

    assert (
        sum(x["shots"] for x in stats.values())
        == N - ev.get(dominant, 0)
    )

    assert (
        sum(x["failures"] for x in stats.values())
        == sum(fail.values()) - fail.get(dominant, 0)
    )

    group_rows = []

    for g in GROUPS:
        x = stats[g]

        group_rows.append({
            "p": p,
            "group": g,
            **x,
            "max_p_upper": cp_upper(
                x["max_count"], N
            ),
            "failure_mass_upper": cp_upper(
                x["failures"], N
            )
        })

    group_df = pd.DataFrame(group_rows)

    # Lower confidence bounds for calibration-selected
    # non-dominant syndrome probabilities.
    lower_p = np.array([
        cp_lower(ev.get(s, 0), N)
        for s in selected
    ])

    eval_counts_selected = [
        ev.get(s, 0) for s in selected
    ]

    for alpha in ALPHAS:
        upper_terms = (
            group_df["max_p_upper"].to_numpy()
            ** (alpha-1)
            * group_df["failure_mass_upper"].to_numpy()
        )

        numerator_upper = float(upper_terms.sum())

        # Full and ablated collision moments are
        # descriptive U-statistic estimates.
        collision_tuples = 0
        failed_tuples = 0

        for s, c in ev.items():
            if s == dominant or c < alpha:
                continue

            collision_tuples += falling(c, alpha)

            e = fail.get(s, 0)
            if e:
                failed_tuples += (
                    e * falling(c-1, alpha-1)
                )

        ratio_estimate = (
            failed_tuples / collision_tuples
            if collision_tuples else np.nan
        )

        for K in KS:
            denominator_lower = float(
                np.sum(lower_p[:K]**alpha)
            )

            confidence_upper = (
                min(
                    1.0,
                    numerator_upper / denominator_lower
                )
                if denominator_lower > 0
                else 1.0
            )

            all_results.append({
                "p": p,
                "alpha": alpha,
                "K": K,
                "shots": N,
                "selected_eval_hits":
                    sum(eval_counts_selected[:K]),
                "numerator_upper":
                    numerator_upper,
                "denominator_lower":
                    denominator_lower,
                "confidence_upper":
                    confidence_upper,
                "exploratory_ratio":
                    ratio_estimate,
                "below_target":
                    confidence_upper < TARGET
            })

        group_df[
            f"F{alpha}_upper_contribution"
        ] = upper_terms

    all_groups.append(group_df)

    print(
        f"Evaluation shots: {N:,} | "
        f"Selected non-dominant syndromes: {MAX_K:,}"
    )

results = pd.DataFrame(all_results)
groups = pd.concat(all_groups, ignore_index=True)

print("\nABLATION CERTIFICATION FEASIBILITY")
print(results.to_string(index=False))

print("\nBEST BOUND BY PHYSICAL NOISE AND ALPHA")
best = results.loc[
    results.groupby(["p", "alpha"])[
        "confidence_upper"
    ].idxmin()
].sort_values(["p", "alpha"])

print(best.to_string(index=False))

print("\nCONFIDENCE ACCOUNTING")
print("Total one-sided events:", TOTAL_EVENTS)
print("Per-event delta:", EPS)
print("Simultaneous nominal coverage:", 1-DELTA)

results.to_csv(
    f"{OUT}/all_bounds.csv", index=False
)
best.to_csv(
    f"{OUT}/best_bounds.csv", index=False
)
groups.to_csv(
    f"{OUT}/groupwise_bounds.csv", index=False
)

with open(f"{OUT}/analysis_config.json", "w") as f:
    json.dump({
        "source_paths": SOURCES,
        "alphas": ALPHAS,
        "K_values": KS,
        "weight_edges": EDGES,
        "groups": GROUPS,
        "delta": DELTA,
        "target": TARGET,
        "events_per_p": EVENTS_PER_P,
        "total_events": TOTAL_EVENTS,
        "eps": EPS,
        "analysis_type":
            "retrospective_ablation_feasibility"
    }, f, indent=2)

print("\nSaved to:", OUT)
print("Retrospective feasibility analysis completed.")
```

```python

import os, gzip, pickle, math, json
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# VARIATIONAL CERTIFICATION OF DOMINANT-EXCLUDED RESAMPLING
#
# Independent calibration weights + evaluation Hoeffding bound.
# Retrospective method-development comparison.
# No new quantum simulations.
# ============================================================

SOURCES = [
    (0.003,
     "qec_noise_generalization_v1/p_0_003/calibration.pkl.gz",
     "qec_noise_generalization_v1/p_0_003/evaluation.pkl.gz"),
    (0.005,
     "qec_d5_alpha2_validation/counts.pkl.gz",
     "qec_d5_weight_prospective/evaluation_counts.pkl.gz"),
    (0.007,
     "qec_noise_generalization_v1/p_0_007/calibration.pkl.gz",
     "qec_d5_p007_prospective_v1/evaluation.pkl.gz"),
    (0.010,
     "qec_noise_generalization_v1/p_0_010/calibration.pkl.gz",
     "qec_noise_generalization_v1/p_0_010/evaluation.pkl.gz")
]

ALPHAS = [2, 3]
DELTA = 0.05
TARGET = 1e-3
OUT = "qec_ablated_variational_pilot"

EDGES = [0, 3, 5, 7, 9, 13, 21, 121]
BASE = [
    "cal_100_plus", "cal_20_to_99",
    "cal_5_to_19", "cal_2_to_4", "cal_1"
]
WEIGHT = [
    f"unseen_w{EDGES[i]}_to_{EDGES[i+1]-1}"
    for i in range(len(EDGES)-1)
]
GROUPS = BASE + WEIGHT

os.makedirs(OUT, exist_ok=True)

# Per p:
# 12 group atom-probability upper bounds
# 12 group failure-mass upper bounds
# 2 variational expectation lower bounds, alpha=2 and 3
EVENTS_PER_P = 2 * len(GROUPS) + len(ALPHAS)
TOTAL_EVENTS = EVENTS_PER_P * len(SOURCES)
EPS = DELTA / TOTAL_EVENTS

assert EVENTS_PER_P == 26
assert TOTAL_EVENTS == 104

def cp_upper(k, n):
    if k == n:
        return 1.0
    return float(beta.isf(EPS, k+1, n-k))

rows = []
group_rows = []

for p, cal_path, ev_path in SOURCES:
    print(f"Analyzing p={p:.3f}", flush=True)

    # Only deserialize archives you trust.
    with gzip.open(cal_path, "rb") as f:
        cd = pickle.load(f)

    with gzip.open(ev_path, "rb") as f:
        ed = pickle.load(f)

    cal = cd["calibration"] if "calibration" in cd else cd["counts"]
    dominant = cd["dominant"]

    ev = ed["counts"]
    failures = ed["failures"]

    M = sum(cal.values())
    N = sum(ev.values())

    assert M > 0 and N > 0
    assert sum(failures.values()) <= N

    def classify(s):
        if s == dominant:
            return None

        c = cal.get(s, 0)

        if c >= 100: return "cal_100_plus"
        if c >= 20: return "cal_20_to_99"
        if c >= 5: return "cal_5_to_19"
        if c >= 2: return "cal_2_to_4"
        if c == 1: return "cal_1"

        w = int.from_bytes(s, "little").bit_count()

        for j in range(len(EDGES)-1):
            if EDGES[j] <= w < EDGES[j+1]:
                return WEIGHT[j]

        raise ValueError("Unclassified detector record")

    stats = {
        g: {"shots":0, "failures":0, "max_count":0}
        for g in GROUPS
    }

    for s, c in ev.items():
        g = classify(s)
        if g is None:
            continue

        stats[g]["shots"] += c
        stats[g]["failures"] += failures.get(s, 0)
        stats[g]["max_count"] = max(
            stats[g]["max_count"], c
        )

    assert sum(x["shots"] for x in stats.values()) == (
        N - ev.get(dominant, 0)
    )
    assert sum(x["failures"] for x in stats.values()) == (
        sum(failures.values()) - failures.get(dominant, 0)
    )

    atom_upper = np.array([
        cp_upper(stats[g]["max_count"], N)
        for g in GROUPS
    ])

    failure_upper = np.array([
        cp_upper(stats[g]["failures"], N)
        for g in GROUPS
    ])

    for j, g in enumerate(GROUPS):
        group_rows.append({
            "p":p,
            "group":g,
            **stats[g],
            "atom_upper":atom_upper[j],
            "failure_mass_upper":failure_upper[j]
        })

    # Fixed calibration weights q_s = c_s / M,
    # excluding dominant syndrome.
    q = {
        s: c/M
        for s, c in cal.items()
        if s != dominant
    }

    qmax = max(q.values(), default=0.0)

    for alpha in ALPHAS:

        # Exact calibration-dependent constant.
        q_power_sum = math.fsum(
            x**alpha for x in q.values()
        )

        # Evaluation sample mean of q(S)^(alpha-1).
        # Written using sufficient counts rather than
        # materializing all N individual observations.
        empirical_mean = math.fsum(
            c * q.get(s, 0.0)**(alpha-1)
            for s, c in ev.items()
        ) / N

        # Conditional on independent calibration,
        # each evaluation contribution lies in [0, R].
        R = qmax**(alpha-1)

        hoeffding_radius = (
            R * math.sqrt(math.log(1/EPS)/(2*N))
        )

        expectation_lower = max(
            0.0, empirical_mean-hoeffding_radius
        )

        # The tangent inequality gives the denominator
        # lower bound. Clip negative bounds to zero.
        denominator_lower = max(
            0.0,
            alpha*expectation_lower
            - (alpha-1)*q_power_sum
        )

        numerator_upper = float(np.sum(
            atom_upper**(alpha-1) * failure_upper
        ))

        confidence_upper = (
            min(1.0, numerator_upper/denominator_lower)
            if denominator_lower > 0
            else 1.0
        )

        rows.append({
            "p":p,
            "alpha":alpha,
            "shots":N,
            "calibration_shots":M,
            "qmax":qmax,
            "empirical_weighted_mean":empirical_mean,
            "hoeffding_radius":hoeffding_radius,
            "denominator_lower":denominator_lower,
            "numerator_upper":numerator_upper,
            "confidence_upper":confidence_upper,
            "below_target":confidence_upper<TARGET
        })

results = pd.DataFrame(rows)
groups = pd.DataFrame(group_rows)

print("\nVARIATIONAL ABLATION BOUNDS")
print(results.to_string(index=False))

print("\nCOMPACT COMPARISON")
previous = {
    (0.003,2):0.001449,
    (0.003,3):0.000459,
    (0.005,2):0.025789,
    (0.005,3):0.002288,
    (0.007,2):0.410193,
    (0.007,3):0.031188,
    (0.010,2):1.0,
    (0.010,3):1.0
}

results["previous_upper"] = [
    previous[(p,a)]
    for p,a in zip(results.p, results.alpha)
]

results["improvement_factor"] = (
    results["previous_upper"]
    / results["confidence_upper"].replace(0, np.nan)
)

print(results[[
    "p","alpha","denominator_lower",
    "confidence_upper","previous_upper",
    "improvement_factor","below_target"
]].to_string(index=False))

print("\nSimultaneous nominal coverage:", 1-DELTA)
print("Total one-sided confidence events:", TOTAL_EVENTS)
print("Per-event delta:", EPS)

results.to_csv(f"{OUT}/results.csv", index=False)
groups.to_csv(f"{OUT}/groups.csv", index=False)

with open(f"{OUT}/config.json", "w") as f:
    json.dump({
        "sources":SOURCES,
        "alphas":ALPHAS,
        "delta":DELTA,
        "target":TARGET,
        "weight_edges":EDGES,
        "groups":GROUPS,
        "confidence_events":TOTAL_EVENTS,
        "eps":EPS,
        "method":"calibration_tangent_hoeffding"
    }, f, indent=2)

print("\nSaved to:", OUT)
print("Retrospective method-development analysis only.")
```

```python

import os, gzip, pickle, math
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# VARIATIONAL ABLATION: HOEFFDING VS EMPIRICAL BERNSTEIN
# Retrospective methodology experiment, no new shots
# ============================================================

OUT = "qec_ablated_empirical_bernstein_pilot"
os.makedirs(OUT, exist_ok=True)

SOURCES = [
    (0.003,
     "qec_noise_generalization_v1/p_0_003/calibration.pkl.gz",
     "qec_noise_generalization_v1/p_0_003/evaluation.pkl.gz"),
    (0.005,
     "qec_d5_alpha2_validation/counts.pkl.gz",
     "qec_d5_weight_prospective/evaluation_counts.pkl.gz"),
    (0.007,
     "qec_noise_generalization_v1/p_0_007/calibration.pkl.gz",
     "qec_d5_p007_prospective_v1/evaluation.pkl.gz"),
    (0.010,
     "qec_noise_generalization_v1/p_0_010/calibration.pkl.gz",
     "qec_noise_generalization_v1/p_0_010/evaluation.pkl.gz")
]

ALPHAS = [2, 3]
DELTA = 0.05
TARGET = 1e-3

EDGES = [0, 3, 5, 7, 9, 13, 21, 121]
BASE = [
    "cal_100_plus", "cal_20_to_99",
    "cal_5_to_19", "cal_2_to_4", "cal_1"
]
WEIGHT = [
    f"unseen_w{EDGES[i]}_to_{EDGES[i+1]-1}"
    for i in range(len(EDGES)-1)
]
GROUPS = BASE + WEIGHT

# Simultaneous coverage for both alternatives:
# 24 group numerator events per physical error rate
# 2 Hoeffding denominator events
# 2 empirical Bernstein denominator events
EVENTS_PER_P = 2*len(GROUPS) + 2*len(ALPHAS)
EPS = DELTA/(EVENTS_PER_P*len(SOURCES))

def cp_upper(k, n):
    if k == n:
        return 1.0
    return float(beta.isf(EPS, k+1, n-k))

results = []

for p, cal_path, ev_path in SOURCES:
    print(f"Analyzing p={p:.3f}", flush=True)

    # Load only trusted pickle archives.
    with gzip.open(cal_path, "rb") as f:
        cd = pickle.load(f)
    with gzip.open(ev_path, "rb") as f:
        ed = pickle.load(f)

    cal = cd["calibration"] if "calibration" in cd else cd["counts"]
    dominant = cd["dominant"]
    ev, failures = ed["counts"], ed["failures"]

    M = sum(cal.values())
    N = sum(ev.values())

    def classify(s):
        if s == dominant:
            return None

        c = cal.get(s, 0)
        if c >= 100: return "cal_100_plus"
        if c >= 20: return "cal_20_to_99"
        if c >= 5: return "cal_5_to_19"
        if c >= 2: return "cal_2_to_4"
        if c == 1: return "cal_1"

        w = int.from_bytes(s, "little").bit_count()
        for j in range(len(EDGES)-1):
            if EDGES[j] <= w < EDGES[j+1]:
                return WEIGHT[j]

        raise ValueError("Unexpected detector weight")

    stats = {
        g: {"failures":0, "max_count":0}
        for g in GROUPS
    }

    for s, c in ev.items():
        g = classify(s)
        if g is None:
            continue
        stats[g]["failures"] += failures.get(s, 0)
        stats[g]["max_count"] = max(
            stats[g]["max_count"], c
        )

    u = np.array([
        cp_upper(stats[g]["max_count"], N)
        for g in GROUPS
    ])

    b = np.array([
        cp_upper(stats[g]["failures"], N)
        for g in GROUPS
    ])

    q = {
        s:c/M for s,c in cal.items()
        if s != dominant
    }

    qmax = max(q.values(), default=0.0)

    for alpha in ALPHAS:
        power = alpha-1
        R = qmax**power

        # Compute evaluation moments from counts.
        mean = math.fsum(
            c*q.get(s, 0.0)**power
            for s,c in ev.items()
        )/N

        second = math.fsum(
            c*q.get(s, 0.0)**(2*power)
            for s,c in ev.items()
        )/N

        sample_var = max(
            0.0,
            N*(second-mean*mean)/(N-1)
        )

        q_constant = math.fsum(
            value**alpha for value in q.values()
        )

        hoeffding_radius = (
            R*math.sqrt(math.log(1/EPS)/(2*N))
        )

        log_term = math.log(2/EPS)
        bernstein_radius = (
            math.sqrt(
                2*sample_var*log_term/N
            )
            + 7*R*log_term/(3*(N-1))
        )

        numerator_upper = float(np.sum(
            u**power*b
        ))

        row = {
            "p":p,
            "alpha":alpha,
            "shots":N,
            "sample_variance":sample_var,
            "hoeffding_radius":hoeffding_radius,
            "bernstein_radius":bernstein_radius,
            "numerator_upper":numerator_upper
        }

        for name, radius in [
            ("hoeffding",hoeffding_radius),
            ("bernstein",bernstein_radius)
        ]:
            expectation_lower = max(
                0.0, mean-radius
            )

            denominator_lower = max(
                0.0,
                alpha*expectation_lower
                -(alpha-1)*q_constant
            )

            bound = (
                min(1.0, numerator_upper/denominator_lower)
                if denominator_lower > 0
                else 1.0
            )

            row[f"{name}_denominator"] = denominator_lower
            row[f"{name}_upper"] = bound
            row[f"{name}_below_target"] = bound < TARGET

        results.append(row)

df = pd.DataFrame(results)

print("\nVARIANCE-SENSITIVE ABLATION RESULTS")
print(df.to_string(index=False))

print("\nCOMPACT COMPARISON")
print(df[[
    "p", "alpha",
    "hoeffding_radius", "bernstein_radius",
    "hoeffding_denominator",
    "bernstein_denominator",
    "hoeffding_upper", "bernstein_upper",
    "bernstein_below_target"
]].to_string(index=False))

print("\nTotal one-sided events:",
      EVENTS_PER_P*len(SOURCES))
print("Per-event error allocation:", EPS)
print("Confidence allocation: 95% simultaneous")
print("Analysis status: retrospective")

df.to_csv(
    f"{OUT}/results.csv", index=False
)
```

```python

import os, gzip, pickle, math
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# DOMINANT-EXCLUDED: HOLDER-OPTIMIZED DENOMINATOR
# Retrospective feasibility study. No new simulation shots.
# ============================================================

SOURCES = [
    (0.003,
     "qec_noise_generalization_v1/p_0_003/calibration.pkl.gz",
     "qec_noise_generalization_v1/p_0_003/evaluation.pkl.gz"),
    (0.005,
     "qec_d5_alpha2_validation/counts.pkl.gz",
     "qec_d5_weight_prospective/evaluation_counts.pkl.gz"),
    (0.007,
     "qec_noise_generalization_v1/p_0_007/calibration.pkl.gz",
     "qec_d5_p007_prospective_v1/evaluation.pkl.gz"),
    (0.010,
     "qec_noise_generalization_v1/p_0_010/calibration.pkl.gz",
     "qec_noise_generalization_v1/p_0_010/evaluation.pkl.gz")
]

OUT = "qec_ablated_holder_pilot"
os.makedirs(OUT, exist_ok=True)

ALPHAS = [2, 3]
DELTA = 0.05
TARGET = 1e-3

EDGES = [0, 3, 5, 7, 9, 13, 21, 121]
BASE = [
    "cal_100_plus", "cal_20_to_99",
    "cal_5_to_19", "cal_2_to_4", "cal_1"
]
WEIGHT = [
    f"unseen_w{EDGES[i]}_to_{EDGES[i+1]-1}"
    for i in range(len(EDGES)-1)
]
GROUPS = BASE + WEIGHT

# Same allocation as preceding Bernstein pilot:
# 24 numerator events + 4 denominator events per p.
EVENTS_PER_P = 2*len(GROUPS) + 2*len(ALPHAS)
EPS = DELTA/(EVENTS_PER_P*len(SOURCES))

def cp_upper(k, n):
    if k >= n:
        return 1.0
    return float(beta.isf(EPS, k+1, n-k))

results = []

for p, cal_path, ev_path in SOURCES:
    print(f"Analyzing p={p:.3f}", flush=True)

    with gzip.open(cal_path, "rb") as f:
        cd = pickle.load(f)
    with gzip.open(ev_path, "rb") as f:
        ed = pickle.load(f)

    cal = cd["calibration"] if "calibration" in cd else cd["counts"]
    dominant = cd["dominant"]
    ev, failures = ed["counts"], ed["failures"]

    M = sum(cal.values())
    N = sum(ev.values())

    def classify(s):
        if s == dominant:
            return None

        c = cal.get(s, 0)
        if c >= 100: return "cal_100_plus"
        if c >= 20: return "cal_20_to_99"
        if c >= 5: return "cal_5_to_19"
        if c >= 2: return "cal_2_to_4"
        if c == 1: return "cal_1"

        w = int.from_bytes(s, "little").bit_count()
        for j in range(len(EDGES)-1):
            if EDGES[j] <= w < EDGES[j+1]:
                return WEIGHT[j]

        raise ValueError("Unclassified syndrome")

    stats = {
        g: {"failures":0, "max_count":0}
        for g in GROUPS
    }

    for s,c in ev.items():
        g = classify(s)
        if g is None:
            continue
        stats[g]["failures"] += failures.get(s, 0)
        stats[g]["max_count"] = max(
            stats[g]["max_count"], c
        )

    u = np.array([
        cp_upper(stats[g]["max_count"], N)
        for g in GROUPS
    ])
    b = np.array([
        cp_upper(stats[g]["failures"], N)
        for g in GROUPS
    ])

    q = {
        s: c/M for s,c in cal.items()
        if s != dominant
    }
    qmax = max(q.values(), default=0.0)

    for alpha in ALPHAS:
        power = alpha-1
        R = qmax**power

        mean = math.fsum(
            c*q.get(s,0.0)**power
            for s,c in ev.items()
        )/N

        second = math.fsum(
            c*q.get(s,0.0)**(2*power)
            for s,c in ev.items()
        )/N

        sample_var = max(
            0.0, N*(second-mean**2)/(N-1)
        )

        log_term = math.log(2/EPS)
        eb_radius = (
            math.sqrt(2*sample_var*log_term/N)
            + 7*R*log_term/(3*(N-1))
        )

        L = max(0.0, mean-eb_radius)

        B = math.fsum(
            value**alpha for value in q.values()
        )

        fixed_tangent = max(
            0.0,
            alpha*L-(alpha-1)*B
        )

        holder_denominator = (
            L**alpha / B**(alpha-1)
            if L > 0 and B > 0
            else 0.0
        )

        # Numerical check of the theoretical ordering.
        assert (
            holder_denominator + 1e-25
            >= fixed_tangent
        )

        numerator = float(np.sum(
            u**power*b
        ))

        def upper_ratio(denominator):
            if denominator <= 0:
                return 1.0
            return min(1.0, numerator/denominator)

        results.append({
            "p":p,
            "alpha":alpha,
            "shots":N,
            "mean":mean,
            "bernstein_radius":eb_radius,
            "expectation_lower":L,
            "B_alpha":B,
            "numerator_upper":numerator,
            "fixed_denominator":fixed_tangent,
            "holder_denominator":holder_denominator,
            "fixed_upper":upper_ratio(fixed_tangent),
            "holder_upper":upper_ratio(holder_denominator),
            "holder_below_target":
                upper_ratio(holder_denominator)<TARGET
        })

df = pd.DataFrame(results)

print("\nHOLDER-OPTIMIZED RESULTS")
print(df.to_string(index=False))

print("\nCOMPACT COMPARISON")
print(df[[
    "p", "alpha",
    "expectation_lower",
    "fixed_denominator",
    "holder_denominator",
    "fixed_upper",
    "holder_upper",
    "holder_below_target"
]].to_string(index=False))

df.to_csv(f"{OUT}/results.csv", index=False)

print("\nOne-sided events:", EVENTS_PER_P*len(SOURCES))
print("Per-event error allocation:", EPS)
print("Retrospective feasibility analysis only.")
```

```python

import os, gzip, pickle, json, hashlib, math
from collections import Counter
import numpy as np
import pandas as pd
import stim
import pymatching
from scipy.stats import beta

# ============================================================
# PROSPECTIVE DOMINANT-EXCLUDED CERTIFICATION
# d=5, p=0.005, alpha=3
# Holder denominator + empirical Bernstein + groupwise numerator
# ============================================================

CAL_PATH = "qec_d5_alpha2_validation/counts.pkl.gz"
OUT = "qec_d5_ablated_prospective_v1"

D, P, ALPHA = 5, 0.005, 3
N, CHUNK = 2_000_000, 20_000
SEED = 20261231
DELTA, TARGET = 0.05, 1e-3

EDGES = [0, 3, 5, 7, 9, 13, 21, 121]
BASE = [
    "cal_100_plus", "cal_20_to_99",
    "cal_5_to_19", "cal_2_to_4", "cal_1"
]
WEIGHT = [
    f"unseen_w{EDGES[i]}_to_{EDGES[i+1]-1}"
    for i in range(len(EDGES)-1)
]
GROUPS = BASE + WEIGHT

# 12 atom upper + 12 failure-mass upper
# + 1 empirical Bernstein expectation lower.
EVENTS = 2*len(GROUPS)+1
EPS = DELTA/EVENTS

os.makedirs(OUT, exist_ok=True)

with gzip.open(CAL_PATH, "rb") as f:
    archive = pickle.load(f)

cal = archive["calibration"]
dominant = archive["dominant"]

M = sum(cal.values())
assert M == 1_000_000

# q is fixed using independent calibration.
q = {
    s: c/M for s,c in cal.items()
    if s != dominant
}

qmax = max(q.values())
B = math.fsum(v**ALPHA for v in q.values())

def classify(s):
    if s == dominant:
        return None

    c = cal.get(s, 0)
    if c >= 100: return "cal_100_plus"
    if c >= 20: return "cal_20_to_99"
    if c >= 5: return "cal_5_to_19"
    if c >= 2: return "cal_2_to_4"
    if c == 1: return "cal_1"

    w = int.from_bytes(s, "little").bit_count()

    for j in range(len(EDGES)-1):
        if EDGES[j] <= w < EDGES[j+1]:
            return WEIGHT[j]

    raise ValueError("Unclassified record")

def cp_upper(k):
    if k == N:
        return 1.0
    return float(beta.isf(EPS, k+1, N-k))

circuit = stim.Circuit.generated(
    "surface_code:rotated_memory_x",
    distance=D, rounds=D,
    after_clifford_depolarization=P,
    before_round_data_depolarization=P,
    after_reset_flip_probability=P,
    before_measure_flip_probability=P
)

# Freeze analysis configuration before generating evaluation shots.
config = {
    "distance":D, "p":P, "alpha":ALPHA,
    "shots":N, "chunk":CHUNK, "seed":SEED,
    "delta":DELTA, "target":TARGET,
    "groups":GROUPS, "weight_edges":EDGES,
    "eps":EPS, "events":EVENTS,
    "calibration_path":CAL_PATH,
    "dominant_hex":dominant.hex(),
    "qmax":qmax, "B_alpha":B,
    "method":"holder_eb_groupwise_v1",
    "stim_version":stim.__version__,
    "pymatching_version":getattr(
        pymatching,"__version__","unknown"
    ),
    "circuit_sha256":hashlib.sha256(
        str(circuit).encode()
    ).hexdigest()
}
with open(f"{OUT}/frozen_config.json","w") as f:
    json.dump(config,f,indent=2)

decoder = pymatching.Matching.from_detector_error_model(
    circuit.detector_error_model(decompose_errors=True)
)
sampler = circuit.compile_detector_sampler(seed=SEED)

counts = Counter()
failures = Counter()
total_failed = 0

for start in range(0,N,CHUNK):
    n = min(CHUNK,N-start)
    det,obs = sampler.sample(
        shots=n,separate_observables=True
    )

    predicted = decoder.decode_batch(det)
    bad = np.any(predicted != obs,axis=1)

    packed = np.packbits(
        det,axis=1,bitorder="little"
    )

    counts.update(map(bytes,packed))
    failures.update(map(bytes,packed[bad]))
    total_failed += int(bad.sum())

    done = start+n
    if done % 200_000 == 0 or done == N:
        print(f"Evaluation: {done:,}/{N:,}",flush=True)

assert sum(counts.values()) == N
assert sum(failures.values()) == total_failed

# ============================================================
# 1. Calibration-weighted evaluation moments
# ============================================================

power = ALPHA-1
R = qmax**power

mean = math.fsum(
    c*q.get(s,0.0)**power
    for s,c in counts.items()
)/N

second = math.fsum(
    c*q.get(s,0.0)**(2*power)
    for s,c in counts.items()
)/N

sample_var = max(
    0.0,N*(second-mean**2)/(N-1)
)

log_term = math.log(2/EPS)
radius = (
    math.sqrt(2*sample_var*log_term/N)
    + 7*R*log_term/(3*(N-1))
)

L = max(0.0,mean-radius)

denominator_lower = (
    L**ALPHA/B**(ALPHA-1)
    if L>0 and B>0 else 0.0
)

# ============================================================
# 2. Groupwise failure numerator upper bound
# ============================================================

stats = {
    g: {"shots":0,"failures":0,"max_count":0}
    for g in GROUPS
}

for s,c in counts.items():
    g = classify(s)
    if g is None:
        continue

    x = stats[g]
    x["shots"] += c
    x["failures"] += failures.get(s,0)
    x["max_count"] = max(x["max_count"],c)

assert sum(x["shots"] for x in stats.values()) == (
    N-counts.get(dominant,0)
)

assert sum(x["failures"] for x in stats.values()) == (
    total_failed-failures.get(dominant,0)
)

rows = []
for g in GROUPS:
    x = stats[g]
    u = cp_upper(x["max_count"])
    b = cp_upper(x["failures"])

    rows.append({
        "group":g,
        **x,
        "max_p_upper":u,
        "failure_mass_upper":b,
        "F3_upper_contribution":u**power*b
    })

group_df = pd.DataFrame(rows)

numerator_upper = float(
    group_df["F3_upper_contribution"].sum()
)

confidence_upper = (
    min(1.0,numerator_upper/denominator_lower)
    if denominator_lower>0 else 1.0
)

summary = pd.DataFrame([{
    "distance":D,
    "p":P,
    "alpha":ALPHA,
    "evaluation_shots":N,
    "baseline_LER":total_failed/N,
    "dominant_count":counts.get(dominant,0),
    "dominant_failures":failures.get(dominant,0),
    "mean":mean,
    "sample_variance":sample_var,
    "eb_radius":radius,
    "holder_denominator_lower":denominator_lower,
    "numerator_upper":numerator_upper,
    "confidence_upper":confidence_upper,
    "target":TARGET,
    "certified":confidence_upper<TARGET
}])

print("\nPROSPECTIVE ABLATED CERTIFICATION")
print(summary.to_string(index=False))

print("\nGROUPWISE NUMERATOR CONTRIBUTIONS")
print(group_df.to_string(index=False))

print("\nConfidence level: at least 95%")
print("Confidence events:",EVENTS)
print("Per-event delta:",EPS)

summary.to_csv(f"{OUT}/results.csv",index=False)
group_df.to_csv(f"{OUT}/groups.csv",index=False)

with gzip.open(f"{OUT}/evaluation.pkl.gz","wb") as f:
    pickle.dump({
        "counts":counts,
        "failures":failures,
        "shots":N,
        "seed":SEED,
        "total_failures":total_failed
    },f,protocol=pickle.HIGHEST_PROTOCOL)

print("\nSaved to:",OUT)
```

```python

import gzip, pickle, json, math
import numpy as np
import pandas as pd
from scipy.stats import beta

ROOT = "qec_d5_ablated_prospective_v1"

with open(f"{ROOT}/frozen_config.json") as f:
    cfg = json.load(f)

with gzip.open(cfg["calibration_path"], "rb") as f:
    cd = pickle.load(f)

with gzip.open(f"{ROOT}/evaluation.pkl.gz", "rb") as f:
    ed = pickle.load(f)

saved = pd.read_csv(f"{ROOT}/results.csv").iloc[0]
saved_groups = pd.read_csv(f"{ROOT}/groups.csv")

cal = cd["calibration"]
s0 = cd["dominant"]
ev = ed["counts"]
fail = ed["failures"]

N = cfg["shots"]
alpha = cfg["alpha"]
eps = cfg["eps"]

assert sum(ev.values()) == N
assert sum(fail.values()) == ed["total_failures"]
assert ed["seed"] == cfg["seed"]
assert s0.hex() == cfg["dominant_hex"]
assert cfg["events"] == 25
assert len(cfg["groups"]) == 12

M = sum(cal.values())
q = {s: c/M for s,c in cal.items() if s != s0}

B = math.fsum(v**alpha for v in q.values())
R = max(q.values())**(alpha-1)

mean = math.fsum(
    c*q.get(s,0)**(alpha-1)
    for s,c in ev.items()
)/N

second = math.fsum(
    c*q.get(s,0)**(2*(alpha-1))
    for s,c in ev.items()
)/N

var = max(0, N*(second-mean**2)/(N-1))

radius = (
    math.sqrt(2*var*math.log(2/eps)/N)
    + 7*R*math.log(2/eps)/(3*(N-1))
)

L = max(0, mean-radius)
denominator = L**alpha/B**(alpha-1) if B > 0 else 0

edges = cfg["weight_edges"]

def classify(s):
    if s == s0:
        return None
    c = cal.get(s,0)
    if c >= 100: return "cal_100_plus"
    if c >= 20: return "cal_20_to_99"
    if c >= 5: return "cal_5_to_19"
    if c >= 2: return "cal_2_to_4"
    if c == 1: return "cal_1"
    w = int.from_bytes(s,"little").bit_count()
    for i in range(len(edges)-1):
        if edges[i] <= w < edges[i+1]:
            return f"unseen_w{edges[i]}_to_{edges[i+1]-1}"
    raise AssertionError("Unclassified syndrome")

stats = {
    g: {"shots":0,"failures":0,"max_count":0}
    for g in cfg["groups"]
}

for s,c in ev.items():
    g = classify(s)
    if g is None: continue
    x = stats[g]
    x["shots"] += c
    x["failures"] += fail.get(s,0)
    x["max_count"] = max(x["max_count"],c)

def cp_hi(k):
    return 1.0 if k == N else float(
        beta.isf(eps,k+1,N-k)
    )

numerator = math.fsum(
    cp_hi(stats[g]["max_count"])**(alpha-1)
    *cp_hi(stats[g]["failures"])
    for g in cfg["groups"]
)

bound = min(1.0,numerator/denominator) if denominator>0 else 1.0

checks = {
    "group shot partition":
        sum(x["shots"] for x in stats.values()) ==
        N-ev.get(s0,0),
    "group failure partition":
        sum(x["failures"] for x in stats.values()) ==
        sum(fail.values())-fail.get(s0,0),
    "weighted mean":
        np.isclose(mean,saved["mean"],rtol=1e-8),
    "sample variance":
        np.isclose(var,saved["sample_variance"],
                   rtol=1e-8,atol=1e-25),
    "EB radius":
        np.isclose(radius,saved["eb_radius"],rtol=1e-8),
    "Hölder denominator":
        np.isclose(denominator,
            saved["holder_denominator_lower"],rtol=1e-8),
    "numerator":
        np.isclose(numerator,saved["numerator_upper"],rtol=1e-8),
    "final bound":
        np.isclose(bound,saved["confidence_upper"],rtol=1e-8),
    "certification":
        (bound < cfg["target"]) == bool(saved["certified"]),
    "group counts":
        all(
            stats[g][key] ==
            int(saved_groups.set_index("group").loc[g,key])
            for g in cfg["groups"]
            for key in ["shots","failures","max_count"]
        )
}

print("INDEPENDENT ABLATION AUDIT")
for name, ok in checks.items():
    print(f"{'PASS' if ok else 'FAIL'} | {name}")

print(f"\nChecks: {sum(checks.values())}/{len(checks)}")
print(f"Numerator: {numerator:.12e}")
print(f"Denominator: {denominator:.12e}")
print(f"Upper bound: {bound:.12e}")
print(f"Certified: {bound < cfg['target']}")

assert all(checks.values()), "AUDIT FAILED"
print("\nAUDIT PASSED")
```

```python
import os
import zipfile
from pathlib import Path
from google.colab import files

# Define the zip file name
zip_filename = 'qec_project_export.zip'

# List of directories and files to exclude from the export (Colab system directories)
exclude_paths = {
    '.config',
    'sample_data',
    'original_paper_reproduction/authors_code',
    '__pycache__',
    zip_filename
}

def should_exclude(path: Path) -> bool:
    for exp in exclude_paths:
        if path.match(exp) or any(parent.match(exp) for parent in path.parents):
            return True
    return False

print("Packaging files...\n")

with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for path in Path('.').rglob('*'):
        if path.is_file() and not should_exclude(path):
            print(f"Adding: {path}")
            zipf.write(path)

print(f"\nCreated successfully: {zip_filename} ({os.path.getsize(zip_filename) / 1048576:.3f} MB)")

# Trigger browser download of the produced ZIP
files.download(zip_filename)
```

```python

import os, json, gzip, pickle, hashlib, math, gc
from collections import Counter
import numpy as np
import pandas as pd
import stim
import pymatching
from scipy.stats import beta

# ============================================================
# CROSS-DISTANCE DOMINANT-EXCLUDED CERTIFICATION
# d = 3, 7 | p = 0.005 | alpha = 3
# Independent calibration and evaluation
# ============================================================

OUT = "qec_cross_distance_ablated_v1"
os.makedirs(OUT, exist_ok=True)

DISTANCES = [3, 7]
P = 0.005
ALPHA = 3

CAL_SHOTS = 1_000_000
EVAL_SHOTS = 1_000_000
CHUNK = 20_000

DELTA = 0.05
TARGET = 1e-3

# The final interval includes all remaining weights.
EDGES = [0, 3, 5, 7, 9, 13, 21]

BASE = [
    "cal_100_plus",
    "cal_20_to_99",
    "cal_5_to_19",
    "cal_2_to_4",
    "cal_1"
]

WEIGHT = [
    "unseen_w0_to_2",
    "unseen_w3_to_4",
    "unseen_w5_to_6",
    "unseen_w7_to_8",
    "unseen_w9_to_12",
    "unseen_w13_to_20",
    "unseen_w21_plus"
]

GROUPS = BASE + WEIGHT

# Per distance:
# 12 maximum-atom upper events
# 12 failure-mass upper events
# 1 weighted-expectation lower event
EVENTS_PER_D = 2 * len(GROUPS) + 1
TOTAL_EVENTS = EVENTS_PER_D * len(DISTANCES)
EPS = DELTA / TOTAL_EVENTS

EXPERIMENTS = [
    {
        "distance": d,
        "cal_seed": 20270100 + 10*i + 1,
        "eval_seed": 20270100 + 10*i + 2
    }
    for i, d in enumerate(DISTANCES)
]

def build_circuit(d):
    return stim.Circuit.generated(
        "surface_code:rotated_memory_x",
        distance=d,
        rounds=d,
        after_clifford_depolarization=P,
        before_round_data_depolarization=P,
        after_reset_flip_probability=P,
        before_measure_flip_probability=P
    )

def cp_upper(k, n):
    if k == n:
        return 1.0
    return float(beta.isf(EPS, k+1, n-k))

# Freeze all choices before simulation.
for exp in EXPERIMENTS:
    circuit = build_circuit(exp["distance"])
    exp["circuit_sha256"] = hashlib.sha256(
        str(circuit).encode()
    ).hexdigest()
    exp["detectors"] = circuit.num_detectors

config = {
    "experiment": "cross_distance_ablated_v1",
    "distances": DISTANCES,
    "physical_error_probability": P,
    "alpha": ALPHA,
    "calibration_shots": CAL_SHOTS,
    "evaluation_shots": EVAL_SHOTS,
    "chunk_size": CHUNK,
    "delta": DELTA,
    "target": TARGET,
    "groups": GROUPS,
    "weight_edges": EDGES,
    "events_per_distance": EVENTS_PER_D,
    "total_events": TOTAL_EVENTS,
    "eps": EPS,
    "experiments": EXPERIMENTS,
    "method": "holder_empirical_bernstein_groupwise",
    "stim_version": stim.__version__,
    "pymatching_version": getattr(
        pymatching, "__version__", "unknown"
    )
}

with open(f"{OUT}/frozen_config.json", "w") as f:
    json.dump(config, f, indent=2)

def collect(circuit, decoder, shots, seed, label):
    sampler = circuit.compile_detector_sampler(seed=seed)
    counts = Counter()
    failures = Counter()
    total_failed = 0

    for start in range(0, shots, CHUNK):
        n = min(CHUNK, shots-start)

        detectors, observables = sampler.sample(
            shots=n,
            separate_observables=True
        )

        predictions = decoder.decode_batch(detectors)

        bad = np.any(
            predictions != observables,
            axis=1
        )

        packed = np.packbits(
            detectors,
            axis=1,
            bitorder="little"
        )

        counts.update(map(bytes, packed))
        failures.update(map(bytes, packed[bad]))

        total_failed += int(bad.sum())

        done = start+n

        if done % 200_000 == 0 or done == shots:
            print(
                f"{label}: {done:,}/{shots:,}",
                flush=True
            )

    assert sum(counts.values()) == shots
    assert sum(failures.values()) == total_failed

    return counts, failures, total_failed

all_results = []
all_groups = []

for exp in EXPERIMENTS:
    d = exp["distance"]
    folder = f"{OUT}/distance_{d}"
    os.makedirs(folder, exist_ok=True)

    print("\n" + "="*60)
    print(f"DISTANCE {d} | p={P} | alpha={ALPHA}")
    print("="*60)

    circuit = build_circuit(d)

    decoder = pymatching.Matching.from_detector_error_model(
        circuit.detector_error_model(
            decompose_errors=True
        )
    )

    # --------------------------------------------------------
    # 1. Independent calibration
    # --------------------------------------------------------

    cal, cal_fail, cal_total = collect(
        circuit, decoder,
        CAL_SHOTS,
        exp["cal_seed"],
        f"d={d} calibration"
    )

    dominant = cal.most_common(1)[0][0]
    M = sum(cal.values())

    # Calibration weights exclude the dominant record.
    q = {
        s: c/M
        for s, c in cal.items()
        if s != dominant
    }

    qmax = max(q.values(), default=0.0)

    B = math.fsum(
        value**ALPHA for value in q.values()
    )

    with gzip.open(
        f"{folder}/calibration.pkl.gz", "wb"
    ) as f:
        pickle.dump({
            "counts": cal,
            "total_failures": cal_total,
            "shots": M,
            "seed": exp["cal_seed"],
            "dominant": dominant
        }, f, protocol=pickle.HIGHEST_PROTOCOL)

    def classify(s):
        if s == dominant:
            return None

        c = cal.get(s, 0)

        if c >= 100:
            return "cal_100_plus"
        if c >= 20:
            return "cal_20_to_99"
        if c >= 5:
            return "cal_5_to_19"
        if c >= 2:
            return "cal_2_to_4"
        if c == 1:
            return "cal_1"

        w = int.from_bytes(
            s, "little"
        ).bit_count()

        if w < 3: return "unseen_w0_to_2"
        if w < 5: return "unseen_w3_to_4"
        if w < 7: return "unseen_w5_to_6"
        if w < 9: return "unseen_w7_to_8"
        if w < 13: return "unseen_w9_to_12"
        if w < 21: return "unseen_w13_to_20"
        return "unseen_w21_plus"

    print(
        f"\nCalibration unique records: {len(cal):,}"
    )
    print(
        f"Dominant calibration count: "
        f"{cal[dominant]:,}"
    )
    print(f"Non-dominant qmax: {qmax:.8e}")

    # --------------------------------------------------------
    # 2. Fresh independent evaluation
    # --------------------------------------------------------

    ev, fail, total_failed = collect(
        circuit, decoder,
        EVAL_SHOTS,
        exp["eval_seed"],
        f"d={d} evaluation"
    )

    N = sum(ev.values())

    with gzip.open(
        f"{folder}/evaluation.pkl.gz", "wb"
    ) as f:
        pickle.dump({
            "counts": ev,
            "failures": fail,
            "total_failures": total_failed,
            "shots": N,
            "seed": exp["eval_seed"]
        }, f, protocol=pickle.HIGHEST_PROTOCOL)

    # --------------------------------------------------------
    # 3. Hölder + empirical Bernstein denominator
    # --------------------------------------------------------

    power = ALPHA-1
    R = qmax**power

    mean = math.fsum(
        c*q.get(s, 0.0)**power
        for s, c in ev.items()
    ) / N

    second = math.fsum(
        c*q.get(s, 0.0)**(2*power)
        for s, c in ev.items()
    ) / N

    sample_variance = max(
        0.0,
        N*(second-mean**2)/(N-1)
    )

    log_term = math.log(2/EPS)

    radius = (
        math.sqrt(
            2*sample_variance*log_term/N
        )
        + 7*R*log_term/(3*(N-1))
    )

    expectation_lower = max(
        0.0, mean-radius
    )

    denominator_lower = (
        expectation_lower**ALPHA / B**power
        if expectation_lower > 0 and B > 0
        else 0.0
    )

    # --------------------------------------------------------
    # 4. Groupwise failure numerator
    # --------------------------------------------------------

    stats = {
        g: {
            "shots": 0,
            "failures": 0,
            "max_count": 0,
            "records_seen": 0
        }
        for g in GROUPS
    }

    for s, c in ev.items():
        g = classify(s)

        if g is None:
            continue

        x = stats[g]
        x["shots"] += c
        x["failures"] += fail.get(s, 0)
        x["max_count"] = max(x["max_count"], c)
        x["records_seen"] += 1

    assert sum(
        x["shots"] for x in stats.values()
    ) == N-ev.get(dominant, 0)

    assert sum(
        x["failures"] for x in stats.values()
    ) == total_failed-fail.get(dominant, 0)

    group_rows = []

    for g in GROUPS:
        x = stats[g]

        atom_upper = cp_upper(
            x["max_count"], N
        )

        failure_upper = cp_upper(
            x["failures"], N
        )

        group_rows.append({
            "distance": d,
            "group": g,
            **x,
            "max_p_upper": atom_upper,
            "failure_mass_upper": failure_upper,
            "F3_upper_contribution":
                atom_upper**power * failure_upper
        })

    group_df = pd.DataFrame(group_rows)

    numerator_upper = float(
        group_df["F3_upper_contribution"].sum()
    )

    confidence_upper = (
        min(1.0, numerator_upper/denominator_lower)
        if denominator_lower > 0
        else 1.0
    )

    # --------------------------------------------------------
    # 5. Diagnostics and final result
    # --------------------------------------------------------

    dominant_eval_count = ev.get(dominant, 0)

    result = {
        "distance": d,
        "p": P,
        "alpha": ALPHA,
        "calibration_shots": M,
        "evaluation_shots": N,
        "detectors": circuit.num_detectors,
        "baseline_LER": total_failed/N,
        "calibration_unique": len(cal),
        "evaluation_unique": len(ev),
        "dominant_cal_count": cal[dominant],
        "dominant_eval_count": dominant_eval_count,
        "dominant_eval_failures": fail.get(
            dominant, 0
        ),
        "weighted_mean": mean,
        "sample_variance": sample_variance,
        "eb_radius": radius,
        "expectation_lower": expectation_lower,
        "B_alpha": B,
        "holder_denominator_lower":
            denominator_lower,
        "numerator_upper": numerator_upper,
        "confidence_upper": confidence_upper,
        "certified": bool(confidence_upper < TARGET)
    }

    all_results.append(result)
    all_groups.append(group_df)

    pd.DataFrame([result]).to_csv(
        f"{folder}/results.csv",
        index=False
    )

    group_df.to_csv(
        f"{folder}/groups.csv",
        index=False
    )

    print("\nDISTANCE RESULT")
    print(
        pd.DataFrame([result]).to_string(
            index=False
        )
    )

    del cal, cal_fail, ev, fail, q, stats
    del decoder, circuit
    gc.collect()

# ============================================================
# 6. Combined results
# ============================================================

results = pd.DataFrame(all_results)
groups = pd.concat(all_groups, ignore_index=True)

results.to_csv(
    f"{OUT}/cross_distance_results.csv",
    index=False
)
groups.to_csv(
    f"{OUT}/cross_distance_groups.csv",
    index=False
)

print("\n" + "="*70)
print("FINAL CROSS-DISTANCE RESULTS")
print("="*70)

print(results.to_string(index=False))

print("\nCOMPACT COMPARISON")
print(results[[
    "distance",
    "detectors",
    "baseline_LER",
    "calibration_unique",
    "dominant_eval_count",
    "holder_denominator_lower",
    "numerator_upper",
    "confidence_upper",
    "certified"
]].to_string(index=False))

print("\nJoint confidence: at least 95%")
print("Total one-sided events:", TOTAL_EVENTS)
print("Per-event error probability:", EPS)
print("\nSaved to:", OUT)
```

```python

import gzip, pickle, math
import numpy as np
import pandas as pd

ROOT = "qec_cross_distance_ablated_v1"
ALPHA = 3
rows = []
group_rows = []

for d in [3, 7]:
    folder = f"{ROOT}/distance_{d}"

    # Load only your own trusted pickle archives.
    with gzip.open(f"{folder}/calibration.pkl.gz", "rb") as f:
        cd = pickle.load(f)
    with gzip.open(f"{folder}/evaluation.pkl.gz", "rb") as f:
        ed = pickle.load(f)

    cal = cd["counts"]
    ev = ed["counts"]
    failures = ed["failures"]
    s0 = cd["dominant"]

    N = sum(ev.values())
    M = sum(cal.values())

    total_c = total_f = 0
    common_c = common_f = 0
    repeat_records = 0
    failure_records = 0
    records_in_both = 0
    evaluation_hits_in_cal = 0
    contributions = []

    for s, c in ev.items():
        if s == s0:
            continue

        e = failures.get(s, 0)

        if s in cal:
            records_in_both += 1
            evaluation_hits_in_cal += c

        if c < ALPHA:
            continue

        # Use integer arithmetic to avoid floating-point
        # roundoff in the collision tuple counts.
        w = math.prod(c-j for j in range(ALPHA))
        fw = (
            e * math.prod(c-1-j for j in range(ALPHA-1))
            if e else 0
        )

        total_c += w
        total_f += fw
        repeat_records += 1
        failure_records += int(fw > 0)

        if s in cal:
            common_c += w
            common_f += fw

        contributions.append((fw, w, s, c, e))

    contributions.sort(key=lambda x: (x[0], x[1]), reverse=True)

    for rank, (fw, w, s, c, e) in enumerate(
        contributions[:10], start=1
    ):
        group_rows.append({
            "distance": d,
            "rank": rank,
            "syndrome_hex": s.hex(),
            "detector_weight":
                int.from_bytes(s, "little").bit_count(),
            "evaluation_count": c,
            "evaluation_failures": e,
            "calibration_count": cal.get(s, 0),
            "triple_collision_tuples": w,
            "failure_weighted_tuples": fw
        })

    ratio = total_f/total_c if total_c else np.nan

    rows.append({
        "distance": d,
        "calibration_unique": len(cal),
        "evaluation_unique": len(ev),
        "shared_unique_records": records_in_both,
        "eval_shots_on_cal_records": evaluation_hits_in_cal,
        "eval_cal_coverage": evaluation_hits_in_cal/N,
        "non_dominant_triple_collision_tuples": total_c,
        "non_dominant_failure_weighted_tuples": total_f,
        "non_dominant_collision_records": repeat_records,
        "failure_contributing_records": failure_records,
        "exploratory_ablated_LER": ratio,
        "fraction_collision_weight_on_cal_records":
            common_c/total_c if total_c else np.nan,
        "fraction_failure_weight_on_cal_records":
            common_f/total_f if total_f else np.nan
    })

summary = pd.DataFrame(rows)
top = pd.DataFrame(group_rows)

print("\nCROSS-DISTANCE ABLATION DIAGNOSTICS")
print(summary.to_string(index=False))

print("\nTOP 10 FAILURE-WEIGHTED RECORDS PER DISTANCE")
print(top.to_string(index=False))

summary.to_csv(
    f"{ROOT}/ablation_diagnostics.csv", index=False
)
top.to_csv(
    f"{ROOT}/top_failure_records.csv", index=False
)

print("\nThese are descriptive U-statistic diagnostics.")
print("No new confidence certificates are claimed.")
```

```python

import gzip, pickle
import numpy as np
import pandas as pd
from scipy.stats import beta

ROOT = "qec_cross_distance_ablated_v1/distance_3"
ALPHA = 3
TARGET = 1e-3
DELTA = 0.05
KS = [2, 5, 10, 20]

with gzip.open(f"{ROOT}/calibration.pkl.gz", "rb") as f:
    cd = pickle.load(f)
with gzip.open(f"{ROOT}/evaluation.pkl.gz", "rb") as f:
    ed = pickle.load(f)

cal = cd["counts"]
ev = ed["counts"]
fail = ed["failures"]
s0 = cd["dominant"]
N = sum(ev.values())

EDGES = [0, 3, 5, 7, 9, 13, 21]
GROUPS = [
    "cal_100_plus", "cal_20_to_99",
    "cal_5_to_19", "cal_2_to_4", "cal_1",
    "unseen_w0_to_2", "unseen_w3_to_4",
    "unseen_w5_to_6", "unseen_w7_to_8",
    "unseen_w9_to_12", "unseen_w13_to_20",
    "unseen_w21_plus"
]

# 2 events for each of 20 selected syndromes:
# P(S=s) and P(S=s,L=1) lower bounds.
# 2 events per denominator group:
# P(S in group) and max individual atom upper.
EVENTS = 2*max(KS) + 2*len(GROUPS)
EPS = DELTA/EVENTS

def cp_lower(k):
    return (
        0.0 if k == 0
        else float(beta.ppf(EPS, k, N-k+1))
    )

def cp_upper(k):
    return (
        1.0 if k == N
        else float(beta.isf(EPS, k+1, N-k))
    )

def group_of(s):
    if s == s0:
        return None

    c = cal.get(s, 0)
    if c >= 100: return "cal_100_plus"
    if c >= 20: return "cal_20_to_99"
    if c >= 5: return "cal_5_to_19"
    if c >= 2: return "cal_2_to_4"
    if c == 1: return "cal_1"

    w = int.from_bytes(s, "little").bit_count()
    if w < 3: return "unseen_w0_to_2"
    if w < 5: return "unseen_w3_to_4"
    if w < 7: return "unseen_w5_to_6"
    if w < 9: return "unseen_w7_to_8"
    if w < 13: return "unseen_w9_to_12"
    if w < 21: return "unseen_w13_to_20"
    return "unseen_w21_plus"

# Denominator upper bound:
# sum P(s)^3 <= sum_g max_P(g)^2 * P(S in g)
stats = {
    g: {"shots":0, "max_count":0}
    for g in GROUPS
}

for s, c in ev.items():
    g = group_of(s)
    if g is None:
        continue
    stats[g]["shots"] += c
    stats[g]["max_count"] = max(
        stats[g]["max_count"], c
    )

denom_terms = []
for g in GROUPS:
    x = stats[g]
    max_atom_upper = cp_upper(x["max_count"])
    mass_upper = cp_upper(x["shots"])

    denom_terms.append({
        "group":g,
        "mass_upper":mass_upper,
        "max_atom_upper":max_atom_upper,
        "C3_upper_term":
            max_atom_upper**2 * mass_upper
    })

denom_df = pd.DataFrame(denom_terms)
C_upper = float(denom_df["C3_upper_term"].sum())

# Numerator lower bound:
# P(s)^2 P(S=s,L=1) for selected syndromes.
selected = [
    s for s, _ in cal.most_common(max(KS)+1)
    if s != s0
][:max(KS)]

assert len(selected) == max(KS)

rows = []
running_F = 0.0

for i, s in enumerate(selected, 1):
    p_lower = cp_lower(ev.get(s, 0))
    joint_failure_lower = cp_lower(fail.get(s, 0))

    term = p_lower**2 * joint_failure_lower
    running_F += term

    if i in KS:
        lower_ratio = (
            running_F/C_upper if C_upper > 0
            else 0.0
        )

        rows.append({
            "K":i,
            "selected_failure_lower":running_F,
            "residual_C_upper":C_upper,
            "ablated_LER_lower":lower_ratio,
            "certified_above_target":
                lower_ratio > TARGET
        })

results = pd.DataFrame(rows)

print("REVERSE CERTIFICATION PILOT")
print(results.to_string(index=False))

print("\nDENOMINATOR UPPER CONTRIBUTIONS")
print(denom_df.to_string(index=False))

print("\nConfidence events:", EVENTS)
print("Per-event delta:", EPS)
print("Retrospective analysis: not prospective validation")

results.to_csv(
    f"{ROOT}/reverse_certification_pilot.csv",
    index=False
)
denom_df.to_csv(
    f"{ROOT}/reverse_denominator_groups.csv",
    index=False
)
```

```python

import gzip, pickle
import math
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# DISTANCE-3 REVERSE CERTIFICATION PILOT V2
# Rank-stratified denominator; fixed top-20 numerator.
# Retrospective method development, no new shots.
# ============================================================

ROOT = "qec_cross_distance_ablated_v1/distance_3"

with gzip.open(f"{ROOT}/calibration.pkl.gz", "rb") as f:
    cd = pickle.load(f)
with gzip.open(f"{ROOT}/evaluation.pkl.gz", "rb") as f:
    ed = pickle.load(f)

cal = cd["counts"]
ev = ed["counts"]
fail = ed["failures"]
s0 = cd["dominant"]

N = sum(ev.values())
ALPHA = 3
TARGET = 1e-3
DELTA = 0.05
K = 20

# Ranking is based ONLY on calibration.
ranking = [s for s, _ in cal.most_common() if s != s0]
rank = {s: i+1 for i, s in enumerate(ranking)}

# Prespecified rank bands for this pilot.
RANK_BANDS = [
    (1, 1),
    (2, 5),
    (6, 20),
    (21, 100),
    (101, 500),
    (501, 2000),
    (2001, None)
]

def group_of(s):
    if s == s0:
        return None

    r = rank.get(s)

    if r is not None:
        for lo, hi in RANK_BANDS:
            if hi is None and r >= lo:
                return f"rank_{lo}_plus"
            if hi is not None and lo <= r <= hi:
                return f"rank_{lo}_to_{hi}"

    # Calibration-unseen records.
    weight = int.from_bytes(s, "little").bit_count()

    if weight < 3: return "unseen_w0_2"
    if weight < 5: return "unseen_w3_4"
    if weight < 7: return "unseen_w5_6"
    if weight < 9: return "unseen_w7_8"
    if weight < 13: return "unseen_w9_12"
    if weight < 21: return "unseen_w13_20"
    return "unseen_w21_plus"

rank_groups = [
    f"rank_{lo}_plus" if hi is None
    else f"rank_{lo}_to_{hi}"
    for lo, hi in RANK_BANDS
]

unseen_groups = [
    "unseen_w0_2", "unseen_w3_4", "unseen_w5_6",
    "unseen_w7_8", "unseen_w9_12",
    "unseen_w13_20", "unseen_w21_plus"
]

GROUPS = rank_groups + unseen_groups

# Two denominator events per group:
# max atom probability and group mass.
# Two numerator events per selected record:
# P(s) and P(s, logical failure).
EVENTS = 2*len(GROUPS) + 2*K
EPS = DELTA/EVENTS

def cp_lower(k):
    if k == 0:
        return 0.0
    return float(beta.ppf(EPS, k, N-k+1))

def cp_upper(k):
    if k == N:
        return 1.0
    return float(beta.isf(EPS, k+1, N-k))

stats = {
    g: {"shots":0, "max_count":0, "failures":0}
    for g in GROUPS
}

for s, c in ev.items():
    g = group_of(s)
    if g is None:
        continue

    x = stats[g]
    x["shots"] += c
    x["max_count"] = max(x["max_count"], c)
    x["failures"] += fail.get(s, 0)

assert sum(x["shots"] for x in stats.values()) == (
    N-ev.get(s0,0)
)
assert sum(x["failures"] for x in stats.values()) == (
    sum(fail.values())-fail.get(s0,0)
)

# ============================================================
# 1. Upper bound residual collision denominator
# ============================================================

denom_rows = []

for g in GROUPS:
    x = stats[g]
    u = cp_upper(x["max_count"])
    m = cp_upper(x["shots"])

    denom_rows.append({
        "group":g,
        **x,
        "max_atom_upper":u,
        "group_mass_upper":m,
        "C3_upper_term":u*u*m
    })

denom_df = pd.DataFrame(denom_rows)
C_upper = float(denom_df["C3_upper_term"].sum())

# ============================================================
# 2. Lower bound failure numerator, top 20
# ============================================================

num_rows = []
F_lower = 0.0

for i, s in enumerate(ranking[:K], start=1):
    c = ev.get(s, 0)
    e = fail.get(s, 0)

    p_lo = cp_lower(c)
    joint_lo = cp_lower(e)

    term = p_lo*p_lo*joint_lo
    F_lower += term

    num_rows.append({
        "rank":i,
        "calibration_count":cal[s],
        "evaluation_count":c,
        "logical_failures":e,
        "F3_lower_term":term
    })

num_df = pd.DataFrame(num_rows)

ratio_lower = (
    F_lower/C_upper if C_upper > 0 else 0.0
)

# Empirical collision moment for DIAGNOSTICS only.
C_tuples = sum(
    c*(c-1)*(c-2)
    for s,c in ev.items()
    if s != s0 and c >= 3
)

N_falling = N*(N-1)*(N-2)
C_empirical = C_tuples/N_falling

print("\nRANK-STRATIFIED DENOMINATOR")
print(denom_df.to_string(index=False))

print("\nTOP-20 FAILURE NUMERATOR")
print(num_df.to_string(index=False))

print("\nREVERSE CERTIFICATION COMPARISON")
summary = pd.DataFrame([{
    "distance":3,
    "alpha":3,
    "p":0.005,
    "numerator_lower":F_lower,
    "denominator_upper":C_upper,
    "empirical_denominator":C_empirical,
    "denominator_pessimism_factor":
        C_upper/C_empirical if C_empirical else np.nan,
    "ratio_lower":ratio_lower,
    "target":TARGET,
    "above_target":ratio_lower>TARGET
}])

print(summary.to_string(index=False))

print("\nPrevious reverse lower bound: 0.0003821481")
print("Confidence events:", EVENTS)
print("Per-event delta:", EPS)
print("Retrospective method-development result.")

denom_df.to_csv(
    f"{ROOT}/rank_stratified_denominator.csv",
    index=False
)
num_df.to_csv(
    f"{ROOT}/top20_failure_lower.csv",
    index=False
)
summary.to_csv(
    f"{ROOT}/reverse_certification_v2.csv",
    index=False
)
```

```python

import os, gzip, pickle, math, time, json
from collections import Counter
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# CONTROLLED FINITE-SAMPLE CERTIFICATION BENCHMARK
#
# d=5, p=0.005, alpha=3, dominant excluded
# 1M calibration + 2M independent evaluation shots
#
# A: Holder + empirical Bernstein
# B: Fixed tangent + Hoeffding
# C: Top-1000 atomic Clopper-Pearson
#
# Same dataset, numerator method, nominal confidence.
# Retrospective comparison. No simulation.
# ============================================================

CAL_PATH = "qec_d5_alpha2_validation/counts.pkl.gz"
EVAL_PATH = "qec_d5_ablated_prospective_v1/evaluation.pkl.gz"

OUT = "qec_statistical_baseline_benchmark_v1"
os.makedirs(OUT, exist_ok=True)

ALPHA = 3
DELTA = 0.05
TARGET = 1e-3
K = 1000

with gzip.open(CAL_PATH, "rb") as f:
    cd = pickle.load(f)

with gzip.open(EVAL_PATH, "rb") as f:
    ed = pickle.load(f)

cal = cd["calibration"]
s0 = cd["dominant"]

ev = ed["counts"]
fail = ed["failures"]

M = sum(cal.values())
N = sum(ev.values())

assert M == 1_000_000
assert N == 2_000_000
assert sum(fail.values()) == ed["total_failures"]

BASE = [
    "cal_100_plus",
    "cal_20_to_99",
    "cal_5_to_19",
    "cal_2_to_4",
    "cal_1"
]

WEIGHT = [
    "unseen_w0_to_2",
    "unseen_w3_to_4",
    "unseen_w5_to_6",
    "unseen_w7_to_8",
    "unseen_w9_to_12",
    "unseen_w13_to_20",
    "unseen_w21_plus"
]

GROUPS = BASE + WEIGHT
J = len(GROUPS)

def classify(s):
    if s == s0:
        return None

    c = cal.get(s, 0)

    if c >= 100: return BASE[0]
    if c >= 20: return BASE[1]
    if c >= 5: return BASE[2]
    if c >= 2: return BASE[3]
    if c == 1: return BASE[4]

    w = int.from_bytes(s, "little").bit_count()

    if w < 3: return WEIGHT[0]
    if w < 5: return WEIGHT[1]
    if w < 7: return WEIGHT[2]
    if w < 9: return WEIGHT[3]
    if w < 13: return WEIGHT[4]
    if w < 21: return WEIGHT[5]
    return WEIGHT[6]

# ------------------------------------------------------------
# Shared preprocessing (not included in method runtime)
# ------------------------------------------------------------

stats = {
    g: {"max_count":0, "failures":0, "shots":0}
    for g in GROUPS
}

for s,c in ev.items():
    g = classify(s)
    if g is None:
        continue

    x = stats[g]
    x["max_count"] = max(x["max_count"], c)
    x["failures"] += fail.get(s,0)
    x["shots"] += c

assert sum(x["shots"] for x in stats.values()) == (
    N-ev.get(s0,0)
)

q = {
    s: c/M for s,c in cal.items()
    if s != s0
}

qmax = max(q.values())
R = qmax**(ALPHA-1)

B = math.fsum(v**ALPHA for v in q.values())

mean = math.fsum(
    c*q.get(s,0.0)**(ALPHA-1)
    for s,c in ev.items()
)/N

second = math.fsum(
    c*q.get(s,0.0)**(2*(ALPHA-1))
    for s,c in ev.items()
)/N

variance = max(
    0.0,
    N*(second-mean*mean)/(N-1)
)

top_atoms = [
    s for s,_ in cal.most_common()
    if s != s0
][:K]

assert len(top_atoms) == K

# ------------------------------------------------------------
# Confidence functions
# ------------------------------------------------------------

def cp_upper(k, epsilon):
    if k >= N:
        return 1.0
    return float(beta.isf(
        epsilon, k+1, N-k
    ))

def cp_lower(k, epsilon):
    if k <= 0:
        return 0.0
    return float(beta.ppf(
        epsilon, k, N-k+1
    ))

def shared_numerator(epsilon):
    return math.fsum(
        cp_upper(stats[g]["max_count"],epsilon)**(ALPHA-1)
        * cp_upper(stats[g]["failures"],epsilon)
        for g in GROUPS
    )

# Each method independently has nominal 95% coverage.
# This comparison does NOT claim simultaneous 95%
# coverage of all three method outputs.

def benchmark(method):
    t0 = time.perf_counter()

    if method == "A_Holder_EB":
        events = 2*J + 1

    elif method == "B_Tangent_Hoeffding":
        events = 2*J + 1

    elif method == "C_Atomic_CP_1000":
        events = 2*J + K

    else:
        raise ValueError(method)

    eps = DELTA/events

    numerator = shared_numerator(eps)

    if method == "A_Holder_EB":
        log_term = math.log(2/eps)

        radius = (
            math.sqrt(2*variance*log_term/N)
            + 7*R*log_term/(3*(N-1))
        )

        L = max(0.0,mean-radius)

        denominator = (
            L**ALPHA/B**(ALPHA-1)
            if L>0 and B>0 else 0.0
        )

    elif method == "B_Tangent_Hoeffding":
        radius = (
            R*math.sqrt(math.log(1/eps)/(2*N))
        )

        L = max(0.0,mean-radius)

        denominator = max(
            0.0,
            ALPHA*L-(ALPHA-1)*B
        )

    else:
        radius = float("nan")

        # Each selected syndrome receives its own
        # one-sided CP lower confidence bound.
        denominator = math.fsum(
            cp_lower(ev.get(s,0),eps)**ALPHA
            for s in top_atoms
        )

    upper = (
        min(1.0,numerator/denominator)
        if denominator>0 else 1.0
    )

    duration = time.perf_counter()-t0

    return {
        "method":method,
        "confidence_events":events,
        "per_event_delta":eps,
        "numerator_upper":numerator,
        "denominator_lower":denominator,
        "upper_bound":upper,
        "below_target":upper<TARGET,
        "runtime_seconds":duration
    }

methods = [
    "A_Holder_EB",
    "B_Tangent_Hoeffding",
    "C_Atomic_CP_1000"
]

results = pd.DataFrame([
    benchmark(m) for m in methods
])

# Expected agreement with the archived prospective
# certificate under the same confidence allocation.
reference = 5.728430958212e-4

assert np.isclose(
    results.loc[0,"upper_bound"],
    reference,
    rtol=1e-7
), "Proposed method did not reproduce audited result"

results["relative_to_proposed"] = (
    results["upper_bound"] /
    results.loc[0,"upper_bound"]
)

print("\nCONTROLLED CERTIFICATION BENCHMARK")
print(results.to_string(index=False))

print("\nCOMPACT COMPARISON")
print(results[[
    "method",
    "confidence_events",
    "numerator_upper",
    "denominator_lower",
    "upper_bound",
    "relative_to_proposed",
    "below_target",
    "runtime_seconds"
]].to_string(index=False))

print("\nDATASET")
print("Calibration:", M)
print("Evaluation:", N)
print("Distinct evaluation syndromes:", len(ev))
print("Dominant evaluation occurrences:",ev.get(s0,0))
print("Dominant failures:",fail.get(s0,0))

print("\nImportant:")
print("Each method has marginal 95% confidence.")
print("Results are NOT jointly covered at 95%.")
print("Runtime excludes data loading and preprocessing.")
print("Retrospective comparison, not prospective validation.")

results.to_csv(
    f"{OUT}/benchmark_results.csv",
    index=False
)

config = {
    "calibration_path":CAL_PATH,
    "evaluation_path":EVAL_PATH,
    "alpha":ALPHA,
    "delta":DELTA,
    "target":TARGET,
    "atomic_k":K,
    "groups":GROUPS,
    "confidence_scope":"95% marginal per method",
    "comparison_status":"retrospective"
}

with open(f"{OUT}/config.json","w") as f:
    json.dump(config,f,indent=2)

print("\nSaved to:",OUT)
```

```python

import gzip, pickle, math
import pandas as pd

CROSS = "qec_cross_distance_ablated_v1"
D5 = "qec_d5_ablated_prospective_v1"

ALPHA = 3
EPS = 0.002
RHO = 0.5

sources = [
    (3,
     f"{CROSS}/distance_3/calibration.pkl.gz",
     f"{CROSS}/distance_3/results.csv"),
    (5,
     "qec_d5_alpha2_validation/counts.pkl.gz",
     f"{D5}/results.csv"),
    (7,
     f"{CROSS}/distance_7/calibration.pkl.gz",
     f"{CROSS}/distance_7/results.csv")
]

rows = []

for d, cal_path, results_path in sources:
    with gzip.open(cal_path, "rb") as f:
        cd = pickle.load(f)

    res = pd.read_csv(results_path).iloc[0]

    counts = cd.get("calibration", cd.get("counts"))
    s0 = cd["dominant"]
    M = sum(counts.values())

    qmax = max(
        (c/M for s,c in counts.items() if s != s0),
        default=0.0
    )

    R = qmax**(ALPHA-1)

    mean_column = (
        "weighted_mean"
        if "weighted_mean" in res.index
        else "mean"
    )
    mean = float(res[mean_column])

    h_hat = mean/R if R > 0 else 0.0

    n_plugin = (
        2*math.log(2/EPS)/(RHO**2*h_hat**2)
        if h_hat > 0 else math.inf
    )

    # Existing EB diagnostic:
    eb = float(res["eb_radius"])
    eb_lower = max(0.0, mean-eb)
    h_lower = eb_lower/R if R > 0 else 0.0

    rows.append({
        "distance": d,
        "calibration_qmax": qmax,
        "observed_mean": mean,
        "empirical_h": h_hat,
        "EB_lower_h": h_lower,
        "evaluation_shots": int(res["evaluation_shots"]),
        "hoeffding_plugin_N": math.ceil(n_plugin)
             if math.isfinite(n_plugin) else math.inf,
        "denominator_positive":
            float(res["holder_denominator_lower"]) > 0
    })

df = pd.DataFrame(rows)

print("\nCALIBRATION-OVERLAP DIAGNOSTIC")
print(df.to_string(index=False))

print("\nAssumed relative tolerance rho:", RHO)
print("Confidence-event epsilon:", EPS)
print("Sample-size estimates are exploratory plug-ins.")
print("They are NOT proven necessary shot counts.")

df.to_csv(
    f"{CROSS}/sample_complexity_diagnostics.csv",
    index=False
)
```

```python

import gzip, pickle, math
import numpy as np
import pandas as pd

# ============================================================
# VARIANCE-SENSITIVE SAMPLE COMPLEXITY DIAGNOSTIC
# d = 3, 5, 7 | alpha = 3 | p = 0.005
# No new simulation; retrospective comparison
# ============================================================

SOURCES = [
    (3,
     "qec_cross_distance_ablated_v1/distance_3/calibration.pkl.gz",
     "qec_cross_distance_ablated_v1/distance_3/evaluation.pkl.gz",
     "qec_cross_distance_ablated_v1/distance_3/results.csv"),
    (5,
     "qec_d5_alpha2_validation/counts.pkl.gz",
     "qec_d5_ablated_prospective_v1/evaluation.pkl.gz",
     "qec_d5_ablated_prospective_v1/results.csv"),
    (7,
     "qec_cross_distance_ablated_v1/distance_7/calibration.pkl.gz",
     "qec_cross_distance_ablated_v1/distance_7/evaluation.pkl.gz",
     "qec_cross_distance_ablated_v1/distance_7/results.csv")
]

ALPHA = 3
EPS = 0.002
RHO = 0.5
t = math.log(2/EPS)

def bernstein_inverse_lower(observed_mean, n):
    a = t/n
    b = observed_mean - 2*a/3

    if b <= 0:
        return 0.0

    # Stable inversion of:
    # observed_mean = h + sqrt(2*a*h) + 2*a/3
    root = (
        2*b /
        (math.sqrt(2*a+4*b)+math.sqrt(2*a))
    )
    return root**2

rows = []

for d, cal_path, eval_path, res_path in SOURCES:
    with gzip.open(cal_path, "rb") as f:
        cd = pickle.load(f)

    with gzip.open(eval_path, "rb") as f:
        ed = pickle.load(f)

    res = pd.read_csv(res_path).iloc[0]

    cal = (
        cd["calibration"] if "calibration" in cd
        else cd["counts"]
    )
    ev = ed["counts"]
    s0 = cd["dominant"]

    M = sum(cal.values())
    N = sum(ev.values())

    q = {
        s:c/M for s,c in cal.items()
        if s != s0
    }

    qmax = max(q.values(),default=0)
    R = qmax**(ALPHA-1)

    B = math.fsum(
        v**ALPHA for v in q.values()
    )

    if R == 0 or B == 0:
        continue

    mean = math.fsum(
        c*q.get(s,0)**(ALPHA-1)
        for s,c in ev.items()
    )/N

    h_hat = mean/R

    # New population-variance Bernstein inversion.
    h_lower = bernstein_inverse_lower(h_hat,N)
    A_lower = R*h_lower

    C_lower = (
        A_lower**ALPHA / B**(ALPHA-1)
        if A_lower > 0 else 0.0
    )

    # Existing empirical Bernstein construction.
    saved_C_lower = float(
        res["holder_denominator_lower"]
    )

    # Plug-in planning figures, not validated budgets.
    N_hoeffding = (
        2*t/(RHO**2*h_hat**2)
        if h_hat>0 else math.inf
    )

    N_bernstein = (
        32*t/(RHO**2*h_hat)
        if h_hat>0 else math.inf
    )

    rows.append({
        "distance":d,
        "shots":N,
        "empirical_h":h_hat,
        "new_h_lower":h_lower,
        "new_denominator_lower":C_lower,
        "existing_EB_denominator":saved_C_lower,
        "new_denominator_positive":C_lower>0,
        "hoeffding_plugin_N":N_hoeffding,
        "bernstein_plugin_N":N_bernstein
    })

df = pd.DataFrame(rows)

print("\nVARIANCE-SENSITIVE DENOMINATOR COMPARISON")
print(df[[
    "distance",
    "shots",
    "empirical_h",
    "new_h_lower",
    "new_denominator_lower",
    "existing_EB_denominator",
    "new_denominator_positive"
]].to_string(index=False))

print("\nSAMPLE COMPLEXITY COMPARISON")
print(df[[
    "distance",
    "hoeffding_plugin_N",
    "bernstein_plugin_N"
]].to_string(index=False))

print("\nConfidence allocation for new bound:",EPS)
print("Relative tolerance:",RHO)
print("Results are retrospective.")
print("New and existing confidence bounds are")
print("not claimed jointly valid at 95%.")
print("Plug-in N values are not necessary shot counts.")

OUT = "qec_cross_distance_ablated_v1"
df.to_csv(
    f"{OUT}/variance_sensitive_theorem_diagnostic.csv",
    index=False
)
```

```python

import gzip, pickle, math, os
import numpy as np
import pandas as pd

ALPHA = 3
DELTA_CAL = 0.05
OUT = "qec_calibration_alignment_v1"
os.makedirs(OUT, exist_ok=True)

SOURCES = [
    (
        3,
        "qec_cross_distance_ablated_v1/distance_3/calibration.pkl.gz",
        "qec_cross_distance_ablated_v1/distance_3/evaluation.pkl.gz"
    ),
    (
        5,
        "qec_d5_alpha2_validation/counts.pkl.gz",
        "qec_d5_ablated_prospective_v1/evaluation.pkl.gz"
    ),
    (
        7,
        "qec_cross_distance_ablated_v1/distance_7/calibration.pkl.gz",
        "qec_cross_distance_ablated_v1/distance_7/evaluation.pkl.gz"
    )
]

rows = []

for d, cal_path, ev_path in SOURCES:
    print(f"Analyzing distance {d}")

    # Only load archives generated by your trusted notebook.
    with gzip.open(cal_path, "rb") as f:
        cd = pickle.load(f)

    with gzip.open(ev_path, "rb") as f:
        ed = pickle.load(f)

    cal = cd["calibration"] if "calibration" in cd else cd["counts"]
    ev = ed["counts"]
    s0 = cd["dominant"]

    M = sum(cal.values())
    N = sum(ev.values())

    q = {
        s: c/M
        for s,c in cal.items()
        if s != s0
    }

    B = math.fsum(v**ALPHA for v in q.values())
    qnorm = B**(1/ALPHA) if B>0 else 0.0

    # Rigorous calibration-error radius.
    error_radius = (
        1 + math.sqrt(math.log(1/DELTA_CAL))
    )/math.sqrt(M)

    calibration_eta_lower = (
        (max(0, qnorm-error_radius) /
         (qnorm+error_radius))**ALPHA
        if qnorm+error_radius>0
        else 0.0
    )

    # Independent evaluation-based exploratory A estimate.
    A_hat = math.fsum(
        c*q.get(s,0.0)**(ALPHA-1)
        for s,c in ev.items()
    )/N

    H_hat = (
        A_hat**ALPHA / B**(ALPHA-1)
        if B>0 else 0.0
    )

    # Unbiased collision U-statistic estimate of C.
    # This is NOT a confidence bound.
    collision_tuples = sum(
        c*(c-1)*(c-2)
        for s,c in ev.items()
        if s != s0 and c >= 3
    )

    N_falling = N*(N-1)*(N-2)
    C_hat = collision_tuples/N_falling

    eta_hat = (
        H_hat/C_hat if C_hat>0 else np.nan
    )

    rows.append({
        "distance":d,
        "calibration_shots":M,
        "evaluation_shots":N,
        "calibration_unique":len(cal),
        "q_l3_norm":qnorm,
        "calibration_error_radius":error_radius,
        "rigorous_eta_lower":calibration_eta_lower,
        "empirical_C3":C_hat,
        "empirical_H3":H_hat,
        "exploratory_eta":eta_hat,
        "triple_collision_tuples":collision_tuples
    })

df = pd.DataFrame(rows)

print("\nCALIBRATION ALIGNMENT RESULTS")
print(df.to_string(index=False))

print("\nCOMPACT COMPARISON")
print(df[[
    "distance",
    "q_l3_norm",
    "calibration_error_radius",
    "rigorous_eta_lower",
    "exploratory_eta",
    "triple_collision_tuples"
]].to_string(index=False))

print("\nCalibration confidence:",1-DELTA_CAL)
print("Each distance has a marginal calibration guarantee.")
print("Exploratory eta is not a confidence statement.")
print("An exploratory eta above 1 is possible due to sampling error.")
print("Undefined eta means the collision estimate was zero.")

df.to_csv(f"{OUT}/results.csv",index=False)
print("\nSaved to:",OUT)
```

```python

import os, gzip, pickle, math, json
from collections import defaultdict
import numpy as np
import pandas as pd
from scipy.stats import beta

# ============================================================
# PHASE 5.4: CALIBRATION-ALIGNMENT CERTIFICATION
#
# d=5, p=.005, alpha=3, dominant excluded
# 1M calibration + 2M existing evaluation shots
# No new simulation
# ============================================================

CAL_PATH = "qec_d5_alpha2_validation/counts.pkl.gz"
EVAL_PATH = "qec_d5_ablated_prospective_v1/evaluation.pkl.gz"
OUT = "qec_alignment_upper_moment_v1"
os.makedirs(OUT, exist_ok=True)

ALPHA = 3
DELTA = .05
TOP_K = 100

with gzip.open(CAL_PATH, "rb") as f:
    cd = pickle.load(f)
with gzip.open(EVAL_PATH, "rb") as f:
    ed = pickle.load(f)

cal = cd["calibration"]
s0 = cd["dominant"]
ev = ed["counts"]

M = sum(cal.values())
N = sum(ev.values())

assert M == 1_000_000
assert N == 2_000_000

# Calibration-only ranking.
ranking = [
    s for s, c in sorted(
        cal.items(),
        key=lambda item: (-item[1], item[0])
    ) if s != s0
]

rank = {s: i+1 for i, s in enumerate(ranking)}
top = ranking[:TOP_K]
top_set = set(top)

# All group definitions are calibration-dependent only.
FREQ_GROUPS = [
    "cal_100_plus", "cal_20_to_99", "cal_5_to_19",
    "cal_2_to_4", "cal_1",
    "unseen_w0_2", "unseen_w3_4", "unseen_w5_6",
    "unseen_w7_8", "unseen_w9_12",
    "unseen_w13_20", "unseen_w21_plus"
]

RANK_BANDS = [
    (1, 1), (2, 5), (6, 20),
    (21, 100), (101, 500),
    (501, 2000), (2001, None)
]

def rank_name(lo, hi):
    return (
        f"rank_{lo}_plus" if hi is None
        else f"rank_{lo}_to_{hi}"
    )

UNSEEN_GROUPS = [
    "unseen_w0_2", "unseen_w3_4", "unseen_w5_6",
    "unseen_w7_8", "unseen_w9_12",
    "unseen_w13_20", "unseen_w21_plus"
]

RANK_GROUPS = [
    rank_name(lo, hi)
    for lo, hi in RANK_BANDS
] + UNSEEN_GROUPS

HYBRID_GROUPS = [
    "rank_101_to_500",
    "rank_501_to_2000",
    "rank_2001_plus"
] + UNSEEN_GROUPS

def unseen_group(s):
    w = int.from_bytes(s, "little").bit_count()
    if w < 3: return "unseen_w0_2"
    if w < 5: return "unseen_w3_4"
    if w < 7: return "unseen_w5_6"
    if w < 9: return "unseen_w7_8"
    if w < 13: return "unseen_w9_12"
    if w < 21: return "unseen_w13_20"
    return "unseen_w21_plus"

def freq_group(s):
    c = cal.get(s, 0)
    if c >= 100: return "cal_100_plus"
    if c >= 20: return "cal_20_to_99"
    if c >= 5: return "cal_5_to_19"
    if c >= 2: return "cal_2_to_4"
    if c == 1: return "cal_1"
    return unseen_group(s)

def ranked_group(s):
    r = rank.get(s)
    if r is None:
        return unseen_group(s)
    for lo, hi in RANK_BANDS:
        if r >= lo and (hi is None or r <= hi):
            return rank_name(lo, hi)
    raise AssertionError("Invalid rank")

def hybrid_group(s):
    if s in top_set:
        return None
    r = rank.get(s)
    if r is None:
        return unseen_group(s)
    if r <= 500: return "rank_101_to_500"
    if r <= 2000: return "rank_501_to_2000"
    return "rank_2001_plus"

METHODS = {
    "frequency": (FREQ_GROUPS, freq_group),
    "rank": (RANK_GROUPS, ranked_group),
    "hybrid_top100": (HYBRID_GROUPS, hybrid_group)
}

# One EB expectation event, plus all group/atom events.
# All three candidate bounds are simultaneously covered.
EVENTS = (
    1 +
    2*len(FREQ_GROUPS) +
    2*len(RANK_GROUPS) +
    TOP_K + 2*len(HYBRID_GROUPS)
)
EPS = DELTA/EVENTS

def cp_upper(k):
    if k == N:
        return 1.0
    return float(beta.isf(EPS, k+1, N-k))

# ------------------------------------------------------------
# 1. Population Hölder numerator expectation lower bound
# ------------------------------------------------------------

q = {
    s: c/M for s, c in cal.items()
    if s != s0
}

B = math.fsum(x**ALPHA for x in q.values())
qmax = max(q.values())
R = qmax**(ALPHA-1)

mean = math.fsum(
    c*q.get(s,0.0)**(ALPHA-1)
    for s,c in ev.items()
)/N

second = math.fsum(
    c*q.get(s,0.0)**(2*(ALPHA-1))
    for s,c in ev.items()
)/N

sample_var = max(
    0.0,
    N*(second-mean**2)/(N-1)
)

t = math.log(2/EPS)

eb_radius = (
    math.sqrt(2*sample_var*t/N)
    + 7*R*t/(3*(N-1))
)

A_lower = max(0.0,mean-eb_radius)

H_lower = (
    A_lower**ALPHA/B**(ALPHA-1)
    if A_lower>0 and B>0 else 0.0
)

# ------------------------------------------------------------
# 2. Upper bound C3 for each candidate method
# ------------------------------------------------------------

group_stats = {
    method: {
        g: {"shots":0, "max_count":0}
        for g in names
    }
    for method, (names, _) in METHODS.items()
}

for s,c in ev.items():
    if s == s0:
        continue

    for method,(names,classifier) in METHODS.items():
        g = classifier(s)
        if g is None:
            continue
        x = group_stats[method][g]
        x["shots"] += c
        x["max_count"] = max(x["max_count"],c)

results = []
details = []

for method,(names,_) in METHODS.items():
    C_upper = 0.0

    for g in names:
        x = group_stats[method][g]
        atom_upper = cp_upper(x["max_count"])
        mass_upper = cp_upper(x["shots"])
        term = atom_upper**(ALPHA-1)*mass_upper
        C_upper += term

        details.append({
            "method":method,
            "group":g,
            **x,
            "atom_upper":atom_upper,
            "mass_upper":mass_upper,
            "C_upper_term":term
        })

    if method == "hybrid_top100":
        for s in top:
            u = cp_upper(ev.get(s,0))
            C_upper += u**ALPHA

        assert (
            sum(group_stats[method][g]["shots"]
                for g in names)
            + sum(ev.get(s,0) for s in top)
            + ev.get(s0,0)
            == N
        )
    else:
        assert (
            sum(group_stats[method][g]["shots"]
                for g in names)
            + ev.get(s0,0)
            == N
        )

    eta_lower = (
        H_lower/C_upper if C_upper>0 else 0.0
    )

    results.append({
        "method":method,
        "C3_upper":C_upper,
        "H3_lower":H_lower,
        "eta3_lower":eta_lower,
        "positive_alignment_certificate":
            eta_lower>0,
        "eta_above_half":eta_lower>0.5
    })

results_df = pd.DataFrame(results)
details_df = pd.DataFrame(details)

# Descriptive collision estimate, not a confidence bound.
collision_tuples = math.fsum(
    c*(c-1)*(c-2)
    for s,c in ev.items()
    if s != s0 and c>=3
)

empirical_C3 = (
    collision_tuples/(N*(N-1)*(N-2))
)

print("\nCALIBRATION ALIGNMENT CERTIFICATION")
print(results_df.to_string(index=False))

print("\nDENOMINATOR UPPER CONTRIBUTIONS")
print(
    details_df.sort_values(
        ["method","C_upper_term"],
        ascending=[True,False]
    ).to_string(index=False)
)

print("\nDIAGNOSTICS")
print("Evaluation shots:",N)
print("Calibration shots:",M)
print("H3 confidence lower:",H_lower)
print("Empirical C3 (not a bound):",empirical_C3)
print("Previously exploratory eta3: 0.988216")
print("Total confidence events:",EVENTS)
print("Per-event delta:",EPS)
print("Joint nominal confidence: 0.95")
print("Retrospective method-development analysis.")

results_df.to_csv(f"{OUT}/results.csv",index=False)
details_df.to_csv(f"{OUT}/groups.csv",index=False)

with open(f"{OUT}/config.json","w") as f:
    json.dump({
        "delta":DELTA,
        "events":EVENTS,
        "epsilon":EPS,
        "alpha":ALPHA,
        "top_k":TOP_K,
        "calibration_path":CAL_PATH,
        "evaluation_path":EVAL_PATH,
        "status":"retrospective"
    },f,indent=2)

print("\nSaved to:",OUT)
```

```python
# Repeated-sampling coverage test for the dominant-excluded QEC certificate.
# Synthetic data only. Requires numpy, pandas, scipy.
import json, math, os
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import beta

OUT = Path('qec_synthetic_coverage_v1')
ALPHA = 3
DELTA = 0.05
M = 6_000
N = 12_000
REPS = 600              # raise to 2000+ for a more precise coverage estimate
SEED = 20261114
TARGET = 1e-3
EPS = DELTA / 25         # 12 group maxima, 12 group failure masses, 1 EB bound
GROUP_NAMES = ['cal_100_plus', 'cal_20_to_99', 'cal_5_to_19', 'cal_2_to_4', 'cal_1',
               'unseen_w0_2', 'unseen_w3_4', 'unseen_w5_6', 'unseen_w7_8',
               'unseen_w9_12', 'unseen_w13_20', 'unseen_w21_plus']
assert len(GROUP_NAMES) == 12

# Fixed finite alphabets with population syndrome probabilities and conditional failure probabilities.
def make_cases():
    result = {}
    # p0 is the dominant syndrome, but f0 is deliberately varied across cases.
    K=64
    raw=(np.arange(1,K, dtype=float))**(-1.2)
    p=np.r_[0.50, 0.50*raw/raw.sum()]
    f=np.full(K, 0.015); f[0]=0.00001
    f[1:8]=0.0002
    result['concentrated_benign']=(p,f)
    f2=f.copy(); f2[0]=0.80
    result['concentrated_harmful_dominant']=(p,f2)

    K=512
    raw=(np.arange(1,K+1, dtype=float))**(-1.0)
    p=raw/raw.sum()
    f=np.where(np.arange(K)<12, 0.002, 0.12).astype(float)
    result['zipf_tail_harmful']=(p,f)

    K=512
    p=np.full(K,1/K)
    f=np.where((np.arange(K)%7)==0, 0.45, 0.015).astype(float)
    result['uniform_unstable_dominant']=(p,f)

    K=4096
    p=np.full(K,1/K)
    f=np.where((np.arange(K)%23)==0, 0.5, 0.02).astype(float)
    result['sparse_uniform']=(p,f)
    for name,(p,f) in result.items():
        assert np.isclose(p.sum(),1) and np.all(p>=0) and np.all((0<=f)&(f<=1)), name
    return result


def cp_up(k):
    k=np.asarray(k)
    return beta.isf(EPS,k+1,N-k)


def wilson_interval(k,n,z=1.959963984540054):
    ph=k/n
    den=1+z*z/n
    ctr=(ph+z*z/(2*n))/den
    half=z*math.sqrt(ph*(1-ph)/n+z*z/(4*n*n))/den
    return max(0,ctr-half),min(1,ctr+half)


def one_replicate(p,f,weight_bins,rng):
    K=len(p)
    # Split calibration and evaluation by separate independent RNG draws.
    cal=np.bincount(rng.choice(K,size=M,p=p),minlength=K)
    s0=int(np.argmax(cal))
    labels=rng.choice(K,size=N,p=p)
    failed=rng.random(N) < f[labels]
    cnt=np.bincount(labels,minlength=K)
    err=np.bincount(labels[failed],minlength=K)

    g=np.zeros(K,dtype=np.int8)
    g[cal>=100]=0
    g[(cal>=20)&(cal<100)]=1
    g[(cal>=5)&(cal<20)]=2
    g[(cal>=2)&(cal<5)]=3
    g[cal==1]=4
    g[cal==0]=5+weight_bins[cal==0]
    g[s0]=-1
    # Every non-dominant syndrome belongs to exactly one of 12 groups.
    assert np.all((g[np.arange(K)!=s0]>=0)&(g[np.arange(K)!=s0]<12))

    q=cal.astype(float)/M
    q[s0]=0
    B=float(np.sum(q**ALPHA))
    R=float(np.max(q)**(ALPHA-1))
    X=q[labels]**(ALPHA-1)
    m=float(X.mean())
    sample_var=float(X.var(ddof=1))
    t=math.log(2/EPS)
    radius=math.sqrt(2*sample_var*t/N)+7*R*t/(3*(N-1))
    A_lower=max(0.0,m-radius)
    C_lower=A_lower**ALPHA/B**(ALPHA-1) if B>0 and A_lower>0 else 0.0

    # Fixed true parameters, conditional on the calibration-selected s0 and groups.
    true_C=float(np.sum(p[np.arange(K)!=s0]**ALPHA))
    true_F=float(np.sum(p[np.arange(K)!=s0]**(ALPHA-1)*p[np.arange(K)!=s0]*f[np.arange(K)!=s0]))
    true_ratio=true_F/true_C
    true_A=float(np.dot(p,q**(ALPHA-1)))

    max_counts=np.zeros(12,dtype=int)
    group_fail=np.zeros(12,dtype=int)
    true_u=np.zeros(12)
    true_b=np.zeros(12)
    for j in range(12):
        mask=g==j
        if np.any(mask):
            max_counts[j]=int(cnt[mask].max())
            group_fail[j]=int(err[mask].sum())
            true_u[j]=float(p[mask].max())
            true_b[j]=float(np.dot(p[mask],f[mask]))
    atom_up=cp_up(max_counts)
    failure_up=cp_up(group_fail)
    numerator_up=float(np.sum(atom_up**(ALPHA-1)*failure_up))
    upper=min(1.0,numerator_up/C_lower) if C_lower>0 else 1.0
    bad_atom=int(np.sum(true_u>atom_up+1e-14))
    bad_fail=int(np.sum(true_b>failure_up+1e-14))
    bad_EB=int(true_A<A_lower-1e-14)
    return {
      'true_ratio':true_ratio, 'upper':upper,'covers':int(true_ratio<=upper+1e-14),
      'nonvacuous':int(upper<1-1e-12),'certifies_target':int(upper<TARGET),
      'false_target_certificate':int(upper<TARGET and true_ratio>=TARGET),
      'positive_denominator':int(C_lower>0),
      'bad_atom_events':bad_atom,'bad_failure_events':bad_fail,'bad_eb_events':bad_EB,
      'any_component_failure':int(bad_atom+bad_fail+bad_EB>0),
      'cal_dominant':s0, 'actual_max_syndrome':int(np.argmax(p)),
      'cal_selected_actual_dominant':int(s0==int(np.argmax(p))),
      'C_lower':C_lower,'true_C':true_C,
    }


def main():
    OUT.mkdir(parents=True,exist_ok=True)
    cases=make_cases()
    master=np.random.SeedSequence(SEED)
    streams=master.spawn(len(cases))
    records=[]
    for (name,(p,f)),stream in zip(cases.items(),streams):
        rng=np.random.default_rng(stream)
        weights=np.array([int(i).bit_count() for i in range(len(p))])
        # Map Hamming weight to 7 unseen groups, bins 0-2,...,21+.
        bins=np.searchsorted(np.array([3,5,7,9,13,21]),weights,side='right')
        for r in range(REPS):
            o=one_replicate(p,f,bins,rng)
            records.append({'case':name,'rep':r,**o})
        sub=[r for r in records if r['case']==name]
        print(f"{name}: coverage={np.mean([x['covers'] for x in sub]):.4f}, "
              f"nonvacuous={np.mean([x['nonvacuous'] for x in sub]):.4f}, "
              f"component failures={sum(x['any_component_failure'] for x in sub)}")
    raw=pd.DataFrame(records)
    summaries=[]
    for name,part in raw.groupby('case',sort=False):
        k=int(part['covers'].sum())
        lower,hi=wilson_interval(k,len(part))
        summaries.append({
          'case':name,'reps':len(part),'empirical_coverage':k/len(part),
          'coverage_Wilson_95_low':lower,'coverage_Wilson_95_high':hi,
          'noncoverage_count':len(part)-k,
          'mean_upper_bound':part['upper'].mean(),
          'median_upper_bound':part['upper'].median(),
          'nonvacuous_fraction':part['nonvacuous'].mean(),
          'positive_denominator_fraction':part['positive_denominator'].mean(),
          'certifies_target_fraction':part['certifies_target'].mean(),
          'false_target_certificates':int(part['false_target_certificate'].sum()),
          'any_component_failure_count':int(part['any_component_failure'].sum()),
          'atom_event_failures':int(part['bad_atom_events'].sum()),
          'failure_mass_event_failures':int(part['bad_failure_events'].sum()),
          'EB_event_failures':int(part['bad_eb_events'].sum()),
          'selected_population_max_fraction':part['cal_selected_actual_dominant'].mean(),
        })
    summary=pd.DataFrame(summaries)
    print('\nCOVERAGE SUMMARY')
    print(summary.to_string(index=False))
    print('\nChecks: all actual noncoverage events must be contained in component failure events')
    assert ((raw['covers']==0)&(raw['any_component_failure']==0)).sum()==0
    assert (raw['false_target_certificate']<=1-raw['covers']).all()
    assert ((raw['upper']>=0)&(raw['upper']<=1)).all()
    raw.to_csv(OUT/'replicate_results.csv',index=False)
    summary.to_csv(OUT/'coverage_summary.csv',index=False)
    (OUT/'config.json').write_text(json.dumps({
       'alpha':ALPHA,'delta':DELTA,'events':25,'eps':EPS,
       'calibration_shots':M,'evaluation_shots':N,'reps_per_case':REPS,
       'seed':SEED,'target':TARGET, 'status':'synthetic repeated sampling; methods fixed',
       'cases':list(cases)
    },indent=2))
    print('\nSaved to:',OUT)
    print('Interpretation: coverage at least 95% in theory, not necessarily exactly 95%.')
    print('High coverage with upper bound=1 is uninformative; inspect nonvacuous_fraction.')

if __name__=='__main__': main()
```

```python
"""Seeded synthetic benchmark for dominant-excluded alpha=3 certification.
Methods: groupwise Hoeffding/H&#246;lder/empirical-Bernstein, atom CP lower,
and exact disjoint-triple post-selection CP upper.
Each method has *marginal* 95% coverage, not joint 95% coverage.
"""
import argparse, math, json, sys
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import beta

ALPHA=3
DELTA=.05
TARGET=.001
GROUPS=12
K_ATOMS=100


def cases():
    out={}
    k=64
    raw=np.arange(1,k,dtype=float)**-1.2
    p=np.r_[.50,.50*raw/raw.sum()]
    f=np.full(k,.005); f[0]=.00001; f[1:16]=.000001
    out['concentrated_easy']=(p,f)
    out['concentrated_harmful_dominant']=(p,f2) if 'f2' in locals() else (p, np.where(np.arange(k)<16, 0.0005, 0.025)) # fallback fix for undefined f2 if needed
    # Ensuring f2 is defined properly in case case configurations have a typo
    f2=np.full(k,.025); f2[0]=.8; f2[1:16]=.0005
    out['concentrated_harmful_dominant']=(p,f2)
    k=512
    raw=np.arange(1,k+1,dtype=float)**-1.0
    p=raw/raw.sum(); f=np.where(np.arange(k)<12,.0005,.12)
    out['zipf_harmful_tail']=(p,f)
    k=4096
    p=np.full(k,1/k); f=np.where((np.arange(k)%23)==0,.5,.02)
    out['sparse_uniform']=(p,f)
    for name,(p,f) in out.items():
        assert abs(p.sum()-1)<1e-12 and np.all((f>=0)&(f<=1)),name
    return out


def up(k,n,eps):
    k=np.asarray(k)
    out=beta.isf(eps,k+1,n-k)
    return np.where(k>=n,1.0,out)


def low(k,n,eps):
    k=np.asarray(k)
    out=beta.ppf(eps,k,n-k+1)
    return np.where(k<=0,0.0,out)


def grouping(cal,s0):
    k=len(cal)
    w=np.array([i.bit_count() for i in range(k)])
    binidx=np.searchsorted([3,5,7,9,13,21],w,side='right')
    group=np.where(cal>=100,0,np.where(cal>=20,1,np.where(cal>=5,2,
               np.where(cal>=2,3,np.where(cal==1,4,5+binidx))))).astype('int16')
    group[s0]=-1
    assert np.all((group[np.arange(k)!=s0]>=0)&(group[np.arange(k)!=s0]<12))
    return group


def proposed(counts,errors,cal,s0,group,n,eps):
    q=cal.astype(float)/cal.sum(); q[s0]=0
    B=float(np.sum(q**3)); R=float(q.max()**2)
    if B==0 or R==0:return 1.,0.
    mean=float(np.dot(counts,q*q)/n)
    second=float(np.dot(counts,q**4)/n)
    var=max(0.,(second-mean**2)*n/(n-1))
    t=math.log(2/eps)
    rad=math.sqrt(2*var*t/n)+7*R*t/(3*(n-1))
    L=max(0.,mean-rad)
    den=L**3/B**2 if L else 0.
    if den==0:return 1.,den
    group_count=np.bincount(group[group>=0],weights=counts[group>=0],minlength=12).astype(int)
    group_fail=np.bincount(group[group>=0],weights=errors[group>=0],minlength=12).astype(int)
    mx=np.zeros(12,dtype=int)
    np.maximum.at(mx,group[group>=0],counts[group>=0])
    num=float(np.sum(up(mx,n,eps)**2*up(group_fail,n,eps)))
    return min(1.,num/den),den


def atomic(counts,errors,cal,s0,group,n,k):
    # Same groupwise numerator; atomic lower denominator on calibration-selected K labels.
    ranking=np.argsort(-cal,kind='stable')
    ranking=ranking[ranking!=s0][:k]
    eps=DELTA/(24+len(ranking))
    gf=np.bincount(group[group>=0],weights=errors[group>=0],minlength=12).astype(int)
    mx=np.zeros(12,dtype=int)
    np.maximum.at(mx,group[group>=0],counts[group>=0])
    num=float(np.sum(up(mx,n,eps)**2*up(gf,n,eps)))
    den=float(np.sum(low(counts[ranking],n,eps)**3))
    return min(1.,num/den) if den else 1.,den


def disjoint_triples(labels,failed,s0,n):
    # Nonoverlapping triples are iid. Conditional on a matching triple not equal to s0,
    # failure of its first draw has exactly the target R_3 distribution.
    m=n//3
    a=labels[:3*m].reshape(m,3)
    selected=(a[:,0]==a[:,1])&(a[:,1]==a[:,2])&(a[:,0]!=s0)
    k=int(selected.sum())
    successes=int(failed[:3*m:3][selected].sum())
    bound=float(up(successes,k,DELTA)) if k else 1.
    return bound,k,successes


def wilson(k,n,z=1.959963984540054):
    if n==0:return (np.nan,np.nan)
    ph=k/n; den=1+z*z/n
    ctr=(ph+z*z/(2*n))/den
    half=z*math.sqrt(ph*(1-ph)/n+z*z/(4*n*n))/den
    return max(0.,ctr-half),min(1.,ctr+half)


def run(reps, budgets, calshots, seed, out, chosen):
    out=Path(out);out.mkdir(parents=True,exist_ok=True)
    records=[]
    available=cases()
    for case_ix,case in enumerate(chosen):
        p,f=available[case];K=len(p)
        rng=np.random.default_rng(np.random.SeedSequence([seed,case_ix]))
        print(f'Case {case}: K={K}',flush=True)
        for rep in range(reps):
            cal=np.bincount(rng.choice(K,size=calshots,p=p),minlength=K)
            s0=int(np.argmax(cal));group=grouping(cal,s0)
            labels=rng.choice(K,size=max(budgets),p=p)
            failed=rng.random(max(budgets))<f[labels]
            keep=np.arange(K)!=s0
            C=float(np.sum(p[keep]**3))
            F=float(np.sum(p[keep]**3*f[keep]))
            truth=F/C
            for n in budgets:
                lab=labels[:n];bad=failed[:n]
                counts=np.bincount(lab,minlength=K)
                errors=np.bincount(lab[bad],minlength=K)
                a,den_a=proposed(counts,errors,cal,s0,group,n,DELTA/25)
                b,den_b=atomic(counts,errors,cal,s0,group,n,min(K_ATOMS,K-1))
                c,k,colfail=disjoint_triples(lab,bad,s0,n)
                for method,bound,diag in [('holder_EB',a,den_a),('atomic_CP100',b,den_b),('disjoint_triple_CP',c,k)]:
                    records.append(dict(case=case,rep=rep,n=n,calshots=calshots,method=method,
                                        true_ratio=truth,upper=bound,covers=truth<=bound+1e-14,
                                        certifies=bound<TARGET,valid_target=truth<TARGET,
                                        nonvacuous=bound<1-1e-12,diagnostic=diag,
                                        collision_failures=colfail if method=='disjoint_triple_CP' else -1))
            if (rep+1)%50==0:print(f'  {rep+1}/{reps}',flush=True)
    raw=pd.DataFrame(records)
    summary=[]
    for (case,n,method),g in raw.groupby(['case','n','method'],sort=True):
        coverage=int(g['covers'].sum());power=int(g['certifies'].sum()); total=len(g)
        lo,hi=wilson(coverage,total)
        plo,phi=wilson(power,total)
        summary.append(dict(case=case,n=n,method=method,reps=total,
                            true_ratio_median=float(g['true_ratio'].median()),
                            coverage=coverage/total,coverage_wilson_lo=lo,coverage_wilson_hi=hi,
                            certification_rate=power/total,power_wilson_lo=plo,power_wilson_hi=phi,
                            median_upper=float(g['upper'].median()),
                            nonvacuous_fraction=float(g['nonvacuous'].mean()),
                            median_diagnostic=float(g['diagnostic'].median()),
                            false_certificates=int((g['certifies']&~g['valid_target']).sum())))
    sm=pd.DataFrame(summary)
    raw.to_csv(out/'replicate_results.csv',index=False)
    sm.to_csv(out/'benchmark_summary.csv',index=False)
    (out/'config.json').write_text(json.dumps(dict(seed=seed,reps=reps,budgets=budgets,
                 calshots=calshots,cases=chosen,alpha=3,delta=DELTA,target=TARGET,
                 note='95% marginal per method, synthetic truth recomputed for calibration-selected dominant'),indent=2))
    print('\nBENCHMARK SUMMARY\n',sm[['case','n','method','true_ratio_median','coverage','certification_rate','median_upper','nonvacuous_fraction','median_diagnostic','false_certificates']].to_string(index=False),flush=True)
    return sm

if __name__=='__main__':
    parser=argparse.ArgumentParser()
    parser.add_argument('--reps',type=int,default=200)
    parser.add_argument('--budgets',type=int,nargs='+',default=[5000,20000,100000])
    parser.add_argument('--calshots',type=int,default=6000)
    parser.add_argument('--seed',type=int,default=20261114)
    parser.add_argument('--out',default='qec_power_benchmark_v1')
    parser.add_argument('--cases',nargs='+',default=list(cases()))

    # Ignore notebook/ipykernel connection-file parameters (-f ...)
    if any(arg.startswith('-f') for arg in sys.argv):
        args, unknown = parser.parse_known_args(args=[])
    else:
        args = parser.parse_args()

    run(args.reps,args.budgets,args.calshots,args.seed,args.out,args.cases)
```

```python

import os, sys, math, time, importlib
from pathlib import Path
import numpy as np
import pandas as pd
import stim
import pymatching

# Official authors' extracted archive.
ROOT = Path("original_paper_reproduction/authors_code").resolve()
assert (ROOT / "sampler_square.py").exists()
assert (ROOT / "surface_code_simulation/utils_surface_code.py").exists()

# NumPy 2.x compatibility for the authors' np.math.factorial.
if not hasattr(np, "math"):
    np.math = math
    print("Applied NumPy 2 compatibility shim.")

sys.path.insert(0, str(ROOT))
sys.path.insert(0, str(ROOT / "surface_code_simulation"))

from sampler_square import EmpiricalPnSampler, keep_repeated_rows
from utils_surface_code import surface_code_capacity_zero

D = 5
ALPHA = 2
SHOTS = 100_000
PHYSICAL_PS = [0.12, 0.14]
SEED = 20261114

results = []

for j, p in enumerate(PHYSICAL_PS):
    print(f"\nRunning official code-capacity model: d={D}, p={p}")

    start = time.perf_counter()
    circuit = surface_code_capacity_zero(D, D, p)

    sampler = circuit.compile_detector_sampler(seed=SEED + j)
    detectors, observables = sampler.sample(
        SHOTS, separate_observables=True
    )

    matching = pymatching.Matching.from_detector_error_model(
        circuit.detector_error_model(decompose_errors=True)
    )
    predictions = matching.decode_batch(detectors)
    failed = np.any(predictions != observables, axis=1)

    # Authors' empirical resampling acceptance rule:
    # retain records whose syndrome occurs at least alpha times.
    _, kept_indices = keep_repeated_rows(detectors, ALPHA)
    accepted = len(kept_indices)

    # Authors' empirical falling-factorial sampler.
    np.random.seed(SEED + 100 + j)
    empirical = EmpiricalPnSampler(detectors, n=ALPHA)
    selected = empirical.sample_indices(k=accepted)

    # Selected indices refer to the ORIGINAL observations.
    resampled_errors = int(failed[selected].sum())

    # Authors' zero-syndrome post-selection control.
    ps_indices = np.flatnonzero(~detectors.any(axis=1))
    ps_errors = int(failed[ps_indices].sum())

    row = {
        "distance": D,
        "physical_p": p,
        "shots": SHOTS,
        "alpha": ALPHA,
        "detector_count": circuit.num_detectors,
        "baseline_LER": float(failed.mean()),
        "SR_acceptance": accepted / SHOTS,
        "SR_LER": (
            resampled_errors / accepted
            if accepted else np.nan
        ),
        "PS_acceptance": len(ps_indices) / SHOTS,
        "PS_LER": (
            ps_errors / len(ps_indices)
            if len(ps_indices) else np.nan
        ),
        "runtime_seconds": time.perf_counter() - start
    }

    results.append(row)
    print(pd.DataFrame([row]).to_string(index=False))

df = pd.DataFrame(results)

out = Path("original_paper_reproduction/figure4_pilot")
out.mkdir(parents=True, exist_ok=True)
df.to_csv(out / "results.csv", index=False)

print("\nOFFICIAL METHOD PILOT")
print(df.to_string(index=False))
print("\nSaved to:", out / "results.csv")
print("Pilot only: not full Figure 4 reproduction.")
```

```python

import os, sys, math
from pathlib import Path
from collections import Counter
import numpy as np
import pandas as pd
import stim
import pymatching

ROOT = Path("original_paper_reproduction/authors_code").resolve()
sys.path.insert(0, str(ROOT))
sys.path.insert(0, str(ROOT / "surface_code_simulation"))

if not hasattr(np, "math"):
    np.math = math

from utils_surface_code import surface_code_capacity_zero

DISTANCES = [5, 7, 9]
PS = [0.12, 0.14]
BUDGETS = [1_000, 10_000, 100_000]
REPEATS = 3
ALPHA = 2
BASE_SEED = 20261114

OUT = Path("original_paper_reproduction/figure4_sweep_v1")
OUT.mkdir(parents=True, exist_ok=True)

results = []

for d in DISTANCES:
    for pidx, p in enumerate(PS):
        circuit = surface_code_capacity_zero(d, d, p)
        decoder = pymatching.Matching.from_detector_error_model(
            circuit.detector_error_model(decompose_errors=True)
        )

        for rep in range(REPEATS):
            seed = BASE_SEED + d*1000 + pidx*100 + rep

            sampler = circuit.compile_detector_sampler(seed=seed)
            detectors, observables = sampler.sample(
                max(BUDGETS), separate_observables=True
            )

            prediction = decoder.decode_batch(detectors)
            failed = np.any(prediction != observables, axis=1)

            for N in BUDGETS:
                # Count syndrome occurrences and failures.
                packed = np.packbits(
                    detectors[:N], axis=1, bitorder="little"
                )

                counts = Counter(map(bytes, packed))
                error_counts = Counter(map(bytes, packed[failed[:N]]))

                eligible = sum(
                    c for c in counts.values() if c >= ALPHA
                )

                weights = {}
                weight_sum = 0

                for syndrome, c in counts.items():
                    if c >= ALPHA:
                        w = c*(c-1)
                        weights[syndrome] = w
                        weight_sum += w

                # Exact conditional mean under the author's
                # empirical alpha=2 resampling distribution.
                if weight_sum:
                    expected_errors = sum(
                        w*error_counts.get(s, 0)/counts[s]
                        for s, w in weights.items()
                    )
                    rb_rate = expected_errors/weight_sum

                    # Monte Carlo realization of author-style
                    # resampling, without materializing all indices.
                    keys = list(weights)
                    probabilities = np.array(
                        [weights[s] for s in keys], dtype=float
                    )
                    probabilities /= probabilities.sum()

                    rng = np.random.default_rng(
                        seed + N*17 + 100000
                    )
                    chosen = rng.choice(
                        len(keys),
                        size=eligible,
                        p=probabilities
                    )

                    conditional_failure = np.array([
                        error_counts.get(s, 0)/counts[s]
                        for s in keys
                    ])

                    sampled_failures = (
                        rng.random(eligible)
                        < conditional_failure[chosen]
                    )

                    sampled_rate = float(sampled_failures.mean())
                else:
                    rb_rate = np.nan
                    sampled_rate = np.nan

                zero = bytes(packed.shape[1])
                zero_count = counts.get(zero, 0)
                zero_failures = error_counts.get(zero, 0)

                results.append({
                    "distance":d,
                    "p":p,
                    "replicate":rep,
                    "shots":N,
                    "detectors":circuit.num_detectors,
                    "baseline_LER":float(failed[:N].mean()),
                    "SR_acceptance":eligible/N,
                    "SR_LER_MC":sampled_rate,
                    "SR_LER_conditional_mean":rb_rate,
                    "PS_acceptance":zero_count/N,
                    "PS_LER":(
                        zero_failures/zero_count
                        if zero_count else np.nan
                    ),
                    "distinct_syndromes":len(counts)
                })

            print(
                f"Completed d={d}, p={p}, "
                f"replicate={rep+1}/{REPEATS}",
                flush=True
            )

df = pd.DataFrame(results)
df.to_csv(OUT/"all_replicates.csv", index=False)

summary = df.groupby(
    ["distance","p","shots"], as_index=False
).agg(
    SR_LER_mean=("SR_LER_MC","mean"),
    SR_RB_mean=("SR_LER_conditional_mean","mean"),
    SR_acceptance_mean=("SR_acceptance","mean"),
    PS_LER_mean=("PS_LER","mean"),
    PS_acceptance_mean=("PS_acceptance","mean"),
    baseline_mean=("baseline_LER","mean")
)

summary.to_csv(OUT/"summary.csv",index=False)

print("\nFIGURE 4 FINITE-SAMPLE SWEEP")
print(summary.to_string(index=False))
print("\nSaved to:",OUT)
print("These are pilot averages, not published bootstrap intervals.")
```

```python
"""Independent-batch pilot reproduction of Fig. 4 of Colmenarez et al.

Uses authors' original surface_code_capacity_zero circuit from extracted Zenodo archive.
Independent fresh batches at each (distance, p, N, replicate); percentile bootstrap
across batches (not a claim to exactly match the authors' unpublished bootstrap code).
"""
import sys, os, math, json, hashlib
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

try:
    import stim, pymatching
except ImportError as exc:
    raise SystemExit('Install stim and pymatching: %pip install -q stim pymatching') from exc

ROOT = Path('original_paper_reproduction/authors_code').resolve()
assert (ROOT/'surface_code_simulation/utils_surface_code.py').is_file(), 'First extract the official Zenodo archive.'
sys.path.insert(0, str(ROOT/'surface_code_simulation'))
from utils_surface_code import surface_code_capacity_zero

DISTANCES=(5,7,9)
PHYSICAL_PS=(0.12,0.14)
BUDGETS=(1000,10000,100000)
REPLICATES=12
BOOTSTRAPS=3000
BASE_SEED=20261114
OUT=Path('original_paper_reproduction/figure4_independent_v1')
OUT.mkdir(parents=True, exist_ok=True)

# Independent sample for every condition, with a stable and injective seed mapping.
def seed_for(di,pi,ni,rep):
    return BASE_SEED + di*100000 + pi*30000 + ni*3000 + rep

def summarize_shots(detectors, failed, rng):
    n=len(failed)
    packed=np.packbits(detectors, axis=1, bitorder='little')
    unique, inverse, counts=np.unique(packed,axis=0,return_inverse=True,return_counts=True)
    fail_counts=np.bincount(inverse,weights=failed.astype(np.int64),minlength=len(counts))
    eligible=int(counts[counts>=2].sum())
    weights=counts.astype(np.float64)*(counts-1)
    denom=weights.sum()
    ps_idx=np.flatnonzero(~detectors.any(axis=1))
    ps_count=len(ps_idx)
    ps_err=int(failed[ps_idx].sum())
    out=dict(baseline_LER=float(failed.mean()), SR_acceptance=eligible/n,
             SR_eligible=eligible, distinct_syndromes=len(counts),
             PS_acceptance=ps_count/n, PS_count=ps_count,
             PS_error_count=ps_err,PS_LER=ps_err/ps_count if ps_count else float('nan'))
    if not eligible or not denom:
        out.update(SR_LER_MC=float('nan'),SR_LER_RB=float('nan'),SR_error_count=0)
        return out
    rb=float(np.dot(counts-1,fail_counts)/denom)
    # Every selected output of the author's sampler is an independent draw,
    # conditional on the original source data, with failure probability rb.
    # Therefore an exact, distributionally equivalent MC implementation is Binomial(K, rb).
    sampled_errors=int(rng.binomial(eligible,rb))
    out.update(SR_LER_MC=sampled_errors/eligible,SR_LER_RB=rb,SR_error_count=sampled_errors)
    return out

def boot_ci(x, seed):
    x=np.asarray(x,dtype=float)
    x=x[np.isfinite(x)]
    if len(x)<2: return float('nan'),float('nan'),float('nan'),len(x)
    rng=np.random.default_rng(seed)
    draws=np.empty(BOOTSTRAPS)
    for j in range(BOOTSTRAPS):
        draws[j]=rng.choice(x,size=len(x),replace=True).mean()
    return float(x.mean()),float(np.quantile(draws,.165)),float(np.quantile(draws,.835)),len(x)

config=dict(distances=DISTANCES,physical_ps=PHYSICAL_PS,budgets=BUDGETS,
            replicates=REPLICATES, bootstraps=BOOTSTRAPS,base_seed=BASE_SEED,
            interval='67% percentile bootstrap across independent batches',
            SR='authors alpha=2 collision weights and eligible-shot acceptance',
            SR_MC='conditional Binomial(K,RB) equivalent to author empirical resampler')
(OUT/'config.json').write_text(json.dumps(config,indent=2))
rows=[]; circuit_hashes={}
for di,d in enumerate(DISTANCES):
    for pi,p in enumerate(PHYSICAL_PS):
        circuit=surface_code_capacity_zero(d,d,p)
        circuit_hashes[f'd{d}_p{p}']=hashlib.sha256(str(circuit).encode()).hexdigest()
        matcher=pymatching.Matching.from_detector_error_model(
            circuit.detector_error_model(decompose_errors=True))
        for ni,N in enumerate(BUDGETS):
            for rep in range(REPLICATES):
                seed=seed_for(di,pi,ni,rep)
                detector_sampler=circuit.compile_detector_sampler(seed=seed)
                detectors, observables=detector_sampler.sample(N,separate_observables=True)
                pred=matcher.decode_batch(detectors)
                failed=np.any(pred!=observables,axis=1)
                row=summarize_shots(detectors,failed,np.random.default_rng(seed+1_000_000))
                row.update(distance=d,p=p,shots=N,replicate=rep,seed=seed,
                           num_detectors=circuit.num_detectors)
                rows.append(row)
            print(f'Completed d={d}, p={p}, N={N}: {REPLICATES} independent batches',flush=True)

(OUT/'circuit_sha256.json').write_text(json.dumps(circuit_hashes,indent=2))
df=pd.DataFrame(rows)
df.to_csv(OUT/'replicates.csv',index=False)
metrics=('baseline_LER','SR_LER_MC','SR_LER_RB','SR_acceptance','PS_LER','PS_acceptance')
summary=[]
for gi,((d,p,N),g) in enumerate(df.groupby(['distance','p','shots'],sort=True)):
    record=dict(distance=int(d),p=float(p),shots=int(N),replicates=len(g),
                total_SR_eligible=int(g.SR_eligible.sum()),
                total_PS_count=int(g.PS_count.sum()))
    for metric_i,m in enumerate(metrics):
        mean,lo,hi,defined=boot_ci(g[m],BASE_SEED+10000+gi*20+metric_i)
        record[m+'_mean']=mean
        record[m+'_lo67']=lo
        record[m+'_hi67']=hi
        record[m+'_defined_batches']=defined
    summary.append(record)
res=pd.DataFrame(summary)
res.to_csv(OUT/'summary.csv',index=False)
print('\nINDEPENDENT-BATCH FIGURE 4 SUMMARY')
cols=['distance','p','shots','SR_LER_MC_mean','SR_LER_RB_mean',
      'SR_acceptance_mean','PS_LER_mean','PS_acceptance_mean',
      'SR_LER_MC_defined_batches','PS_LER_defined_batches']
print(res[cols].to_string(index=False))

fig,axes=plt.subplots(2,2,figsize=(13,9),constrained_layout=True)
for col,p in enumerate(PHYSICAL_PS):
    for d in DISTANCES:
        g=res[(res.distance==d)&(res.p==p)].sort_values('shots')
        x=g.shots.to_numpy(dtype=float)
        for method,metric,ls in [('SR','SR_LER_MC','-'),('PS','PS_LER','--')]:
            if d==9 and method=='PS': continue # publication omits this unreliable curve
            y=g[metric+'_mean'].to_numpy(dtype=float)
            lo=g[metric+'_lo67'].to_numpy(dtype=float)
            hi=g[metric+'_hi67'].to_numpy(dtype=float)
            valid=np.isfinite(y)&(y>0)&np.isfinite(lo)&np.isfinite(hi)
            if valid.any():
                axes[0,col].plot(x[valid],y[valid],ls,marker='o',label=f'd={d} {method}')
                axes[0,col].fill_between(x[valid],np.maximum(lo[valid],1e-12),hi[valid],alpha=.12)
        for method,metric,ls in [('SR','SR_acceptance','-'),('PS','PS_acceptance','--')]:
            y=g[metric+'_mean'].to_numpy(dtype=float)
            lo=g[metric+'_lo67'].to_numpy(dtype=float)
            hi=g[metric+'_hi67'].to_numpy(dtype=float)
            valid=np.isfinite(y)&(y>0)&np.isfinite(lo)&np.isfinite(hi)
            if valid.any():
                axes[1,col].plot(x[valid],y[valid],ls,marker='o',label=f'd={d} {method}')
                axes[1,col].fill_between(x[valid],np.maximum(lo[valid],1e-12),hi[valid],alpha=.12)
    axes[0,col].set(xscale='log',yscale='log',title=f'p={p}',ylabel='Logical failure rate',xlabel='Batch shots N')
    axes[1,col].set(xscale='log',yscale='log',ylabel='Acceptance rate',xlabel='Batch shots N')
    axes[0,col].legend(fontsize=8)
    axes[1,col].legend(fontsize=8)
fig.suptitle('Figure 4 reproduction pilot: independent batches, 67% bootstrap CI')
fig.savefig(OUT/'figure4_independent_pilot.png',dpi=180)
plt.show()
print('\nSaved:',OUT.resolve())
print('NOTE: This does not establish a numerical reproduction of the published figure;')
print('the exact author bootstrap implementation and comparison coordinates are not archived.')
```

```python
"""Matched finite-sample syndrome-resampling certification pilot.

Uses authors' original surface_code_capacity_zero and PyMatching MWPM.
Full and calibration-dominant-excluded alpha=2 targets are kept separate.
The author-style conditional point estimator uses falling-factorial weights.
The certificates use independent calibration and evaluation; 12 CP max-atom,
12 CP group-failure and 1 empirical-Bernstein expectation events per target.
For each target the guarantee is marginal 95%, not joint across targets/runs.
"""
from __future__ import annotations
import hashlib
import json
import math
import sys
import time
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import beta

# Faster first run. Increase only after inspecting this pilot.
CONFIG = dict(
    distances=[5, 7, 9],
    ps=[0.12, 0.14],
    calibration_shots=50_000,
    evaluation_shots=100_000,
    replicates=1,
    alpha=2,
    delta=0.05,
    groups=12,
    seed=20261114,
    benchmark_thresholds=[1e-3, 1e-2, 5e-2],
    out_dir='original_paper_reproduction/matched_certification_v1',
    authors_root='original_paper_reproduction/authors_code',
)
GROUP_NAMES = [
    'observed_100plus','observed_20to99','observed_5to19',
    'observed_2to4','observed_1',
    'unseen_w0to2','unseen_w3to4','unseen_w5to6',
    'unseen_w7to8','unseen_w9to12','unseen_w13to20','unseen_w21plus',
]
assert len(GROUP_NAMES)==12


def assign_group(c, syndrome_key):
    """All groups fixed by calibration. Bit weights refer to detector bits."""
    if c >= 100: return 0
    if c >= 20: return 1
    if c >= 5: return 2
    if c >= 2: return 3
    if c == 1: return 4
    w = int.from_bytes(syndrome_key, 'little').bit_count()
    if w <= 2: return 5
    if w <= 4: return 6
    if w <= 6: return 7
    if w <= 8: return 8
    if w <= 12: return 9
    if w <= 20: return 10
    return 11


def upper_cp(k, N, eps):
    # one-sided Clopper-Pearson upper for fixed binomial marginal;
    # k may be max of observed syndrome counts, which dominates true argmax.
    if k >= N: return 1.0
    return float(beta.isf(eps, int(k)+1, int(N)-int(k)))


def aggregate(packed, failed):
    uniques, inverse, counts = np.unique(
        packed, axis=0, return_inverse=True, return_counts=True
    )
    errors = np.bincount(inverse[failed], minlength=len(counts))
    keys = [bytes(row) for row in uniques]
    return keys, counts.astype(np.int64), errors.astype(np.int64)


def rb_and_acceptance(counts, errors, include_mask, total_shots):
    # Exactly the conditional mean of the authors' alpha=2 empirical sampler.
    # This is not an unbiased estimate of the population ratio.
    eligible = int(counts[(counts >= 2) & include_mask].sum())
    w = counts.astype(np.float64)*(counts.astype(np.float64)-1)
    w[~include_mask] = 0
    denom = float(w.sum())
    numerator = float(np.dot(((counts-1)*include_mask).astype(float), errors))
    point = numerator/denom if denom>0 else math.nan
    return point, eligible/total_shots, eligible, denom


def certificate(cal_dict, eval_keys, eval_counts, eval_errors, N, M, dominant, exclude):
    alpha=CONFIG['alpha']
    delta=CONFIG['delta']
    G=CONFIG['groups']
    eps=delta/(2*G+1)
    group_max=np.zeros(G,dtype=np.int64)
    group_fail=np.zeros(G,dtype=np.int64)
    # q contains all calibration weights, optionally excluding its selected dominant.
    q={k: float(v)/M for k,v in cal_dict.items() if not(exclude and k==dominant)}
    B=sum(v**alpha for v in q.values())
    R=max((v**(alpha-1) for v in q.values()),default=0.0)
    # EB expectation X = q[S]^(alpha-1), with 0 for calibration-unseen.
    sum_x=0.0
    sum_x2=0.0
    for k,c,e in zip(eval_keys,eval_counts,eval_errors):
        if exclude and k==dominant:
            continue
        group=assign_group(cal_dict.get(k,0),k)
        group_max[group]=max(int(group_max[group]),int(c))
        group_fail[group]+=e
    for k,c in zip(eval_keys,eval_counts):
        x = q.get(k,0.0)**(alpha-1)
        sum_x += c*x
        sum_x2 += c*x*x
    mean_x=sum_x/N
    var_x=max(0.0,(sum_x2-N*mean_x**2)/(N-1))
    t=math.log(2/eps)
    radius=math.sqrt(2*var_x*t/N)+7*R*t/(3*(N-1))
    A_lo=max(0.0, mean_x-radius)
    C_lo=(A_lo**alpha/B**(alpha-1)) if A_lo>0 and B>0 else 0.0
    U=np.array([upper_cp(k,N,eps) for k in group_max])
    V=np.array([upper_cp(k,N,eps) for k in group_fail])
    F_up=float(np.sum(U**(alpha-1)*V))
    ratio_up=min(1.0,F_up/C_lo) if C_lo>0 else 1.0
    result=dict(
        target='excluded_dominant' if exclude else 'full',
        selected_dominant_hex=dominant.hex(),
        q_support=len(q),q_max=max(q.values(), default=0.0),
        numerator_upper=F_up,denominator_lower=C_lo,
        A_mean=mean_x,A_lower=A_lo,EB_radius=radius,
        q_power_sum_B=B,group_count=G,events=2*G+1,
        per_event_epsilon=eps, confidence_marginal=1-delta,
        certificate_upper=ratio_up,
        nonvacuous=int(ratio_up<1),
    )
    for thresh in CONFIG['benchmark_thresholds']:
        result['certifies_'+str(thresh)]=int(ratio_up<thresh)
    return result, dict(group_max=group_max,group_fail=group_fail,U=U,V=V)


def self_tests():
    assert assign_group(100,b'\x00')==0
    assert assign_group(1,b'\x00')==4
    assert assign_group(0,b'\x00')==5
    assert assign_group(0,b'\xff')==8
    assert assign_group(0,b'\xff\xff\xff')==11
    assert 0.0<upper_cp(0,1000,.002)<1
    assert upper_cp(1000,1000,.002)==1
    # A sample with known repeated-syndrome ratio: A: (3,1 fail), B: (2,0 fail).
    rb,acc,eligible,denom=rb_and_acceptance(
        np.array([3,2,1]),np.array([1,0,1]),np.array([True,True,True]),6
    )
    assert abs(rb-(2/(6+2)))<1e-14
    assert eligible==5 and abs(acc-5/6)<1e-14
    assert denom==8
    # In-place exclusion should not remove excluded syndrome from EB draw count.
    c={'\x00'.encode():5, '\x01'.encode():5}
    ekeys=[b'\x00',b'\x01']
    ec=np.array([5,5]); er=np.array([0,1]);
    full,_=certificate(c,ekeys,ec,er,10,10,b'\x00',False)
    excluded,_=certificate(c,ekeys,ec,er,10,10,b'\x00',True)
    assert full['q_support']==2 and excluded['q_support']==1
    assert full['certificate_upper']>=0 and excluded['certificate_upper']>=0
    print('Internal deterministic tests: PASS')


def main():
    self_tests()
    try:
        import stim, pymatching
    except ImportError as exc:
        raise RuntimeError('Install dependencies first: %pip install stim pymatching scipy pandas') from exc
    root=Path(CONFIG['authors_root']).resolve()
    utils=root/'surface_code_simulation'/'utils_surface_code.py'
    if not utils.exists():
        raise FileNotFoundError(f'Missing official source: {utils}')
    sys.path.insert(0,str(utils.parent))
    from utils_surface_code import surface_code_capacity_zero
    out=Path(CONFIG['out_dir'])
    out.mkdir(parents=True,exist_ok=True)
    metadata={**CONFIG,'source_sha256':hashlib.sha256(utils.read_bytes()).hexdigest(),
              'stim_version':stim.__version__, 'pymatching_version':pymatching.__version__,
              'method':'author code-capacity X_ERROR, MWPM, alpha2, independent cal/eval',
              'note':'marginal 95% per target only; not simultaneous across analyses'}
    (out/'config.json').write_text(json.dumps(metadata,indent=2))
    rows=[]; group_rows=[]
    for d in CONFIG['distances']:
        for pidx,p in enumerate(CONFIG['ps']):
            circuit=surface_code_capacity_zero(d,d,p)
            decoder=pymatching.Matching.from_detector_error_model(
                circuit.detector_error_model(decompose_errors=True))
            c_hash=hashlib.sha256(str(circuit).encode()).hexdigest()
            for rep in range(CONFIG['replicates']):
                seed=CONFIG['seed']+d*10000+pidx*1000+rep*10
                start=time.perf_counter()
                # Two fresh independently seeded Stim samplers.
                M=CONFIG['calibration_shots']; N=CONFIG['evaluation_shots']
                cal_det, _ = circuit.compile_detector_sampler(seed=seed).sample(
                    M,separate_observables=True)
                eval_det,obs = circuit.compile_detector_sampler(seed=seed+1).sample(
                    N,separate_observables=True)
                fail=np.any(decoder.decode_batch(eval_det)!=obs,axis=1)
                cal_pack=np.packbits(cal_det,axis=1,bitorder='little')
                eval_pack=np.packbits(eval_det,axis=1,bitorder='little')
                ck,cc,_=aggregate(cal_pack,np.zeros(M,dtype=bool))
                ek,ec,ee=aggregate(eval_pack,fail)
                cal=dict(zip(ck,map(int,cc)))
                # Deterministic tie-breaking on packed-syndrome bytes.
                dominant=min(ck,key=lambda k:(-cal[k],k))
                full_point, full_acc, full_eligible, full_pairs=rb_and_acceptance(
                    ec,ee,np.ones(len(ec),dtype=bool),N)
                mask=np.array([k!=dominant for k in ek],dtype=bool)
                exc_point,exc_acc,exc_eligible,exc_pairs=rb_and_acceptance(ec,ee,mask,N)
                zero=bytes(eval_pack.shape[1])
                zero_stats={k:(int(c),int(e)) for k,c,e in zip(ek,ec,ee)}.get(zero,(0,0))
                base=dict(distance=d,p=p,replicate=rep,calibration_shots=M,
                          evaluation_shots=N,detectors=circuit.num_detectors,
                          distinct_cal=len(ck),distinct_eval=len(ek),
                          calibration_dominant_count=cal[dominant],
                          evaluation_dominant_count=int(sum(ec[i] for i,k in enumerate(ek) if k==dominant)),
                          circuit_sha256=c_hash,cal_seed=seed,eval_seed=seed+1,
                          baseline_LER=float(fail.mean()),
                          author_full_SR_point=full_point,
                          author_full_SR_acceptance=full_acc,
                          author_full_SR_eligible=full_eligible,
                          excluded_SR_point=exc_point,
                          excluded_SR_acceptance=exc_acc,
                          excluded_SR_eligible=exc_eligible,
                          PS_acceptance=zero_stats[0]/N,
                          PS_LER=zero_stats[1]/zero_stats[0] if zero_stats[0] else math.nan)
                for exclude in [False,True]:
                    cert,g=certificate(cal,ek,ec,ee,N,M,dominant,exclude)
                    r={**base,**cert,
                       'matched_point_estimate':exc_point if exclude else full_point,
                       'runtime_seconds':time.perf_counter()-start}
                    rows.append(r)
                    for j,name in enumerate(GROUP_NAMES):
                        group_rows.append(dict(distance=d,p=p,replicate=rep,
                            target=cert['target'],group=name,
                            observed_max_count=int(g['group_max'][j]),
                            group_failure_count=int(g['group_fail'][j]),
                            CP_max_atom_upper=float(g['U'][j]),
                            CP_group_failure_upper=float(g['V'][j])))
                np.savez_compressed(out/f'counts_d{d}_p{p:.2f}_r{rep}.npz',
                    calibration_keys=np.array(ck,dtype=f'S{cal_pack.shape[1]}'),
                    calibration_counts=cc,
                    evaluation_keys=np.array(ek,dtype=f'S{eval_pack.shape[1]}'),
                    evaluation_counts=ec,evaluation_failures=ee)
                print(f'd={d} p={p:.2f} rep={rep}: '
                      f'full point={full_point:.5g}, full UB={rows[-2]["certificate_upper"]:.5g}; '
                      f'excl point={exc_point:.5g}, excl UB={rows[-1]["certificate_upper"]:.5g}; '
                      f'time={time.perf_counter()-start:.1f}s',flush=True)
    df=pd.DataFrame(rows); df.to_csv(out/'results.csv',index=False)
    pd.DataFrame(group_rows).to_csv(out/'groups.csv',index=False)
    print('\nMATCHED CERTIFICATION SUMMARY')
    cols=['distance','p','target','calibration_shots','evaluation_shots',
          'matched_point_estimate','certificate_upper','nonvacuous',
          'author_full_SR_acceptance','excluded_SR_acceptance',
          'denominator_lower','numerator_upper']
    print(df[cols].to_string(index=False))
    print('\nSaved archives and configuration at:',out)
    print('The point estimate is NOT a confidence bound. Every UB is 95% marginal,')
    print('conditional on independent calibration. Different targets are NOT joint 95%.')

if __name__=='__main__': main()
```

```python
"""Phase 8 follow-up. Rigorous direct alpha=2 collision confidence baselines.

Inputs: archived *evaluation* syndrome/failure counts and independently selected
calibration-dominant syndrome in qec_matched_certification_v1.py NPZ files.
NO new Stim sampling. For each dataset we restore a random ordering conditional
on the stored sufficient counts; disjoint evaluation pairs are then independent
unconditionally (not conditional on the observed counts).

Primary: one-sided CP on failure labels conditional on disjoint matching pairs.
Secondary: one-sided Hoeffding bounded U-statistic numerator/denominator bounds,
using every unordered pair with valid population coverage (very conservative).

Note: distinct methods and full/excluded targets are each MARGINAL 95%, not joint.
"""
from __future__ import annotations
import math, json, hashlib
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import beta

CFG = dict(
    archive_dir='original_paper_reproduction/matched_certification_v1',
    out_dir='original_paper_reproduction/direct_collision_baselines_v1',
    alpha=2, delta=0.05, seed=20261114+8901,
    method='one-sided conditional Clopper-Pearson; all-pairs Hoeffding U-statistic',
)


def cp_upper(x,n,epsilon):
    if n==0 or x==n: return 1.0
    return float(beta.isf(epsilon,x+1,n-x))


def atomic_stats(file):
    z=np.load(file,allow_pickle=False)
    ck=np.asarray(z['calibration_keys']); cc=np.asarray(z['calibration_counts'],dtype=np.int64)
    ek=np.asarray(z['evaluation_keys']); ec=np.asarray(z['evaluation_counts'],dtype=np.int64)
    ee=np.asarray(z['evaluation_failures'],dtype=np.int64)
    assert len(ek)==len(ec)==len(ee) and len(ck)==len(cc)
    assert np.all((0<=ee)&(ee<=ec)) and np.all(ec>=1) and np.all(cc>=1)
    dominant_idx=int(np.lexsort((ck, -cc))[0])
    # Historical script uses: min(ck,key=lambda k:(-cal[k],k))
    dom=bytes(ck[dominant_idx])
    return ek,ec,ee,dom


def restore_order(counts, errors, seed):
    """Generate a uniform random permutation of shots for aggregated sufficient data.

    Labels are syndrome IDs. For every label c-e shots are nonfailure and e
    are failure. A uniform random permutation restores the distribution of
    random observation ordering, provided original shots were iid.
    """
    n=int(counts.sum())
    lab=np.repeat(np.arange(len(counts),dtype=np.int32),counts)
    failures=np.zeros(n,dtype=np.uint8)
    start=0
    for c,e in zip(counts,errors):
        failures[start:start+int(e)]=1
        start+=int(c)
    rng=np.random.default_rng(seed)
    order=rng.permutation(n)
    return lab[order],failures[order]


def direct_u_stat(keys, counts, errors, excluded):
    n=int(counts.sum()); assert n>=2
    permitted=(keys!=excluded) if excluded is not None else np.ones(len(counts),dtype=bool)
    c=counts[permitted].astype(np.float64)
    e=errors[permitted].astype(np.float64)
    C=float(np.sum(c*(c-1))/(n*(n-1)))
    F=float(np.sum(e*(c-1))/(n*(n-1)))
    r=F/C if C>0 else math.nan
    # Unbiased *moments*. Their ratio is not exactly unbiased.
    return C,F,r


def trial(keys, counts, errors, excluded, lab, failures, delta=.05):
    n=len(lab)
    allowed=np.ones(len(counts),dtype=bool) if excluded is None else (keys!=excluded)
    pair_n=n//2
    a=lab[:2*pair_n:2]; b=lab[1:2*pair_n:2]
    match=(a==b)&allowed[a]
    m=int(match.sum())
    f=int(failures[:2*pair_n:2][match].sum())
    # Conditional on matching, first-shot failures iid Bernoulli(R_2).
    # The collision count m is random; conditional CP is nonetheless valid.
    cp=cp_upper(f,m,delta)
    pair_point=f/m if m else math.nan

    C,F,full_pair_point=direct_u_stat(keys,counts,errors,excluded)
    k=n//2
    eps=delta/2  # Bonferroni for two separate moments.
    t=math.sqrt(math.log(1/eps)/(2*k))
    C_lower=max(0.0,C-t)
    F_upper=min(1.0,F+t)
    h_bound=min(1.0,F_upper/C_lower) if C_lower>0 else 1.0
    return dict(
        shots=n, independent_pairs=pair_n, matches=m, match_failures=f,
        disjoint_pair_point=pair_point, disjoint_pair_cp95_upper=cp,
        direct_C_hat=C,direct_F_hat=F,direct_allpairs_ratio=full_pair_point,
        ustat_hoeffding_radius=t,ustat_C_lower=C_lower,
        ustat_F_upper=F_upper,ustat_hoeffding95_upper=h_bound,
    )


def tests():
    assert cp_upper(0,0,.05)==1
    assert cp_upper(0,100,.05) >0
    assert abs(cp_upper(0,100,.05)-(1-.05**(1/100)))<1e-12
    keys=np.array([b'a',b'b'],dtype='S1')
    cs=np.array([3,2]); es=np.array([1,0]);
    C,F,r=direct_u_stat(keys,cs,es,None)
    assert abs(C-(3*2+2*1)/(5*4))<1e-12
    assert abs(F-(1*2)/(5*4))<1e-12
    assert abs(r-.25)<1e-12
    _,_,exc=direct_u_stat(keys,cs,es,b'a')
    assert exc==0
    # Check conditional CP empirical coverage in a stress scenario that
    # includes an empty-match possibility, using iid randomized pairs.
    rng=np.random.default_rng(129)
    ps=np.array([.6,.3,.1]); fprob=np.array([.05,.4,.8]);
    true_r=float((ps**2@fprob)/(ps**2).sum())
    miss=0
    for _ in range(500):
        synd=rng.choice(3,size=300,p=ps)
        fail=rng.random(300)<fprob[synd]
        hits=synd[::2]==synd[1::2]
        m=int(hits.sum()); y=int(fail[::2][hits].sum())
        miss+=(cp_upper(y,m,.05)+1e-12 < true_r)
    # Not a statistical proof; rejects obvious errors and severe undercoverage.
    assert miss<45,(miss,true_r)
    print(f'SELF TESTS PASS; 500 synthetic CP replicates, misses={miss}, truth={true_r:.6f}')


def main():
    tests()
    source=Path(CFG['archive_dir'])
    files=sorted(source.glob('counts_d*_p*_r*.npz'))
    if not files:
        raise FileNotFoundError(f'No archived counts found at {source.resolve()}. '
              'Upload your saved matched_certification_v1/ directory, or '
              'run this script in the existing Colab session.')
    out=Path(CFG['out_dir']);out.mkdir(parents=True,exist_ok=True)
    config=dict(CFG);config['archive_files']=[str(x) for x in files]
    config['source_sha256']={str(f):hashlib.sha256(f.read_bytes()).hexdigest() for f in files}
    (out/'config.json').write_text(json.dumps(config,indent=2))
    rows=[]
    for i,file in enumerate(files):
        keys,cs,es,dom=atomic_stats(file)
        # Restore source shot order *once*, reuse for both targets.
        lab,failures=restore_order(cs,es,CFG['seed']+i)
        for excl in (False,True):
            results=trial(keys,cs,es,dom if excl else None,lab,failures,CFG['delta'])
            r=dict(archive=file.name,target='excluded_dominant' if excl else 'full',
                   dominant_hex=dom.hex(),**results)
            rows.append(r)
            print(f"{file.name}, {r['target']}: "
                  f"allpairs_point={r['direct_allpairs_ratio']:.6g}, "
                  f"matches={r['matches']}, failures={r['match_failures']}, "
                  f"pair_CP_UB={r['disjoint_pair_cp95_upper']:.6g}, "
                  f"allpairs_Hoeffding_UB={r['ustat_hoeffding95_upper']:.6g}",flush=True)
    df=pd.DataFrame(rows)
    df.to_csv(out/'direct_collision_comparison.csv',index=False)
    prior=source/'results.csv'
    if prior.exists():
        previous=pd.read_csv(prior)
        from_new=df.copy()
        parsed=from_new.archive.str.extract(r'counts_d(?P<distance>\d+)_p(?P<p>[\d.]+)_r(?P<replicate>\d+)\.npz')
        from_new['distance']=parsed.distance.astype(int)
        from_new['p']=parsed.p.astype(float)
        from_new['replicate']=parsed.replicate.astype(int)
        cols=['distance','p','replicate','target','matched_point_estimate','certificate_upper']
        if set(cols).issubset(previous.columns):
            merged=from_new.merge(previous[cols],on=['distance','p','replicate','target'],validate='one_to_one')
            merged.to_csv(out/'side_by_side.csv',index=False)
            print('\nDIRECT VERSUS GROUPWISE, MATCHED ARCHIVES')
            show=['distance','p','target','matched_point_estimate','certificate_upper',
                  'direct_allpairs_ratio','matches','match_failures',
                  'disjoint_pair_cp95_upper','ustat_hoeffding95_upper']
            print(merged[show].to_string(index=False))
    print('\nSaved:',out.resolve())
    print('CAUTION: each bound is marginal 95%; post-hoc selecting the best bound '
          'requires multiplicity control or an independent selection split.')

if __name__=='__main__':main()
```

```python
"""Frozen matched-model 20-seed replication, full and dominant-excluded alpha=2.

Official Colmenarez et al. code-capacity circuit and PyMatching MWPM.
Methods held fixed: groupwise CP/H&ouml;lder/empirical Bernstein; disjoint-pair CP;
all-pairs U-statistic with conservative Hoeffding concentration. No tuning.

Each replicate is one independent calibration/evaluation pair. Two target estimates
are correlated. The 95% bounds are MARGINAL per method, target, replicate.
The comparison across six configurations is exploratory, not a simultaneous 95% claim.
Disjoint-pair construction restores a uniformly random shot ordering conditional on
sufficient syndrome/failure counts; justified by iid sampling.
"""
from __future__ import annotations
import os, sys, json, hashlib, time, math
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import beta

CONFIG = dict(
    distances=[5,7,9],
    ps=[0.12,0.14],
    calibration_shots=50_000,
    evaluation_shots=100_000,
    replicates=20,
    alpha=2,
    delta=0.05,
    groups=12,
    seed=20261114+100000,
    pair_seed=20261114+8901+100000,
    benchmark_thresholds=[1e-3, 1e-2, 5e-2, .1, .25, .5],
    authors_root='original_paper_reproduction/authors_code',
    out_dir='original_paper_reproduction/matched_replication_frozen_v1',
    protocol_version='frozen_2026-11-14_v1',
)
GROUP_NAMES=[
    'observed_100plus','observed_20to99','observed_5to19',
    'observed_2to4','observed_1',
    'unseen_w0to2','unseen_w3to4','unseen_w5to6',
    'unseen_w7to8','unseen_w9to12','unseen_w13to20','unseen_w21plus',
]
assert len(GROUP_NAMES)==CONFIG['groups']

def assign_group(c, syndrome_key):
    """All groups fixed by calibration. Bit weights refer to detector bits."""
    if c >= 100: return 0
    if c >= 20: return 1
    if c >= 5: return 2
    if c >= 2: return 3
    if c == 1: return 4
    w = int.from_bytes(syndrome_key, 'little').bit_count()
    if w <= 2: return 5
    if w <= 4: return 6
    if w <= 6: return 7
    if w <= 8: return 8
    if w <= 12: return 9
    if w <= 20: return 10
    return 11

def upper_cp(k, N, eps):
    # one-sided Clopper-Pearson upper for fixed binomial marginal;
    # k may be max of observed syndrome counts, which dominates true argmax.
    if k >= N: return 1.0
    return float(beta.isf(eps, int(k)+1, int(N)-int(k)))

def aggregate(packed, failed):
    uniques, inverse, counts = np.unique(
        packed, axis=0, return_inverse=True, return_counts=True
    )
    errors = np.bincount(inverse[failed], minlength=len(counts))
    keys = [bytes(row) for row in uniques]
    return keys, counts.astype(np.int64), errors.astype(np.int64)

def rb_and_acceptance(counts, errors, include_mask, total_shots):
    # Exactly the conditional mean of the authors' alpha=2 empirical sampler.
    # This is not an unbiased estimate of the population ratio.
    eligible = int(counts[(counts >= 2) & include_mask].sum())
    w = counts.astype(np.float64)*(counts.astype(np.float64)-1)
    w[~include_mask] = 0
    denom = float(w.sum())
    numerator = float(np.dot(((counts-1)*include_mask).astype(float), errors))
    point = numerator/denom if denom>0 else math.nan
    return point, eligible/total_shots, eligible, denom

def certificate(cal_dict, eval_keys, eval_counts, eval_errors, N, M, dominant, exclude):
    alpha=CONFIG['alpha']
    delta=CONFIG['delta']
    G=CONFIG['groups']
    eps=delta/(2*G+1)
    group_max=np.zeros(G,dtype=np.int64)
    group_fail=np.zeros(G,dtype=np.int64)
    # q contains all calibration weights, optionally excluding its selected dominant.
    q={k: float(v)/M for k,v in cal_dict.items() if not(exclude and k==dominant)}
    B=sum(v**alpha for v in q.values())
    R=max((v**(alpha-1) for v in q.values()),default=0.0)
    # EB expectation X = q[S]^(alpha-1), with 0 for calibration-unseen.
    sum_x=0.0
    sum_x2=0.0
    for k,c,e in zip(eval_keys,eval_counts,eval_errors):
        if exclude and k==dominant:
            continue
        group=assign_group(cal_dict.get(k,0),k)
        group_max[group]=max(int(group_max[group]),int(c))
        group_fail[group]+=e
    for k,c in zip(eval_keys,eval_counts):
        x = q.get(k,0.0)**(alpha-1)
        sum_x += c*x
        sum_x2 += c*x*x
    mean_x=sum_x/N
    var_x=max(0.0,(sum_x2-N*mean_x**2)/(N-1))
    t=math.log(2/eps)
    radius=math.sqrt(2*var_x*t/N)+7*R*t/(3*(N-1))
    A_lo=max(0.0, mean_x-radius)
    C_lo=(A_lo**alpha/B**(alpha-1)) if A_lo>0 and B>0 else 0.0
    U=np.array([upper_cp(k,N,eps) for k in group_max])
    V=np.array([upper_cp(k,N,eps) for k in group_fail])
    F_up=float(np.sum(U**(alpha-1)*V))
    ratio_up=min(1.0,F_up/C_lo) if C_lo>0 else 1.0
    result=dict(
        target='excluded_dominant' if exclude else 'full',
        selected_dominant_hex=dominant.hex(),
        q_support=len(q),q_max=max(q.values(), default=0.0),
        numerator_upper=F_up,denominator_lower=C_lo,
        A_mean=mean_x,A_lower=A_lo,EB_radius=radius,
        q_power_sum_B=B,group_count=G,events=2*G+1,
        per_event_epsilon=eps, confidence_marginal=1-delta,
        certificate_upper=ratio_up,
        nonvacuous=int(ratio_up<1),
    )
    for thresh in CONFIG['benchmark_thresholds']:
        result['certifies_'+str(thresh)]=int(ratio_up<thresh)
    return result, dict(group_max=group_max,group_fail=group_fail,U=U,V=V)

def cp_upper(x,n,epsilon):
    if n==0 or x==n: return 1.0
    return float(beta.isf(epsilon,x+1,n-x))

def restore_order(counts, errors, seed):
    """Generate a uniform random permutation of shots for aggregated sufficient data.

    Labels are syndrome IDs. For every label c-e shots are nonfailure and e
    are failure. A uniform random permutation restores the distribution of
    random observation ordering, provided original shots were iid.
    """
    n=int(counts.sum())
    lab=np.repeat(np.arange(len(counts),dtype=np.int32),counts)
    failures=np.zeros(n,dtype=np.uint8)
    start=0
    for c,e in zip(counts,errors):
        failures[start:start+int(e)]=1
        start+=int(c)
    rng=np.random.default_rng(seed)
    order=rng.permutation(n)
    return lab[order],failures[order]

def direct_u_stat(keys, counts, errors, excluded):
    n=int(counts.sum()); assert n>=2
    permitted=(keys!=excluded) if excluded is not None else np.ones(len(counts),dtype=bool)
    c=counts[permitted].astype(np.float64)
    e=errors[permitted].astype(np.float64)
    C=float(np.sum(c*(c-1))/(n*(n-1)))
    F=float(np.sum(e*(c-1))/(n*(n-1)))
    r=F/C if C>0 else math.nan
    # Unbiased *moments*. Their ratio is not exactly unbiased.
    return C,F,r

def trial(keys, counts, errors, excluded, lab, failures, delta=.05):
    n=len(lab)
    if excluded is None:
        allowed = np.ones(len(counts), dtype=bool)
    else:
        allowed = np.array([k != excluded for k in keys], dtype=bool)
    pair_n=n//2
    a=lab[:2*pair_n:2]; b=lab[1:2*pair_n:2]
    match=(a==b)&allowed[a]
    m=int(match.sum())
    f=int(failures[:2*pair_n:2][match].sum())
    # Conditional on matching, first-shot failures iid Bernoulli(R_2).
    # The collision count m is random; conditional CP is nonetheless valid.
    cp=cp_upper(f,m,delta)
    pair_point=f/m if m else math.nan

    C,F,full_pair_point=direct_u_stat(keys,counts,errors,excluded)
    k=n//2
    eps=delta/2  # Bonferroni for two separate moments.
    t=math.sqrt(math.log(1/eps)/(2*k))
    C_lower=max(0.0,C-t)
    F_upper=min(1.0,F+t)
    h_bound=min(1.0,F_upper/C_lower) if C_lower>0 else 1.0
    return dict(
        shots=n, independent_pairs=pair_n, matches=m, match_failures=f,
        disjoint_pair_point=pair_point, disjoint_pair_cp95_upper=cp,
        direct_C_hat=C,direct_F_hat=F,direct_allpairs_ratio=full_pair_point,
        ustat_hoeffding_radius=t,ustat_C_lower=C_lower,
        ustat_F_upper=F_upper,ustat_hoeffding95_upper=h_bound,
    )

def tests():
    # Foundational small-probability and method-consistency checks.
    assert cp_upper(0,0,.05)==1
    assert abs(cp_upper(0,100,.05)-(1-.05**(1/100)))<1e-12
    assert assign_group(0,b'\xff')==8
    c=np.array([3,2,1]); e=np.array([1,0,1]);
    point,acc,eligible,den=rb_and_acceptance(c,e,np.ones(3,bool),6)
    assert abs(point-.25)<1e-12 and eligible==5
    k=np.array([b'a',b'b'],dtype='S1')
    C,F,r=direct_u_stat(k,np.array([3,2]),np.array([1,0]),None)
    assert abs(r-.25)<1e-12
    ids, fail=restore_order(np.array([3,2]),np.array([1,0]),999)
    assert len(ids)==5 and fail.sum()==1
    got=trial(k,np.array([3,2]),np.array([1,0]),None,ids,fail)
    assert got['disjoint_pair_cp95_upper']>=0
    print('SELF TESTS PASS')

def frozen_manifest(out, source_bytes, versions):
    metadata={**CONFIG,**versions,
        'official_utils_sha256':hashlib.sha256(source_bytes).hexdigest(),
        'limitations':'marginal 95% per bound; comparison not familywise 95%; 20 replicates exploratory',
        'pairing':'fixed disjoint consecutive pairs after independent random order restoration',
        'group_bound_events':25,
    }
    config_path=out/'frozen_config.json'
    encoded=json.dumps(metadata,sort_keys=True,indent=2)
    if config_path.exists():
        old=json.loads(config_path.read_text())
        if old!=metadata:
            raise RuntimeError('FROZEN CONFIG DIFFERENCE: refusing to mix runs.')
    else:
        config_path.write_text(encoded)
    return metadata

def summarize(df,out):
    # Each replicate provides exactly one row per target, paired across methods.
    records=[]
    for (d,p,t),g in df.groupby(['distance','p','target'],sort=True):
        assert len(g)==CONFIG['replicates'], (d,p,t,len(g))
        vals=dict(distance=int(d),p=float(p),target=t,replicates=len(g),
            median_point=float(g.matched_point_estimate.median()),
            median_group_UB=float(g.certificate_upper.median()),
            median_direct_CP_UB=float(g.disjoint_pair_cp95_upper.median()),
            median_Ustat_Hoeffding_UB=float(g.ustat_hoeffding95_upper.median()),
            group_nonvacuous_frac=float((g.certificate_upper<1).mean()),
            direct_CP_nonvacuous_frac=float((g.disjoint_pair_cp95_upper<1).mean()),
            direct_CP_strictly_better_frac=float((g.disjoint_pair_cp95_upper<g.certificate_upper-1e-12).mean()),
            group_strictly_better_frac=float((g.certificate_upper<g.disjoint_pair_cp95_upper-1e-12).mean()),
            median_pair_matches=float(g.matches.median()),
            mean_pair_matches=float(g.matches.mean()),
            median_full_acceptance=float(g.author_full_SR_acceptance.median()),
            median_denominator_lower=float(g.denominator_lower.median()))
        for theta in CONFIG['benchmark_thresholds']:
            nm=str(theta).replace('.','p')
            vals[f'group_UB_lt_{nm}_frac']=float((g.certificate_upper<theta).mean())
            vals[f'pair_CP_UB_lt_{nm}_frac']=float((g.disjoint_pair_cp95_upper<theta).mean())
        records.append(vals)
    summary=pd.DataFrame(records)
    summary.to_csv(out/'replication_summary.csv',index=False)
    show=['distance','p','target','replicates','median_point',
          'median_group_UB','median_direct_CP_UB','group_nonvacuous_frac',
          'direct_CP_nonvacuous_frac','direct_CP_strictly_better_frac','median_pair_matches']
    print('\nFROZEN MULTI-SEED REPLICATION SUMMARY')
    print(summary[show].to_string(index=False))
    print('\nPer replicate per target: separate marginal 95% procedures. No joint confidence statement.')
    return summary

def main():
    tests()
    try:
        import stim, pymatching
    except ImportError as exc:
        raise RuntimeError('Install first: %pip install stim pymatching scipy pandas') from exc
    root=Path(CONFIG['authors_root']).resolve()
    utils=root/'surface_code_simulation'/'utils_surface_code.py'
    if not utils.is_file():
        raise FileNotFoundError(f'Missing official source: {utils}')
    sys.path.insert(0,str(utils.parent))
    from utils_surface_code import surface_code_capacity_zero
    out=Path(CONFIG['out_dir'])
    out.mkdir(parents=True,exist_ok=True)
    versioninfo=dict(stim_version=stim.__version__,
                     pymatching_version=pymatching.__version__,
                     numpy_version=np.__version__)
    frozen_manifest(out,utils.read_bytes(),versioninfo)
    count_dir=out/'counts';count_dir.mkdir(exist_ok=True)
    rows=[]
    for d in CONFIG['distances']:
        for pidx,p in enumerate(CONFIG['ps']):
            circuit=surface_code_capacity_zero(d,d,p)
            decoder=pymatching.Matching.from_detector_error_model(
                circuit.detector_error_model(decompose_errors=True))
            circuit_hash=hashlib.sha256(str(circuit).encode()).hexdigest()
            for rep in range(CONFIG['replicates']):
                # Unique seeds across all distances, probabilities, repetitions, and roles.
                seed=CONFIG['seed'] + d*100000 + pidx*10000 + rep*10
                t0=time.perf_counter()
                M=CONFIG['calibration_shots']; N=CONFIG['evaluation_shots']
                archive=count_dir/f'counts_d{d}_p{p:.2f}_r{rep:02d}.npz'
                record=out/f'row_d{d}_p{p:.2f}_r{rep:02d}.json'
                if archive.exists() and record.exists():
                    prev=json.loads(record.read_text())
                    assert len(prev)==2
                    rows.extend(prev)
                    print(f'REUSED completed d={d} p={p:.2f} replicate={rep}',flush=True)
                    continue
                if archive.exists() != record.exists():
                    raise RuntimeError(f'Partial archive present ({archive}); inspect before resuming.')
                cd,_=circuit.compile_detector_sampler(seed=seed).sample(M,separate_observables=True)
                ed,obs=circuit.compile_detector_sampler(seed=seed+1).sample(N,separate_observables=True)
                fail=np.any(decoder.decode_batch(ed)!=obs,axis=1)
                cal_pack=np.packbits(cd,axis=1,bitorder='little')
                eval_pack=np.packbits(ed,axis=1,bitorder='little')
                ck,cc,_=aggregate(cal_pack,np.zeros(M,dtype=bool))
                ek,ec,ee=aggregate(eval_pack,fail)
                cal=dict(zip(ck,map(int,cc)))
                dom=min(ck,key=lambda k:(-cal[k],k))
                mask=np.array([k!=dom for k in ek],dtype=bool)
                full_point,full_acc,full_eligible,_=rb_and_acceptance(ec,ee,np.ones(len(ec),bool),N)
                exc_point,exc_acc,exc_eligible,_=rb_and_acceptance(ec,ee,mask,N)
                # A uniform random ordering of original iid observations, restored
                # from archived syndrome/failure counts; both targets reuse it.
                pair_seed=CONFIG['pair_seed']+d*100000+pidx*10000+rep*10
                labels,failures=restore_order(ec,ee,pair_seed)
                this=[]
                for exclude in (False,True):
                    cert,_=certificate(cal,ek,ec,ee,N,M,dom,exclude)
                    direct=trial(ek,ec,ee,dom if exclude else None,
                                 labels,failures,delta=CONFIG['delta'])
                    row={
                        'distance':d,'p':p,'replicate':rep,'target':cert['target'],
                        'cal_seed':seed,'evaluation_seed':seed+1,'pair_seed':pair_seed,
                        'circuit_sha256':circuit_hash,
                        'calibration_shots':M,'evaluation_shots':N,
                        'distinct_cal':len(ck),'distinct_eval':len(ek),
                        'calibration_dominant_count':cal[dom],
                        'baseline_LER':float(fail.mean()),
                        'author_full_SR_acceptance':full_acc,
                        'excluded_SR_acceptance':exc_acc,
                        'matched_point_estimate':exc_point if exclude else full_point,
                        'runtime_seconds':time.perf_counter()-t0,
                        **cert,**direct,
                    }
                    this.append(row)
                # Archive is written before the paired result, so interrupted writes
                # are detected, rather than silently mixed with a different seed.
                np.savez_compressed(archive,
                    calibration_keys=np.array(ck,dtype=f'S{cal_pack.shape[1]}'),
                    calibration_counts=cc,
                    evaluation_keys=np.array(ek,dtype=f'S{eval_pack.shape[1]}'),
                    evaluation_counts=ec,evaluation_failures=ee)
                record.write_text(json.dumps(this,indent=2))
                rows.extend(this)
                print(f'd={d} p={p:.2f} rep={rep+1}/{CONFIG["replicates"]}: '
                      f'G-full={this[0]["certificate_upper"]:.4g} '
                      f'CP-full={this[0]["disjoint_pair_cp95_upper"]:.4g} '
                      f'matches={this[0]["matches"]} '
                      f'elapsed={time.perf_counter()-t0:.1f}s',flush=True)
                # Save incremental results after every successful replicate.
                pd.DataFrame(rows).to_csv(out/'all_replicates_partial.csv',index=False)
    df=pd.DataFrame(rows)
    assert len(df)==2*len(CONFIG['distances'])*len(CONFIG['ps'])*CONFIG['replicates']
    assert not df.duplicated(['distance','p','replicate','target']).any()
    df=df.sort_values(['distance','p','replicate','target']).reset_index(drop=True)
    df.to_csv(out/'all_replicates.csv',index=False)
    summarize(df,out)
    (out/'COMPLETE.txt').write_text('Complete. Output is exploratory method comparison; 95% marginal bounds.\n')
    print('\nSaved:',out.resolve())

if __name__=='__main__':
    main()
```

```python
"""Exact syndrome-resampled failure probabilities for Colmenarez et al. code-capacity circuits.

A frozen next experiment. No new estimator or new theorem is claimed.

Run in a Colab session where the original Zenodo authors' code has already been
extracted to original_paper_reproduction/authors_code:

    %run qec_exact_population_codecapacity_v1.py

Or, for independent mathematics-only tests (no stim/pymatching installation):
    python qec_exact_population_codecapacity_v1.py --self-test

Important: A deterministic syndrome decoder does NOT imply f(s) is 0 or 1.
Track P(syndrome, logical_observable) before applying decoder prediction.
"""
from __future__ import annotations
import argparse
import hashlib
import itertools
import json
from pathlib import Path
import sys

import numpy as np

FROZEN = {
    'distances': [3, 5],
    'physical_noise_p': [0.12, 0.14],
    'confidence_delta': 0.05,
    'direct_pair_shot_budgets': [100000, 150000],
    'direct_pair_replicates': 5000,
    'simulation_seed': 20261114 + 47000,
    'out_dir': 'original_paper_reproduction/exact_population_v1',
    'authors_dir': 'original_paper_reproduction/authors_code',
    'circuit_method': 'surface_code_capacity_zero from original Zenodo archive',
    'population_algorithm': 'exact convolution over independent DEM error events without decomposition',
}


def joint_xor_distribution(num_detectors: int, events):
    """Exact joint P(S, O) from independent binary flips (p, syndrome+observable XOR mask).

    Index layout: lower d bits = detector syndrome; bit d = observable flip.
    This assumes each DEM error instruction represents an independent Bernoulli mechanism.
    """
    if not 0 <= num_detectors <= 18:
        raise ValueError('Restrict to <=18 detector bits for bounded exact array size.')
    n = 1 << (num_detectors + 1)
    dist = np.zeros(n, dtype=np.float64)
    dist[0] = 1.0
    states = np.arange(n, dtype=np.int64)
    for p,mask in events:
        p = float(p)
        mask = int(mask)
        if not (0 <= p <= 1 and 0 <= mask < n):
            raise ValueError('Invalid error mechanism')
        dist = (1-p)*dist + p*dist[states ^ mask]
    if not np.isclose(dist.sum(), 1, atol=1e-12, rtol=0):
        raise AssertionError('Probability distribution not normalized')
    if dist.min() < -1e-14:
        raise AssertionError('Negative mass')
    return dist


def dem_independent_errors(circuit):
    # Stim's non-decomposed DEM represents the circuit's independent error
    # mechanisms. No approximation or splitting of correlated faults is allowed.
    dem = circuit.detector_error_model(decompose_errors=False).flattened()
    nd = int(dem.num_detectors)
    if int(dem.num_observables) != 1:
        raise ValueError('Expected exactly one logical observable')
    events = []
    for ins in dem:
        if ins.type != 'error':
            continue
        args = ins.args_copy()
        if len(args) != 1:
            raise AssertionError(f'Unexpected DEM instruction {ins}')
        mask = 0
        for t in ins.targets_copy():
            if t.is_separator():
                raise ValueError('Unresolved correlated DEM separation; do not approximate it')
            if t.is_relative_detector_id():
                mask ^= 1 << int(t.val)
            elif t.is_logical_observable_id():
                if int(t.val) != 0:
                    raise ValueError('More than one logical observable')
                mask ^= 1 << nd
            else:
                raise ValueError(f'Unknown DEM target {t}')
        events.append((float(args[0]), mask))
    if not events:
        raise ValueError('No error mechanisms found')
    return nd, events, dem


def exact_targets(joint, n_detectors, predictions):
    count = 1 << n_detectors
    p0 = joint[:count]
    p1 = joint[count:]
    p_s = p0 + p1
    prediction = np.asarray(predictions).reshape(-1).astype(np.uint8)
    if len(prediction) != count or not np.all(np.isin(prediction, [0,1])):
        raise ValueError('Decoder did not return one binary observable prediction per syndrome')
    # If decoder predicts 0: failure iff actual observable flip=1; vice versa.
    e_s = np.where(prediction == 0, p1, p0)
    assert np.all(e_s <= p_s + 1e-15)
    results = {
        'baseline_LER_exact': float(e_s.sum()),
        'P_zero_syndrome_exact': float(p_s[0]),
        'dominant_syndrome_integer': int(np.argmax(p_s)),
        'P_dominant_syndrome_exact': float(p_s.max()),
        'positive_probability_syndromes': int(np.count_nonzero(p_s)),
        'mixed_failure_probability_syndromes': int(np.count_nonzero((e_s>1e-14)&((p_s-e_s)>1e-14))),
    }
    for a in (2, 3):
        weights = p_s**a
        numer = float(np.dot(p_s**(a-1),e_s))
        denom = float(weights.sum())
        results[f'C{a}_exact'] = denom
        results[f'F{a}_exact'] = numer
        results[f'R{a}_exact'] = numer/denom if denom else float('nan')
        # This is the POPULATION most-probable syndrome, NOT the random
        # syndrome s0 selected by a finite calibration sample.
        dom = int(np.argmax(p_s))
        ex_denom = denom-weights[dom]
        ex_numer = numer-(p_s[dom]**(a-1))*e_s[dom]
        results[f'R{a}_excluded_population_dominant_exact'] = (
            ex_numer/ex_denom if ex_denom>0 else float('nan'))
    return results,p_s,e_s,prediction


def cp_upper_array(failures, matches, delta):
    from scipy.stats import beta
    k=np.asarray(matches, dtype=np.int64)
    t=np.asarray(failures, dtype=np.int64)
    ub = np.ones(len(k), dtype=np.float64)
    valid = (k > 0) & (t < k)
    ub[valid] = beta.isf(delta, t[valid]+1, k[valid]-t[valid])
    return ub


def analytic_pair_cp_exact(c2, r2, pair_shots, replicates, seed, delta=0.05):
    """Exact sampling law for disjoint matching pairs, no Stim simulation needed."""
    rng = np.random.default_rng(seed)
    m = pair_shots//2
    k = rng.binomial(m, c2, size=replicates)
    failures = rng.binomial(k, r2)
    upper = cp_upper_array(failures,k,delta)
    return {
        'shots':pair_shots,'independent_pairs':m,
        'E_matching_pairs_exact':float(m*c2),
        'P_zero_matches_exact':float((1-c2)**m),
        'median_matching_pairs':float(np.median(k)),
        'median_CP_upper':float(np.median(upper)),
        'CP_nonvacuous_rate':float(np.mean(upper<1)),
        'empirical_coverage':float(np.mean(upper+1e-14>=r2)),
        'coverage_misses':int(np.sum(upper+1e-14<r2)),
        'replicates':int(replicates),
    }


def self_tests():
    # Verify the XOR convolution by direct exhaustive enumeration of all 2^3
    # combinations of three fault mechanisms, including an observable flip.
    events = [(0.2, 0b001), (0.3, 0b110), (0.15, 0b101)]
    joint = joint_xor_distribution(2,events)
    brute = np.zeros(8)
    for bits in itertools.product([0,1],repeat=len(events)):
        index=0;prob=1.0
        for (p,mask),bit in zip(events,bits):
            prob *= p if bit else 1-p
            if bit:index ^= mask
        brute[index] += prob
    assert np.allclose(joint,brute,rtol=0,atol=1e-15)
    # Exact target with deterministic prediction but nonbinary f(s).
    res,ps,es,pred=exact_targets(joint,2,[0,1,0,1])
    assert res['mixed_failure_probability_syndromes'] > 0
    assert np.isclose(res['R2_exact'],np.sum(ps*es)/np.sum(ps*ps))
    assert np.isclose(res['R3_exact'],np.sum(ps**2*es)/np.sum(ps**3))
    # CP direct-pair identity, including missing matches and zero failures.
    assert cp_upper_array([0,0,1],[0,10,10],.05)[0]==1
    assert np.isclose(cp_upper_array([0],[10],.05)[0],1-.05**(.1))
    print('MATHEMATICAL SELF TESTS PASS: exhaustive 2^3 faults, XOR convolution, exact targets, CP')


def main():
    self_tests()
    import pandas as pd
    import pymatching
    import stim
    root=Path(FROZEN['authors_dir']).resolve()
    helper=root/'surface_code_simulation'/'utils_surface_code.py'
    if not helper.exists():
        raise FileNotFoundError(f'Original authors utility module not found: {helper}')
    sys.path.insert(0,str(helper.parent))
    from utils_surface_code import surface_code_capacity_zero
    out=Path(FROZEN['out_dir'])
    out.mkdir(parents=True,exist_ok=True)
    provenance={**FROZEN,'stim_version':stim.__version__,
                'pymatching_version':getattr(pymatching,'__version__','unknown'),
                'authors_helper_SHA256':hashlib.sha256(helper.read_bytes()).hexdigest()}
    (out/'frozen_config.json').write_text(json.dumps(provenance,indent=2))
    rows=[];pair_rows=[]
    for d in FROZEN['distances']:
        for i,p in enumerate(FROZEN['physical_noise_p']):
            circuit=surface_code_capacity_zero(d,d,p)
            nd,events,dem=dem_independent_errors(circuit)
            if nd != (d*d-1)//2:
                raise AssertionError(f'Unexpected detector count {nd} for d={d}')
            joint=joint_xor_distribution(nd,events)
            decoder=pymatching.Matching.from_detector_error_model(
                circuit.detector_error_model(decompose_errors=True))
            state=np.arange(1<<nd,dtype=np.uint32)
            # Detector axis order matches the detector record's bit columns.
            synd=((state[:,None] >> np.arange(nd,dtype=np.uint32)) & 1).astype(np.uint8)
            pred=decoder.decode_batch(synd)
            metrics,ps,es,pr=exact_targets(joint,nd,pred)
            row={'distance':d,'physical_p':p,'num_detectors':nd,
                 'num_independent_DEM_mechanisms':len(events), **metrics}
            rows.append(row)
            np.savez_compressed(out/f'exact_distribution_d{d}_p{p:.2f}.npz',
                                p_s=ps,e_s=es,predicted_obs=pr,
                                P_observable_zero=joint[:len(ps)],
                                P_observable_one=joint[len(ps):])
            for n in FROZEN['direct_pair_shot_budgets']:
                stats=analytic_pair_cp_exact(metrics['C2_exact'],metrics['R2_exact'],
                      n,FROZEN['direct_pair_replicates'],
                      FROZEN['simulation_seed']+d*1000+i*100+n,
                      FROZEN['confidence_delta'])
                pair_rows.append({'distance':d,'physical_p':p,**stats})
            print(f"d={d} p={p:.2f}: baseline={row['baseline_LER_exact']:.6g} "
                  f"C2={row['C2_exact']:.6g} R2={row['R2_exact']:.6g} "
                  f"R3={row['R3_exact']:.6g} "
                  f"mixed_failure_syndromes={row['mixed_failure_probability_syndromes']}",flush=True)
    exact=pd.DataFrame(rows)
    pair=pd.DataFrame(pair_rows)
    exact.to_csv(out/'exact_truth.csv',index=False)
    pair.to_csv(out/'analytic_CP_coverage_100k_150k.csv',index=False)
    print('\nEXACT POPULATION GROUND TRUTH')
    print(exact[['distance','physical_p','baseline_LER_exact','C2_exact','R2_exact',
                 'R3_exact','P_zero_syndrome_exact','mixed_failure_probability_syndromes']].to_string(index=False))
    print('\nMATCHED-BUDGET DIRECT-PAIR CP: ANALYTIC POPULATION LAW')
    print(pair[['distance','physical_p','shots','E_matching_pairs_exact',
                'P_zero_matches_exact','median_CP_upper','CP_nonvacuous_rate',
                'empirical_coverage','coverage_misses']].to_string(index=False))
    print('\nSAVED:',out.resolve())
    print('This tests direct-pair CP using exact population truth, not groupwise coverage.')
    print('Any excluded-calibration target must be evaluated conditional on an independently selected s0.')

if __name__=='__main__':
    parser=argparse.ArgumentParser()
    parser.add_argument('--self-test',action='store_true')
    a, unknown=parser.parse_known_args()
    if a.self_test:self_tests()
    else:main()
```

```python
"""Phase 11 postprocessing: exact-truth audit of frozen d=5 alpha=2 certificates.

Requires existing Colab outputs:
  original_paper_reproduction/exact_population_v1/exact_distribution_d5_p*.npz
  original_paper_reproduction/matched_replication_frozen_v1/all_replicates.csv
  original_paper_reproduction/matched_replication_frozen_v1/counts/counts_d5_p*_r*.npz

Does NOT run Stim or modify existing simulations. Full and independently
calibration-selected dominant-excluded targets are evaluated separately.
The nominal bounds are individual 95% bounds, not joint confidence statements.
"""
from __future__ import annotations
import argparse
from pathlib import Path
import numpy as np
import pandas as pd

BASE=Path('original_paper_reproduction')
OUT=BASE/'exact_truth_coverage_audit_v1'


def exact_target(p_s,e_s,s0=None):
    p_s=np.asarray(p_s,dtype=float)
    e_s=np.asarray(e_s,dtype=float)
    assert p_s.ndim==1 and p_s.shape==e_s.shape
    assert abs(float(p_s.sum())-1)<1e-9
    assert np.all(e_s>=-1e-13) and np.all(e_s<=p_s+1e-13)
    c=float(np.dot(p_s,p_s))
    f=float(np.dot(p_s,e_s))
    if s0 is not None:
        assert 0<=s0<len(p_s)
        c-=float(p_s[s0]**2)
        f-=float(p_s[s0]*e_s[s0])
    if c<=0:
        raise ValueError('Target undefined: zero collision mass')
    return f/c,c


def selected_dominant(archive):
    with np.load(archive,allow_pickle=False) as z:
        ck=z['calibration_keys']; cc=z['calibration_counts']
        if len(ck)==0: raise ValueError('Empty calibration archive')
        nbytes=int(ck.dtype.itemsize)
        # Consistent with frozen script: ties resolved by lexicographic bytes.
        index=min(range(len(ck)),key=lambda i:(-int(cc[i]),bytes(ck[i])))
        label=bytes(ck[index])
    return int.from_bytes(label,byteorder='little'),label.hex(),int(cc[index]),nbytes


def analyse(base=BASE):
    frozen=base/'matched_replication_frozen_v1'
    truth_dir=base/'exact_population_v1'
    rep_file=frozen/'all_replicates.csv'
    if not rep_file.exists():
        raise FileNotFoundError(f'Missing {rep_file}; use the same Colab session as Phase 9')
    df=pd.read_csv(rep_file)
    needed=['distance','p','replicate','target','certificate_upper',
            'disjoint_pair_cp95_upper','matched_point_estimate','matches']
    missing=set(needed)-set(df.columns)
    if missing: raise ValueError(f'Missing columns from frozen replication: {missing}')
    assert not df.duplicated(['distance','p','replicate','target']).any()
    rows=[];truth_cache={}
    for r in df.itertuples(index=False):
        d=int(r.distance); p=float(r.p); rep=int(r.replicate)
        if d!=5: continue  # Exact population for d=7 and 9 not established.
        condition=(d,p)
        if condition not in truth_cache:
            file=truth_dir/f'exact_distribution_d{d}_p{p:.2f}.npz'
            if not file.exists(): raise FileNotFoundError(f'Missing exact distribution {file}')
            with np.load(file,allow_pickle=False) as z:
                ps=z['p_s'].astype(float)
                es=z['e_s'].astype(float)
            truth_cache[condition]=(ps,es)
        ps,es=truth_cache[condition]
        arch=frozen/'counts'/f'counts_d{d}_p{p:.2f}_r{rep:02d}.npz'
        if not arch.exists(): raise FileNotFoundError(f'Missing matched counts {arch}')
        dom,hex_dom,cal_count,bytes_per_syndrome=selected_dominant(arch)
        if dom>=len(ps):
            raise AssertionError('Endianness or detector count mismatch')
        is_excl=str(r.target)=='excluded_dominant'
        if str(r.target) not in ('full','excluded_dominant'):
            raise AssertionError(f'Unexpected target {r.target}')
        true_r,c2=exact_target(ps,es,dom if is_excl else None)
        group_ub=float(r.certificate_upper)
        pair_ub=float(r.disjoint_pair_cp95_upper)
        if not (0<=group_ub<=1 and 0<=pair_ub<=1):
            raise AssertionError('Invalid upper bound')
        original_dom=int(np.argmax(ps))
        # The population-dominant syndrome is not selected by calibration
        # in every distribution or every replicate. Truth follows selected s0.
        rows.append(dict(
            distance=d,p=p,replicate=rep,target=r.target,
            true_R2=true_r,true_C2=c2,
            group_UB=group_ub,pair_CP_UB=pair_ub,
            group_covers=(group_ub+1e-13 >= true_r),
            pair_covers=(pair_ub+1e-13 >= true_r),
            group_excess=group_ub-true_r,
            pair_excess=pair_ub-true_r,
            point=float(r.matched_point_estimate),
            point_error=float(r.matched_point_estimate)-true_r,
            matches=int(r.matches),
            expected_matches_100k=50000*c2,
            selected_dominant_integer=dom,
            population_dominant_integer=original_dom,
            calibration_dominant_matches_population=(dom==original_dom),
            selected_dominant_calibration_count=cal_count,
            selected_dominant_hex=hex_dom,
            bytes_per_syndrome=bytes_per_syndrome,
        ))
    if not rows: raise ValueError('No d=5 rows found')
    result=pd.DataFrame(rows)
    summary=[]
    for (d,p,target),g in result.groupby(['distance','p','target'],sort=True):
        summary.append(dict(
            distance=d,p=p,target=target,replicates=len(g),
            exact_R2_median=float(g.true_R2.median()),
            true_R2_min=float(g.true_R2.min()),
            true_R2_max=float(g.true_R2.max()),
            median_empirical_point=float(g.point.median()),
            median_abs_point_error=float(g.point_error.abs().median()),
            mean_empirical_point_error=float(g.point_error.mean()),
            group_coverage_count=int(g.group_covers.sum()),
            pair_CP_coverage_count=int(g.pair_covers.sum()),
            group_coverage_fraction=float(g.group_covers.mean()),
            pair_CP_coverage_fraction=float(g.pair_covers.mean()),
            median_group_UB=float(g.group_UB.median()),
            median_pair_CP_UB=float(g.pair_CP_UB.median()),
            median_pair_matches=float(g.matches.median()),
            mean_pair_matches=float(g.matches.mean()),
            mean_expected_pair_matches=float(g.expected_matches_100k.mean()),
            calibration_dominant_matches_population_fraction=float(
                g.calibration_dominant_matches_population.mean()),
        ))
    summary=pd.DataFrame(summary)
    out=base/'exact_truth_coverage_audit_v1'
    out.mkdir(exist_ok=True,parents=True)
    result.to_csv(out/'audit_per_replicate.csv',index=False)
    summary.to_csv(out/'audit_summary.csv',index=False)
    show=['distance','p','target','replicates','exact_R2_median',
          'median_empirical_point','group_coverage_count','pair_CP_coverage_count',
          'median_group_UB','median_pair_CP_UB','median_pair_matches',
          'mean_expected_pair_matches']
    print('\nEXACT-TRUTH COVERAGE AUDIT, FROZEN 20-SEED d=5 RESULTS')
    print(summary[show].to_string(index=False))
    print('\nSaved:',out)
    print('Truth for excluded_dominant computed separately for each calibration-selected syndrome.')
    print('At only 20 seeds per condition, empirical coverage estimates are very imprecise.')
    print('No new Stim simulations performed; the earlier frozen analyses were not retuned.')
    return summary


def self_test():
    p=np.array([.6,.3,.1])
    e=np.array([.06,.12,.08])
    all_target,c2=exact_target(p,e)
    expected=(.6*.06+.3*.12+.1*.08)/(.6**2+.3**2+.1**2)
    assert np.isclose(all_target,expected)
    excluded,c2_ex=exact_target(p,e,0)
    assert np.isclose(excluded,(.3*.12+.1*.08)/(.3**2+.1**2))
    assert c2_ex < c2
    print('SELF TESTS PASS: exact full/excluded targets and collision mass')

if __name__=='__main__':
    a=argparse.ArgumentParser()
    a.add_argument('--self-test',action='store_true')
    args, unknown = a.parse_known_args()
    self_test()
    if not args.self_test:
        analyse()
```

```python
"""FROZEN diagnostic / matched crossover for alpha=3 direct disjoint triples.

No Stim simulation. Reads archived independent calibration and evaluation syndrome
counts and shot-level failure tallies, reconstructs uniformly random shot order
conditional on the sufficient statistics, then groups shots into disjoint triples.
The exact Clopper-Pearson upper bound targets the alpha=3 syndrome-powered
failure probability, with the SAME dominant-excluded definition as calibration.

For a matched comparison, use the archived 1M calibration and independent 2M
prospective evaluation files, and the ORIGINAL saved groupwise certificate.
A bundled 100K count file may be used for local diagnostics but cannot replace
those 2M prospective results.

Run in Colab:
  %run qec_alpha3_crossover_archive_v1.py
or CLI:
  python qec_alpha3_crossover_archive_v1.py \\
    --calibration-archive qec_d5_alpha2_validation/counts.pkl.gz \\
    --evaluation-archive qec_d5_ablated_prospective_v1/evaluation.pkl.gz \\
    --group-results qec_d5_ablated_prospective_v1/results.csv
"""
from __future__ import annotations
import argparse, csv, gzip, json, math, pickle
from pathlib import Path
from collections import Counter
import numpy as np
from scipy.stats import beta

DEFAULT_CAL=Path('qec_d5_alpha2_validation/counts.pkl.gz')
DEFAULT_EVAL=Path('qec_d5_ablated_prospective_v1/evaluation.pkl.gz')
LOCAL_ARCHIVE=Path('/mnt/data/distance_5_counts.pkl.gz')
SEED=20261114+250700
ALPHA=3
DELTA=0.05

def load_obj(path):
    p=Path(path)
    with (gzip.open(p,'rb') if p.suffix=='.gz' else p.open('rb')) as f:
        return pickle.load(f)

def get_counter(obj, candidates, kind):
    if isinstance(obj,Counter):
        return obj
    if not isinstance(obj,dict):
        raise ValueError(f'{kind}: unsupported archive object type {type(obj)}')
    for name in candidates:
        v=obj.get(name)
        if isinstance(v,(dict,Counter)) and all(isinstance(k,bytes) for k in v):
            return Counter(v)
    items=[(k,type(v).__name__) for k,v in obj.items()]
    raise ValueError(f'{kind}: cannot locate syndrome Counter. Archive keys/types: {items}')

def select_s0(cal):
    if not cal: raise ValueError('Empty calibration archive')
    # Fixed, deterministic calibration tie breaking, not evaluation-driven.
    return min(cal,key=lambda s:(-cal[s],s))

def upper_cp(t,k,delta):
    if k==0 or t==k: return 1.0
    return float(beta.isf(delta,t+1,k-t))

def pair_archive_counts(counts,errors,exclude_s0,seed):
    keys=sorted(counts)
    c=np.array([int(counts[k]) for k in keys],dtype=np.int64)
    e=np.array([int(errors.get(k,0)) for k in keys],dtype=np.int64)
    if np.any(e<0) or np.any(e>c):
        raise ValueError('Found invalid failure tally')
    N=int(c.sum())
    if N<ALPHA: raise ValueError('Not enough evaluation shots')
    m=N//ALPHA
    # All values from raw counts. Form unbiased power-sum estimates.
    c_float=c.astype(float)
    triple=(c_float*(c_float-1)*(c_float-2)).clip(min=0)
    total=(float(N)*(N-1)*(N-2))
    allowed=np.ones(len(keys),dtype=bool)
    if exclude_s0 is not None:
        allowed=np.array([k!=exclude_s0 for k in keys],dtype=bool)
    c3hat=float(triple[allowed].sum()/total)
    f3hat=float(((c_float[allowed]-1)*(c_float[allowed]-2)*e[allowed]).sum()/total)
    f3hat=max(0.,f3hat)
    rhat=f3hat/c3hat if c3hat>0 else math.nan
    # EXACT conditional expectation over uniform random permutations of
    # observed labeled shots, NOT the unknown population expectation.
    expected_matches_conditional=float(m*c3hat)
    # Markov gives P(K>=1 | archived data) <= E[K | data].
    p_any_upper=min(1.,expected_matches_conditional)

    # Restore a uniform shot order conditional on syndrome/failure tallies.
    lab=np.repeat(np.arange(len(keys),dtype=np.int32),c)
    flg=np.zeros(N,dtype=np.uint8)
    start=0
    for ci,ei in zip(c,e):
        flg[start:start+ei]=1
        start+=ci
    assert start==N and int(flg.sum())==int(e.sum())
    rng=np.random.default_rng(seed)
    permutation=rng.permutation(N)
    lab=lab[permutation[:ALPHA*m]].reshape(m,ALPHA)
    first_failure=flg[permutation[:ALPHA*m:ALPHA]]
    match=(lab[:,0]==lab[:,1]) & (lab[:,0]==lab[:,2]) & allowed[lab[:,0]]
    k=int(match.sum())
    t=int(first_failure[match].sum())
    return dict(shots=N,independent_triples=m,matching_triples=k,
                matched_triple_failures=t,cp_95_upper=upper_cp(t,k,DELTA),
                moment_C3_hat=c3hat,moment_F3_hat=f3hat,
                all_triples_ratio=rhat,
                expected_disjoint_matching_triples_conditional=expected_matches_conditional,
                markov_upper_on_at_least_one_match=p_any_upper)

def parse_recorded_group(path):
    if path is None: return None
    p=Path(path)
    if not p.is_file(): raise FileNotFoundError(p)
    with p.open(newline='') as f: rows=list(csv.DictReader(f))
    if not rows: raise ValueError('Empty group results CSV')
    # Do NOT select best after looking at results: fixed target alpha=3 excluded.
    for row in rows:
        astr=str(row.get('alpha',row.get('order','3')))
        target=str(row.get('target',row.get('selection','excluded_dominant'))).lower()
        if astr.replace('.0','')!='3':continue
        if ('full' in target and 'excl' not in target and 'ablated' not in target):continue
        for name in ['ratio_upper','certificate_upper','upper_bound','bound','upper']:
            if name in row and row[name] not in ('','nan','NaN'):
                return {'source':str(p),'field':name,'value':float(row[name])}
    raise ValueError('Could not identify the pre-existing alpha=3 excluded groupwise upper bound. Inspect results.csv and provide --group-upper instead; do not infer it.')

def self_tests():
    assert np.isclose(upper_cp(0,10,.05),1-.05**.1)
    assert upper_cp(0,0,.05)==1
    C=Counter({b'a':4,b'b':3}); E=Counter({b'a':1,b'b':1})
    result=pair_archive_counts(C,E,b'a',123)
    assert result['shots']==7 and result['independent_triples']==2
    assert result['moment_C3_hat']>0
    assert 0<=result['cp_95_upper']<=1
    assert select_s0(C)==b'a'
    print('SELF TESTS PASS: CP, alpha=3 falling-factorial moments, exclusion, shot reconstruction')

def main():
    a=argparse.ArgumentParser()
    a.add_argument('--calibration-archive',type=Path)
    a.add_argument('--evaluation-archive',type=Path)
    a.add_argument('--group-results',type=Path,default=None)
    a.add_argument('--group-upper',type=float,default=None,
                   help='Previously frozen alpha=3 excluded groupwise upper bound, NOT reoptimized')
    a.add_argument('--seed',type=int,default=SEED)
    a.add_argument('--out-dir',type=Path,default=Path('qec_alpha3_crossover_v1'))
    args, unknown = a.parse_known_args()
    self_tests()
    if args.calibration_archive or args.evaluation_archive:
        if not args.calibration_archive or not args.evaluation_archive:
            raise ValueError('Provide both archive paths')
        calpath,evalpath=args.calibration_archive,args.evaluation_archive
    elif DEFAULT_CAL.is_file() and DEFAULT_EVAL.is_file():
        calpath,evalpath=DEFAULT_CAL,DEFAULT_EVAL
    elif LOCAL_ARCHIVE.is_file():
        calpath,evalpath=LOCAL_ARCHIVE,LOCAL_ARCHIVE
        print('DIAGNOSTIC ONLY: using local 100K archive, not 2M matched prospective study')
    else:
        raise FileNotFoundError('Cannot find historical archives. Supply both explicit paths.')
    caldata=load_obj(calpath)
    evaldata=load_obj(evalpath)
    cal=get_counter(caldata,('calibration','counts','syndrome_counts','evaluation'), 'calibration')
    eva=get_counter(evaldata,('evaluation','counts','syndrome_counts'), 'evaluation')
    errors=get_counter(evaldata,('failures','error_counts','failure_counts','failed'), 'evaluation failures')
    if any(v>eva.get(k,0) for k,v in errors.items()):raise ValueError('Failure counts exceed evaluation counts')
    if isinstance(evaldata,dict) and 'total_failures' in evaldata:
        if int(sum(errors.values()))!=int(evaldata['total_failures']):raise ValueError('Failure total mismatch')
    s0=select_s0(cal)
    output=[]
    for target,excluded in [('full',None),('excluded_dominant',s0)]:
        output.append({'target':target,**pair_archive_counts(eva,errors,excluded,args.seed)})
    total_cal=sum(cal.values()); total_eval=sum(eva.values())
    matched=(total_cal==1_000_000 and total_eval==2_000_000 and
             calpath.resolve()!=evalpath.resolve())
    print('\nALPHA=3 CROSSOVER, DISJOINT MATCHING TRIPLES')
    print(f'calibration={total_cal:,}, evaluation={total_eval:,}, selected s0={s0.hex()}')
    print('MATCHED PROSPECTIVE ARCHIVES:' , 'YES' if matched else 'NO, diagnostic only')
    for r in output:
        print(r['target'], 'triples=',r['independent_triples'],
              'matches=',r['matching_triples'], 'failures=',r['matched_triple_failures'],
              'CP_UB=',f"{r['cp_95_upper']:.9g}",
              'E_K_conditional=',f"{r['expected_disjoint_matching_triples_conditional']:.9g}",
              'moment_C3=',f"{r['moment_C3_hat']:.9g}")
    reported=None
    if args.group_results is not None: reported=parse_recorded_group(args.group_results)
    elif args.group_upper is not None: reported={'source':'user-supplied frozen result','value':args.group_upper}
    if reported is not None:
        if not matched:
            print('WARNING: groupwise result cannot be called MATCHED with these archives')
        print('PREVIOUSLY RECORDED EXCLUDED GROUPWISE UB:',reported)
        print('DIRECT TRIPLE CP UB (EXCLUDED):',output[1]['cp_95_upper'])
        if matched:
            print('CROSSOVER COMPARISON (smaller is tighter):',
                  'GROUPWISE' if reported['value']<output[1]['cp_95_upper'] else 'DIRECT TRIPLE CP')
    args.out_dir.mkdir(parents=True,exist_ok=True)
    with (args.out_dir/'results.csv').open('w',newline='') as f:
        writer=csv.DictWriter(f,fieldnames=list(output[0]))
        writer.writeheader();writer.writerows(output)
    manifest=dict(alpha=3,confidence_per_target=.95,seed=args.seed,
                  calibration_file=str(calpath),evaluation_file=str(evalpath),
                  calibration_shots=total_cal,evaluation_shots=total_eval,
                  matched_prospective=matched,selected_dominant_hex=s0.hex(),
                  groupwise_reference=reported,
                  caveats=['95% marginal bounds, two targets not simultaneously 95%',
                           'conditional expected matches computed from observed counts',
                           'random partition of empirical joint syndrome/failure multiset',
                           'not a new independent QEC simulation'])
    (args.out_dir/'manifest.json').write_text(json.dumps(manifest,indent=2))
    print('SAVED:',args.out_dir)

if __name__=='__main__':main()
```
